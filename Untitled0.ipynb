```python
# ===== Paste this entire cell into Google Colab and run =====
# Title: Gold (GC=F) — success rate of price reaching the closer of PDH/PDL to day open
# Installs, downloads data, computes results, plots and saves CSV.

!pip install --quiet yfinance pandas matplotlib

import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime, timedelta
import os

# -------------------- User parameters --------------------
TICKER = "GC=F"          # Yahoo ticker for Gold futures continuous (change if needed)
LOOKBACK_YEARS = 5       # how many years of daily data to fetch
END_DATE = None          # None for today; or "YYYY-MM-DD" to fix end date
MIN_PREV_RANGE = 0       # optional filter: minimum previous-day range (in USD) to include day
OUTPUT_CSV = "/mnt/data/gold_pdh_pdl_target_hits.csv"
# ---------------------------------------------------------

# Prepare dates for yfinance
period_days = int(LOOKBACK_YEARS * 365.25)
start_date = (datetime.utcnow() - timedelta(days=period_days)).strftime("%Y-%m-%d")
end_date = END_DATE if END_DATE else datetime.utcnow().strftime("%Y-%m-%d")

print(f"Downloading {TICKER} daily data from {start_date} to {end_date} ...")
df = yf.download(TICKER, start=start_date, end=end_date, interval="1d", progress=True)

if df.empty:
    raise SystemExit("No data downloaded — check ticker or date range.")

# Keep relevant columns & fix index
df = df[['Open','High','Low','Close','Volume']].copy()
df.index = pd.to_datetime(df.index).tz_localize(None)

# Build analysis frame
a = df.copy()
a['PDH'] = a['High'].shift(1)
a['PDL'] = a['Low'].shift(1)
a['Open_today'] = a['Open']
a = a.dropna(subset=['PDH','PDL'])

# Distances to previous high/low
a['dist_to_PDH'] = (a['PDH'] - a['Open_today']).abs()
a['dist_to_PDL'] = (a['Open_today'] - a['PDL']).abs()

# Choose nearer target (ties -> PDH)
a['target_side'] = np.where(a['dist_to_PDH'] <= a['dist_to_PDL'], 'PDH', 'PDL')
a['target_price'] = np.where(a['target_side']=='PDH', a['PDH'], a['PDL'])

# Check whether the chosen target was hit during the day
a['target_hit'] = np.where(
    a['target_side']=='PDH',
    a['High'] >= a['PDH'],
    a['Low'] <= a['PDL']
)

# Prev day range filter if requested
a['prev_range'] = a['PDH'] - a['PDL']
if MIN_PREV_RANGE > 0:
    a = a[a['prev_range'] >= MIN_PREV_RANGE]

# Summary statistics
total_signals = len(a)
hits = int(a['target_hit'].sum())
success_rate = hits / total_signals if total_signals>0 else float('nan')

by_side = a.groupby('target_side')['target_hit'].agg(['count','sum'])
by_side['success_rate'] = by_side['sum'] / by_side['count']

print("\n=== Summary ===")
print(f"Total days analyzed: {total_signals}")
print(f"Targets hit: {hits}")
print(f"Overall success rate: {success_rate:.2%}\n")
print("By target side:")
print(by_side)

# Add extras for analysis and save CSV
a['day_of_week'] = a.index.day_name()
a['month'] = a.index.month
a['hit_int'] = a['target_hit'].astype(int)
a['cum_hits'] = a['hit_int'].cumsum()
a['cum_signals'] = np.arange(1, len(a)+1)
a['cum_success_rate'] = a['cum_hits'] / a['cum_signals']
a['rolling_30'] = a['hit_int'].rolling(window=30, min_periods=5).mean()

# Save results
os.makedirs(os.path.dirname(OUTPUT_CSV), exist_ok=True)
a.to_csv(OUTPUT_CSV)
print(f"\nDetailed results saved to: {OUTPUT_CSV} (use Colab Files pane to download)")

# Show a sample of rows
print("\nLatest 10 rows (most recent first):")
display(a[['Open_today','PDH','PDL','dist_to_PDH','dist_to_PDL','target_side','target_price','High','Low','target_hit']].iloc[::-1].head(10))

# Plot cumulative success rate and rolling 30-day success rate
plt.figure(figsize=(11,4))
plt.plot(a.index, a['cum_success_rate'])
plt.title("Cumulative success rate: target (closest PDH/PDL) hit")
plt.xlabel("Date")
plt.ylabel("Cumulative success rate")
plt.grid(True)
plt.tight_layout()
plt.show()

plt.figure(figsize=(11,4))
plt.plot(a.index, a['rolling_30'])
plt.title("30-day rolling success rate")
plt.xlabel("Date")
plt.ylabel("Rolling success rate")
plt.grid(True)
plt.tight_layout()
plt.show()

# Weekday summary and distance quintiles
weekday_table = a.groupby('day_of_week')['target_hit'].agg(['count','sum'])
weekday_table['success_rate'] = weekday_table['sum'] / weekday_table['count']
print("\nSuccess rate by weekday:")
print(weekday_table.sort_values('count', ascending=False))

# Distance-to-target bins (quintiles)
a['dist_to_chosen'] = np.where(a['target_side']=='PDH', a['dist_to_PDH'], a['dist_to_PDL'])
a['dist_bin'] = pd.qcut(a['dist_to_chosen'], q=5, duplicates='drop')
bin_table = a.groupby('dist_bin')['target_hit'].agg(['count','sum'])
bin_table['success_rate'] = bin_table['sum'] / bin_table['count']
print("\nSuccess rate by distance-to-target quintiles:")
print(bin_table)

# Final counts: PDH vs PDL chosen
print("\nCount of chosen targets (PDH vs PDL):")
print(a['target_side'].value_counts())
# ===== End of cell =====

```

```python
# ===== Colab-ready: GC=F — lowest 40% distance filter + market-on-open simulation =====
!pip install --quiet yfinance pandas matplotlib

import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime, timedelta, timezone
import os

# -------------------- Parameters --------------------
TICKER = "GC=F"
LOOKBACK_YEARS = 5
END_DATE = None          # or "YYYY-MM-DD"
CUTOFF_QUANTILE = 0.40   # lowest 40% distances
STOP_BUFFER = 1.0        # USD buffer added to stop distance (price units)
SLIPPAGE = 0.0           # USD slippage per trade
COMMISSION = 0.0         # USD commission per round-trip
OUTPUT_CSV = "/mnt/data/gold_pdh_pdl_40pct_filter_trades.csv"
# ----------------------------------------------------

now_utc = datetime.now(timezone.utc)
period_days = int(LOOKBACK_YEARS * 365.25)
start_date = (now_utc - timedelta(days=period_days)).strftime("%Y-%m-%d")
end_date = END_DATE if END_DATE else now_utc.strftime("%Y-%m-%d")

print(f"Downloading {TICKER} daily data from {start_date} to {end_date} ...")
df = yf.download(TICKER, start=start_date, end=end_date, interval="1d", progress=True, auto_adjust=False)
if df.empty:
    raise SystemExit("No data downloaded — check ticker or date range.")

# Flatten MultiIndex if present
if isinstance(df.columns, pd.MultiIndex):
    df.columns = [c[0] if isinstance(c, tuple) else c for c in df.columns]

# Keep necessary columns and clean index
df = df[['Open','High','Low','Close','Volume']].copy()
df.index = pd.to_datetime(df.index).tz_localize(None)

# Compute PDH, PDL, distances, target side
a = df.copy()
a['PDH'] = a['High'].shift(1)
a['PDL'] = a['Low'].shift(1)
a['Open_today'] = a['Open']
a = a.dropna(subset=['PDH','PDL']).copy()

a['dist_to_PDH'] = (a['PDH'] - a['Open_today']).abs()
a['dist_to_PDL'] = (a['Open_today'] - a['PDL']).abs()
a['target_side'] = np.where(a['dist_to_PDH'] <= a['dist_to_PDL'], 'PDH', 'PDL')
a['target_price'] = np.where(a['target_side']=='PDH', a['PDH'], a['PDL'])
a['dist_to_chosen'] = np.where(a['target_side']=='PDH', a['dist_to_PDH'], a['dist_to_PDL'])

# Determine raw hit (intraday touch)
a['target_hit'] = np.where(a['target_side']=='PDH', a['High'] >= a['PDH'], a['Low'] <= a['PDL'])

# Compute cutoff value for lowest X% distances (quantile)
cutoff_value = a['dist_to_chosen'].quantile(CUTOFF_QUANTILE)
print(f"Cutoff distance value for lowest {int(CUTOFF_QUANTILE*100)}%: {cutoff_value:.4f} (price units)")

# Filter dataset
filtered = a[a['dist_to_chosen'] <= cutoff_value].copy()
print(f"Total days: {len(a)}, Days in lowest {int(CUTOFF_QUANTILE*100)}%: {len(filtered)}")

# Simulation: market-on-open
def simulate_market_on_open(df_days, stop_buffer=STOP_BUFFER, slippage=SLIPPAGE, commission=COMMISSION):
    trades = []
    for idx, row in df_days.iterrows():
        entry = row['Open_today']
        side = 'LONG' if row['target_side']=='PDH' else 'SHORT'
        target = row['target_price']
        opposite = row['PDL'] if side=='LONG' else row['PDH']
        stop_dist = abs(entry - opposite) / 2.0 + stop_buffer
        stop = entry - stop_dist if side=='LONG' else entry + stop_dist

        hit = row['target_hit']
        exit_price = target if hit else stop
        win = bool(hit)

        if side == 'LONG':
            pnl_price = exit_price - entry
        else:
            pnl_price = entry - exit_price

        pnl_after = pnl_price - slippage - commission

        trades.append({
            'date': idx,
            'side': side,
            'entry': entry,
            'exit': exit_price,
            'target': target,
            'stop': stop,
            'win': win,
            'pnl_price': pnl_price,
            'pnl_after_fees': pnl_after,
            'dist_to_chosen': row['dist_to_chosen']
        })
    return pd.DataFrame(trades).set_index('date')

trades_all = simulate_market_on_open(a)
trades_filt = simulate_market_on_open(filtered)

def summarize_trades(trades_df):
    n = len(trades_df)
    wins = trades_df['win'].sum()
    win_rate = wins / n if n>0 else np.nan
    avg_pnl = trades_df['pnl_after_fees'].mean()
    avg_win = trades_df[trades_df['win']==True]['pnl_after_fees'].mean() if wins>0 else np.nan
    avg_loss = trades_df[trades_df['win']==False]['pnl_after_fees'].mean() if n-wins>0 else np.nan
    expectancy = win_rate * (avg_win if not np.isnan(avg_win) else 0) + (1-win_rate) * (avg_loss if not np.isnan(avg_loss) else 0)
    cum_pnl = trades_df['pnl_after_fees'].cumsum()
    return {
        'trades': n,
        'wins': wins,
        'win_rate': win_rate,
        'avg_pnl': avg_pnl,
        'avg_win': avg_win,
        'avg_loss': avg_loss,
        'expectancy_per_trade': expectancy,
        'cumulative_pnl_last': cum_pnl.iloc[-1] if n>0 else 0,
        'cum_pnl_series': cum_pnl
    }

summary_all = summarize_trades(trades_all)
summary_filt = summarize_trades(trades_filt)

def print_summary(label, s):
    print(f"--- {label} ---")
    print(f"Trades: {s['trades']}, Wins: {s['wins']}, Win rate: {s['win_rate']:.2%}")
    print(f"Avg PnL (after fees) per trade: {s['avg_pnl']:.4f} price units")
    print(f"Avg win: {s['avg_win']:.4f}, Avg loss: {s['avg_loss']:.4f}")
    print(f"Expectancy per trade: {s['expectancy_per_trade']:.4f}")
    print(f"Cumulative PnL (price units): {s['cumulative_pnl_last']:.4f}\n")

print_summary("ALL DAYS (no distance filter)", summary_all)
print_summary(f"FILTERED DAYS (lowest {int(CUTOFF_QUANTILE*100)}% dist)", summary_filt)

# Show sample filtered trades
display_cols = ['side','entry','target','stop','exit','win','pnl_price','pnl_after_fees','dist_to_chosen']
print("Sample trades (filtered set, most recent 10):")
display(trades_filt[display_cols].iloc[::-1].head(10))

# Plot cumulative pnl series
plt.figure(figsize=(11,5))
plt.plot(summary_all['cum_pnl_series'].index, summary_all['cum_pnl_series'], label='All days')
plt.plot(summary_filt['cum_pnl_series'].index, summary_filt['cum_pnl_series'], label=f'Filtered (<= {int(CUTOFF_QUANTILE*100)}%)')
plt.legend()
plt.title('Cumulative PnL (price units) - Market-on-open simulation')
plt.xlabel('Date')
plt.ylabel('Cumulative PnL (price units)')
plt.grid(True)
plt.tight_layout()
plt.show()

# Save filtered trades
os.makedirs(os.path.dirname(OUTPUT_CSV), exist_ok=True)
trades_filt.to_csv(OUTPUT_CSV)
print(f"Filtered trades saved to {OUTPUT_CSV}")
# ===== End of cell =====
```

```python
# ===== Colab-ready: sweep stop multipliers and min R:R thresholds =====
!pip install --quiet yfinance pandas matplotlib

import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime, timedelta, timezone
import os

# -------------------- User parameters --------------------
TICKER = "GC=F"
LOOKBACK_YEARS = 5
END_DATE = None           # or "YYYY-MM-DD"
STOP_BUFFER = 1.0         # absolute price units added to stop distance
SLIPPAGE = 0.0            # price units per trade (conservative)
COMMISSION = 0.0          # price units per round-trip
STOP_MULTS = [0.25, 0.5, 0.75, 1.0, 1.5]   # multipliers of half-distance-to-opposite used to compute stop
MIN_RR_LIST = [0.0, 0.5, 1.0, 1.5, 2.0]    # minimum required R:R (distance_to_target / stop_distance) to accept a trade
OUTPUT_ALL_TRADES_CSV = "/mnt/data/gc_stop_rr_sweep_all_trades.csv"
OUTPUT_SUMMARY_CSV = "/mnt/data/gc_stop_rr_sweep_summary.csv"
# ----------------------------------------------------

# Helper: download & prepare data
now_utc = datetime.now(timezone.utc)
period_days = int(LOOKBACK_YEARS * 365.25)
start_date = (now_utc - timedelta(days=period_days)).strftime("%Y-%m-%d")
end_date = END_DATE if END_DATE else now_utc.strftime("%Y-%m-%d")

print(f"Downloading {TICKER} daily data from {start_date} to {end_date} ...")
df = yf.download(TICKER, start=start_date, end=end_date, interval="1d", progress=True, auto_adjust=False)
if df.empty:
    raise SystemExit("No data downloaded — check ticker or date range.")

# Flatten MultiIndex columns if present
if isinstance(df.columns, pd.MultiIndex):
    df.columns = [c[0] if isinstance(c, tuple) else c for c in df.columns]

# Keep core columns
df = df[['Open','High','Low','Close','Volume']].copy()
df.index = pd.to_datetime(df.index).tz_localize(None)

# Compute PDH/PDL & helper distances
a = df.copy()
a['PDH'] = a['High'].shift(1)
a['PDL'] = a['Low'].shift(1)
a['Open_today'] = a['Open']
a = a.dropna(subset=['PDH','PDL']).copy()

a['dist_to_PDH'] = (a['PDH'] - a['Open_today']).abs()
a['dist_to_PDL'] = (a['Open_today'] - a['PDL']).abs()
a['target_side'] = np.where(a['dist_to_PDH'] <= a['dist_to_PDL'], 'PDH', 'PDL')
a['target_price'] = np.where(a['target_side']=='PDH', a['PDH'], a['PDL'])
a['dist_to_chosen'] = np.where(a['target_side']=='PDH', a['dist_to_PDH'], a['dist_to_PDL'])

# half-distance to opposite used as base stop distance in original approach
a['opposite_price'] = np.where(a['target_side']=='PDH', a['PDL'], a['PDH'])
a['half_to_opposite'] = (a['Open_today'] - a['opposite_price']).abs() / 2.0

# whether target was touched during the day (intraday hit)
a['target_hit'] = np.where(a['target_side']=='PDH', a['High'] >= a['PDH'], a['Low'] <= a['PDL'])

# prepare structure to collect simulated trades
all_trades = []

# Sweep combinations
for stop_mult in STOP_MULTS:
    for min_rr in MIN_RR_LIST:
        trades = []
        for idx, row in a.iterrows():
            entry = row['Open_today']
            side = 'LONG' if row['target_side']=='PDH' else 'SHORT'
            target = row['target_price']
            half_op = row['half_to_opposite']
            # Stop distance based on multiplier
            stop_distance = half_op * stop_mult + STOP_BUFFER
            stop_price = entry - stop_distance if side=='LONG' else entry + stop_distance
            # Dist to target
            dist_to_target = abs(target - entry)
            # R:R (target distance / stop distance)
            rr = dist_to_target / stop_distance if stop_distance > 0 else np.nan

            # Optionally skip trades that don't meet min_rr
            if rr < min_rr:
                continue

            # Determine outcome: win if target touched, else loss at stop
            hit = bool(row['target_hit'])
            exit_price = target if hit else stop_price
            pnl_price = (exit_price - entry) if side=='LONG' else (entry - exit_price)
            pnl_after = pnl_price - SLIPPAGE - COMMISSION

            trades.append({
                'date': idx,
                'stop_mult': stop_mult,
                'min_rr': min_rr,
                'side': side,
                'entry': entry,
                'target': target,
                'stop_price': stop_price,
                'stop_distance': stop_distance,
                'dist_to_target': dist_to_target,
                'rr': rr,
                'win': hit,
                'exit': exit_price,
                'pnl_price': pnl_price,
                'pnl_after_fees': pnl_after,
                'dist_to_chosen': row['dist_to_chosen']
            })

        # collect trades into master list
        if trades:
            all_trades.extend(trades)

# Build DataFrame of all simulated trades
if len(all_trades) == 0:
    raise SystemExit("No trades simulated — check STOP_MULTS / MIN_RR settings.")

trades_df = pd.DataFrame(all_trades)
trades_df['date'] = pd.to_datetime(trades_df['date'])
trades_df = trades_df.set_index('date').sort_index()

# Summary by (stop_mult, min_rr)
summary_rows = []
grouped = trades_df.groupby(['stop_mult','min_rr'])
for (sm, mr), g in grouped:
    n = len(g)
    wins = g['win'].sum()
    win_rate = wins / n if n>0 else np.nan
    avg_pnl = g['pnl_after_fees'].mean()
    avg_win = g[g['win']==True]['pnl_after_fees'].mean() if wins>0 else np.nan
    avg_loss = g[g['win']==False]['pnl_after_fees'].mean() if n-wins>0 else np.nan
    expectancy = win_rate * (avg_win if not np.isnan(avg_win) else 0) + (1-win_rate) * (avg_loss if not np.isnan(avg_loss) else 0)
    cum_pnl = g['pnl_after_fees'].sum()
    summary_rows.append({
        'stop_mult': sm,
        'min_rr': mr,
        'trades': n,
        'wins': int(wins),
        'win_rate': win_rate,
        'avg_pnl': avg_pnl,
        'avg_win': avg_win,
        'avg_loss': avg_loss,
        'expectancy': expectancy,
        'cumulative_pnl': cum_pnl
    })

summary_df = pd.DataFrame(summary_rows).sort_values(['stop_mult','min_rr']).reset_index(drop=True)

# Print summary table nicely
pd.set_option('display.float_format', lambda x: f"{x:.4f}")
print("Summary of sweep (stop_mult x min_RR):\n")
display(summary_df)

# Save outputs
os.makedirs(os.path.dirname(OUTPUT_ALL_TRADES_CSV), exist_ok=True)
trades_df.to_csv(OUTPUT_ALL_TRADES_CSV)
summary_df.to_csv(OUTPUT_SUMMARY_CSV, index=False)
print(f"\nAll simulated trades saved to: {OUTPUT_ALL_TRADES_CSV}")
print(f"Summary table saved to: {OUTPUT_SUMMARY_CSV}")

# Quick visual: heatmap-like table of expectancy (pivot)
pivot_expectancy = summary_df.pivot(index='stop_mult', columns='min_rr', values='expectancy')
plt.figure(figsize=(8,5))
plt.title('Expectancy per trade (price units) — stop_mult rows vs min_RR columns')
plt.imshow(pivot_expectancy.astype(float), aspect='auto', cmap='RdYlGn', interpolation='nearest')
plt.colorbar(label='Expectancy (price units)')
plt.yticks(range(len(pivot_expectancy.index)), pivot_expectancy.index)
plt.xticks(range(len(pivot_expectancy.columns)), pivot_expectancy.columns)
plt.xlabel('min_RR')
plt.ylabel('stop_mult')
plt.tight_layout()
plt.show()

# Print top-performing configs by expectancy and by cumulative PnL (if ties, show more)
top_by_expectancy = summary_df.sort_values('expectancy', ascending=False).head(6)
top_by_cum = summary_df.sort_values('cumulative_pnl', ascending=False).head(6)
print("\nTop configs by expectancy (per trade):")
display(top_by_expectancy)
print("\nTop configs by cumulative PnL:")
display(top_by_cum)

# If you want, pick the best config (highest expectancy) and show sample trades:
best_cfg = top_by_expectancy.iloc[0]
best_sm = best_cfg['stop_mult']
best_mr = best_cfg['min_rr']
print(f"\nBest config chosen (by expectancy): stop_mult={best_sm}, min_rr={best_mr}")
best_trades = trades_df[(trades_df['stop_mult']==best_sm) & (trades_df['min_rr']==best_mr)]
print(f"Trades in best config: {len(best_trades)}. Sample most recent 10:")
display(best_trades[['side','entry','target','stop_price','win','pnl_after_fees','rr']].iloc[::-1].head(10))

# Save the best config trades separately
best_csv = "/mnt/data/gc_best_config_trades.csv"
best_trades.to_csv(best_csv)
print(f"Best-config trades saved to: {best_csv}")

# ===== End of cell =====
```

```python
# ===== Colab-ready: Directional filter + compare entry-at-open vs limit-on-retrace =====
!pip install --quiet yfinance pandas matplotlib

import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime, timedelta, timezone
import os

# -------------------- Parameters (tweakable) --------------------
TICKER = "GC=F"
LOOKBACK_YEARS = 5
END_DATE = None
STOP_BUFFER = 1.0            # price units added to stop distance
STOP_MULT = 0.25             # base multiplier of half-distance-to-opposite for stop (use earlier best)
SLIPPAGE = 0.0               # price units per trade
COMMISSION = 0.0             # price units per round-trip
DISTANCE_CUTOFF = None       # None or price units to require dist_to_chosen <= cutoff
RETRACE_FRAC_LIST = [0.0, 0.25, 0.5]   # 0.0 = market at open; >0 are limit retrace fractions
OUTPUT_DIR = "/mnt/data/gc_directional_entry_compare"
# ----------------------------------------------------------------

# Helper: download & prepare data
now_utc = datetime.now(timezone.utc)
period_days = int(LOOKBACK_YEARS * 365.25)
start_date = (now_utc - timedelta(days=period_days)).strftime("%Y-%m-%d")
end_date = END_DATE if END_DATE else now_utc.strftime("%Y-%m-%d")

print(f"Downloading {TICKER} daily data from {start_date} to {end_date} ...")
df = yf.download(TICKER, start=start_date, end=end_date, interval="1d", progress=True, auto_adjust=False)
if df.empty:
    raise SystemExit("No data downloaded — check ticker or date range.")

# Flatten MultiIndex if present
if isinstance(df.columns, pd.MultiIndex):
    df.columns = [c[0] if isinstance(c, tuple) else c for c in df.columns]

# Keep essentials
df = df[['Open','High','Low','Close','Volume']].copy()
df.index = pd.to_datetime(df.index).tz_localize(None)

# Compute PDH/PDL & distances
a = df.copy()
a['PDH'] = a['High'].shift(1)
a['PDL'] = a['Low'].shift(1)
a['prev_open'] = a['Open'].shift(1)
a['prev_close'] = a['Close'].shift(1)
a['Open_today'] = a['Open']
a = a.dropna(subset=['PDH','PDL','prev_open','prev_close']).copy()

a['dist_to_PDH'] = (a['PDH'] - a['Open_today']).abs()
a['dist_to_PDL'] = (a['Open_today'] - a['PDL']).abs()
a['target_side'] = np.where(a['dist_to_PDH'] <= a['dist_to_PDL'], 'PDH', 'PDL')
a['target_price'] = np.where(a['target_side']=='PDH', a['PDH'], a['PDL'])
a['dist_to_chosen'] = np.where(a['target_side']=='PDH', a['dist_to_PDH'], a['dist_to_PDL'])

# Opposite price and half-to-opposite base
a['opposite_price'] = np.where(a['target_side']=='PDH', a['PDL'], a['PDH'])
a['half_to_opposite'] = (a['Open_today'] - a['opposite_price']).abs() / 2.0

# Intraday hit (target touched at any time during day)
a['target_touched'] = np.where(a['target_side']=='PDH', a['High'] >= a['PDH'], a['Low'] <= a['PDL'])

# Directional filter signal: prev_day bullish -> prefer LONG (PDH), prev_day bearish -> prefer SHORT (PDL)
a['prev_day_dir'] = np.where(a['prev_close'] > a['prev_open'], 'BULL',
                             np.where(a['prev_close'] < a['prev_open'], 'BEAR', 'NEUTRAL'))

# Optional distance cutoff filter
if DISTANCE_CUTOFF is not None:
    a = a[a['dist_to_chosen'] <= DISTANCE_CUTOFF].copy()
    print(f"Applied distance cutoff <= {DISTANCE_CUTOFF}: remaining days = {len(a)}")

print(f"Prepared {len(a)} days for simulation.")

# Simulation engine:
# - For each retrace fraction, we either enter at open (0.0) or place a limit order:
#   * LONG (target PDH): limit entry = open - retrace_frac * dist_to_chosen  (i.e., wait for price to pull back)
#   * SHORT (target PDL): limit entry = open + retrace_frac * dist_to_chosen
# - For limit entries: assume it's filled if day's Low <= entry_price (LONG) or day's High >= entry_price (SHORT).
# - **Important**: Using daily OHLC we cannot infer ordering (entry-before-target). We approximate:
#   If entry is filled (based on day's high/low) and target_touched is True that day -> treat as win.
#   This may overcount wins on days where target happened earlier than the entry would have filled.
# - Stop: stop = entry ± (half_to_opposite * STOP_MULT + STOP_BUFFER)
# - If entry not filled (for limit), we skip that day (no trade).

def simulate_entries(df_days, retrace_frac, directional_filter=True):
    trades = []
    for idx, row in df_days.iterrows():
        # Determine allowed side by directional filter
        prev_dir = row['prev_day_dir']
        intended_side = 'LONG' if row['target_side']=='PDH' else 'SHORT'
        if directional_filter:
            if prev_dir == 'BULL' and intended_side != 'LONG':
                continue
            if prev_dir == 'BEAR' and intended_side != 'SHORT':
                continue
            if prev_dir == 'NEUTRAL':
                continue  # skip neutral prev-day candles when filtering
        # Determine entry price
        entry_open = row['Open_today']
        dist = row['dist_to_chosen']
        if retrace_frac == 0.0:
            # market at open (we take open as executed)
            entry_price = entry_open
            entry_filled = True
        else:
            if intended_side == 'LONG':
                entry_price = entry_open - retrace_frac * dist
                # filled if day's Low <= entry_price
                entry_filled = row['Low'] <= entry_price
            else:
                entry_price = entry_open + retrace_frac * dist
                entry_filled = row['High'] >= entry_price
        if not entry_filled:
            continue  # limit didn't fill -> no trade this day

        # compute stop
        stop_dist = row['half_to_opposite'] * STOP_MULT + STOP_BUFFER
        stop_price = entry_price - stop_dist if intended_side == 'LONG' else entry_price + stop_dist

        # Determine outcome: win if target_touched during day AND entry filled
        # (Note: this assumes entry happened before target in intraday sequence. See comment above.)
        touched = bool(row['target_touched'])
        win = touched

        exit_price = row['target_price'] if win else stop_price
        pnl = (exit_price - entry_price) if intended_side=='LONG' else (entry_price - exit_price)
        pnl_after = pnl - SLIPPAGE - COMMISSION

        trades.append({
            'date': idx,
            'retrace_frac': retrace_frac,
            'directional_filter': directional_filter,
            'prev_day_dir': prev_dir,
            'intended_side': intended_side,
            'entry_price': entry_price,
            'entry_filled': entry_filled,
            'target_price': row['target_price'],
            'stop_price': stop_price,
            'stop_dist': stop_dist,
            'touched': touched,
            'win': win,
            'pnl': pnl,
            'pnl_after_fees': pnl_after,
            'dist_to_chosen': row['dist_to_chosen'],
            'half_to_opposite': row['half_to_opposite']
        })
    return pd.DataFrame(trades).set_index('date')

# Run simulations for all retrace_fracs with and without directional filter
results = []
all_trades = []
for retr in RETRACE_FRAC_LIST:
    for dfilt in [True, False]:
        trades_df = simulate_entries(a, retrace_frac=retr, directional_filter=dfilt)
        if len(trades_df)==0:
            summary = {'retrace_frac': retr, 'directional_filter': dfilt, 'trades':0}
        else:
            n = len(trades_df)
            wins = int(trades_df['win'].sum())
            win_rate = wins/n
            avg_pnl = trades_df['pnl_after_fees'].mean()
            avg_win = trades_df[trades_df['win']==True]['pnl_after_fees'].mean() if wins>0 else np.nan
            avg_loss = trades_df[trades_df['win']==False]['pnl_after_fees'].mean() if n-wins>0 else np.nan
            expectancy = win_rate * (avg_win if not np.isnan(avg_win) else 0) + (1-win_rate) * (avg_loss if not np.isnan(avg_loss) else 0)
            cum_pnl = trades_df['pnl_after_fees'].sum()
            summary = {
                'retrace_frac': retr,
                'directional_filter': dfilt,
                'trades': n,
                'wins': wins,
                'win_rate': win_rate,
                'avg_pnl': avg_pnl,
                'avg_win': avg_win,
                'avg_loss': avg_loss,
                'expectancy': expectancy,
                'cumulative_pnl': cum_pnl
            }
        results.append(summary)
        if len(trades_df)>0:
            trades_df['retrace_frac'] = retr
            trades_df['directional_filter'] = dfilt
            all_trades.append(trades_df)

results_df = pd.DataFrame(results).sort_values(['directional_filter','retrace_frac'], ascending=[False,True])
print("\nSimulation summary (retrace_frac rows x directional_filter columns):")
display(results_df)

# Combine all trades for CSV output
if all_trades:
    all_trades_df = pd.concat(all_trades).sort_index()
    os.makedirs(OUTPUT_DIR, exist_ok=True)
    all_trades_df.to_csv(os.path.join(OUTPUT_DIR, "gc_directional_entry_all_trades.csv"))
    results_df.to_csv(os.path.join(OUTPUT_DIR, "gc_directional_entry_summary.csv"), index=False)
    print(f"\nSaved detailed trades and summary to {OUTPUT_DIR}")
else:
    print("\nNo trades were generated by these settings.")

# Quick plots: expectancy & cumulative pnl for each scenario
plt.figure(figsize=(10,5))
for retr in RETRACE_FRAC_LIST:
    for dfilt in [True, False]:
        row = results_df[(results_df['retrace_frac']==retr) & (results_df['directional_filter']==dfilt)]
        if len(row)==0 or row['trades'].iloc[0]==0: continue
        label = f"retr={retr}, dir={dfilt}"
        # show cumulative pnl
        trades_series = all_trades_df[(all_trades_df['retrace_frac']==retr) & (all_trades_df['directional_filter']==dfilt)]['pnl_after_fees']
        if trades_series is not None and len(trades_series)>0:
            plt.plot(trades_series.cumsum().index, trades_series.cumsum().values, label=label)
plt.legend()
plt.title("Cumulative PnL for each retrace_frac and directional_filter scenario")
plt.xlabel("Date")
plt.ylabel("Cumulative PnL (price units)")
plt.grid(True)
plt.tight_layout()
plt.show()

# Print best scenario by expectancy
best = results_df.sort_values('expectancy', ascending=False).iloc[0]
print("\nBest scenario by expectancy:")
display(best)

# Show sample trades for best scenario
best_trades = all_trades_df[(all_trades_df['retrace_frac']==best['retrace_frac']) &
                            (all_trades_df['directional_filter']==best['directional_filter'])]
print(f"\nSample (most recent 10) trades for best scenario (retr={best['retrace_frac']}, dir={best['directional_filter']}):")
display(best_trades[['intended_side','entry_price','target_price','stop_price','win','pnl_after_fees','dist_to_chosen']].iloc[::-1].head(10))

# End of cell
```

```python
# Intraday (5-min) validation for PDH/PDL "closest to open" strategy on GC=F (Gold futures)
# Colab-ready cell: fetches daily and 5-min intraday data (yfinance intraday is usually limited to ~60 days),
# simulates realistic order sequence (entry fill then target/stop after entry) for retrace fractions,
# and summarizes results.
#
# Paste into Google Colab and run. Outputs CSVs under /mnt/data and prints summaries.
# Adjust PARAMETERS below as needed.

!pip install --quiet yfinance pandas matplotlib tqdm

import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime, timedelta, timezone
import os, time
from tqdm import tqdm

# ================= USER PARAMETERS =================
TICKER = "GC=F"                # Yahoo continuous gold futures
LOOKBACK_DAYS_INTRADAY = 60    # max ~60 days for 5m via yfinance; adjust <=60
RETRACE_FRACS = [0.5, 0.25, 0.0]  # test 50% retrace, 25%, or market-at-open (0.0)
STOP_MULT = 0.25               # as earlier
STOP_BUFFER = 1.0
SLIPPAGE = 0.0
COMMISSION = 0.0
OUTPUT_DIR = "/mnt/data/gc_intraday_validation"
SLEEP_BETWEEN_REQUESTS = 0.5   # seconds - polite, avoid throttling
# ==================================================

# Helper: dates for intraday fetch
end_utc = datetime.now(timezone.utc)
start_utc = end_utc - timedelta(days=LOOKBACK_DAYS_INTRADAY*1.3)  # a bit of buffer to get trading days
start_date = start_utc.strftime("%Y-%m-%d")
end_date = end_utc.strftime("%Y-%m-%d")
print(f"Will fetch daily data from {start_date} to {end_date} and intraday for last ~{LOOKBACK_DAYS_INTRADAY} days.")

# Download daily data for PDH/PDL calculation (longer window OK)
daily = yf.download(TICKER, start=start_date, end=end_date, interval="1d", progress=True, auto_adjust=False)
if daily.empty:
    raise SystemExit("No daily data downloaded. Check ticker/dates.")

# Flatten cols if needed
if isinstance(daily.columns, pd.MultiIndex):
    daily.columns = [c[0] if isinstance(c, tuple) else c for c in daily.columns]
daily = daily[['Open','High','Low','Close','Volume']].copy()
daily.index = pd.to_datetime(daily.index).tz_localize(None)

# Build list of target trading days for intraday check: recent N trading days where we have previous day
daily = daily.dropna()
daily['PDH'] = daily['High'].shift(1)
daily['PDL'] = daily['Low'].shift(1)
daily['Open_today'] = daily['Open']
daily = daily.dropna(subset=['PDH','PDL']).copy()

recent_days = daily.iloc[-LOOKBACK_DAYS_INTRADAY:].copy()
print(f"Preparing intraday validation for {len(recent_days)} days (most recent).")

# Helper: fetch intraday 5m for a single day using yfinance by specifying start/end datetimes
def fetch_intraday_day(ticker, day_date):
    # day_date is a pandas Timestamp (naive) representing calendar day in UTC date zone for simplicity
    start = pd.Timestamp(day_date.date(), tz=timezone.utc)
    end = start + pd.Timedelta(days=1)
    # yfinance expects strings; interval 5m
    try:
        df = yf.download(ticker, start=start.strftime("%Y-%m-%d"), end=end.strftime("%Y-%m-%d"), interval="5m", progress=False, auto_adjust=False)
    except Exception as e:
        print(f"Error fetching intraday for {day_date.date()}: {e}")
        return pd.DataFrame()
    if df.empty:
        return df
    # flatten columns
    if isinstance(df.columns, pd.MultiIndex):
        df.columns = [c[0] if isinstance(c, tuple) else c for c in df.columns]
    df = df[['Open','High','Low','Close','Volume']].copy()
    df.index = pd.to_datetime(df.index).tz_localize(None)
    return df

# Function: simulate intraday sequence for a day given retrace_frac and stop settings
def simulate_day_intraday(day_row, intraday_df, retrace_frac, stop_mult=STOP_MULT, stop_buffer=STOP_BUFFER):
    # day_row: a row from daily with PDH, PDL, Open_today etc.
    # intraday_df: 5-min bars for that day
    # Returns: dict with trade result or None if no trade/entry not filled
    if intraday_df is None or intraday_df.empty:
        return None  # no intraday data

    entry_open = float(day_row['Open_today'])
    target_side = 'PDH' if abs(day_row['PDH'] - entry_open) <= abs(entry_open - day_row['PDL']) else 'PDL'
    target_price = float(day_row['PDH'] if target_side=='PDH' else day_row['PDL'])
    dist_to_chosen = abs(target_price - entry_open)

    # intended side LONG if target is PDH (price needs to go up to hit PDH), else SHORT
    intended_side = 'LONG' if target_side=='PDH' else 'SHORT'
    # define intended limit entry
    if retrace_frac == 0.0:
        entry_price = entry_open
    else:
        entry_price = entry_open - retrace_frac * dist_to_chosen if intended_side=='LONG' else entry_open + retrace_frac * dist_to_chosen

    # Determine stop price based on half_to_opposite
    opposite_price = float(day_row['PDL'] if target_side=='PDH' else day_row['PDH'])
    half_to_opposite = abs(entry_open - opposite_price) / 2.0
    stop_dist = half_to_opposite * stop_mult + stop_buffer
    stop_price = entry_price - stop_dist if intended_side=='LONG' else entry_price + stop_dist

    # Now scan intraday bars in chronological order to model realistic sequence:
    # 1) If entry is limit (retrace_frac>0), we wait until intraday low/high crosses entry_price for LONG/SHORT.
    # 2) After entry is filled at the first crossing (we assume filled at the entry_price), we then watch subsequent bars for
    #    target being hit (High >= target for LONG / Low <= target for SHORT) OR stop being hit (Low <= stop for LONG / High >= stop for SHORT).
    # 3) If neither hit by close, we mark as loss at stop (conservative).
    entry_filled = False
    entry_fill_time = None
    # iterate rows
    for ts, bar in intraday_df.iterrows():
        if not entry_filled:
            # check whether entry would be filled on this bar
            if intended_side == 'LONG':
                # limit fill if bar.Low <= entry_price
                if bar['Low'] <= entry_price:
                    entry_filled = True
                    entry_fill_time = ts
            else:
                # SHORT: bar.High >= entry_price
                if bar['High'] >= entry_price:
                    entry_filled = True
                    entry_fill_time = ts
        else:
            # After entry, check for target/stop on the same or subsequent bars
            if intended_side == 'LONG':
                # target hit?
                if bar['High'] >= target_price:
                    return {'entry_filled': True, 'filled_at': entry_fill_time, 'win': True, 'pnl_price': target_price - entry_price, 'entry_price': entry_price, 'exit_price': target_price}
                # stop hit?
                if bar['Low'] <= stop_price:
                    return {'entry_filled': True, 'filled_at': entry_fill_time, 'win': False, 'pnl_price': stop_price - entry_price, 'entry_price': entry_price, 'exit_price': stop_price}
            else: # SHORT
                if bar['Low'] <= target_price:
                    return {'entry_filled': True, 'filled_at': entry_fill_time, 'win': True, 'pnl_price': entry_price - target_price, 'entry_price': entry_price, 'exit_price': target_price}
                if bar['High'] >= stop_price:
                    return {'entry_filled': True, 'filled_at': entry_fill_time, 'win': False, 'pnl_price': entry_price - stop_price, 'entry_price': entry_price, 'exit_price': stop_price}
    # If entry was never filled, return None (no trade)
    if not entry_filled:
        return None
    # If entry filled but neither target nor stop hit by end of day, assume conservative loss at stop
    return {'entry_filled': True, 'filled_at': entry_fill_time, 'win': False, 'pnl_price': (stop_price - entry_price) if intended_side=='LONG' else (entry_price - stop_price), 'entry_price': entry_price, 'exit_price': stop_price}

# Loop over recent days and fetch intraday data per day, then simulate for each retrace fraction
results = []
all_trades = []

days_iter = list(recent_days.index)

for day in tqdm(days_iter, desc="Days"):
    day_row = recent_days.loc[day]
    # fetch intraday for that calendar day
    intraday = fetch_intraday_day(TICKER, day)
    time.sleep(SLEEP_BETWEEN_REQUESTS)
    if intraday.empty:
        # skip day if no intraday bars
        continue
    for retr in RETRACE_FRACS:
        res = simulate_day_intraday(day_row, intraday, retrace_frac=retr)
        if res is None:
            continue
        # convert pnl_price to USD per contract for gold (multiplier = 100 oz per contract)
        pnl_usd = res['pnl_price'] * 100.0 - SLIPPAGE - COMMISSION
        results.append({'date': day, 'retrace_frac': retr, 'win': res['win'], 'pnl_price': res['pnl_price'], 'pnl_usd': pnl_usd})
        all_trades.append({'date': day, 'retrace_frac': retr, **res, 'pnl_usd': pnl_usd})

# Build DataFrame and summary
if len(results) == 0:
    raise SystemExit("No intraday trades simulated. Check intraday availability and LOOKBACK_DAYS_INTRADAY.")

trades_df = pd.DataFrame(all_trades).set_index('date').sort_index()
summary = trades_df.groupby('retrace_frac').agg(trades=('win','count'), wins=('win','sum'), win_rate=('win','mean'), avg_pnl_price=('pnl_price','mean'), avg_pnl_usd=('pnl_usd','mean'))
summary['expectancy_price'] = trades_df.groupby('retrace_frac')['pnl_price'].mean()
summary['cumulative_usd'] = trades_df.groupby('retrace_frac')['pnl_usd'].sum()

print("\nIntraday validation summary (5-min):")
display(summary)

# Save outputs
os.makedirs(OUTPUT_DIR, exist_ok=True)
trades_df.to_csv(os.path.join(OUTPUT_DIR, "gc_intraday_trades.csv"))
summary.to_csv(os.path.join(OUTPUT_DIR, "gc_intraday_summary.csv"))
print(f"\nSaved intraday trades and summary to {OUTPUT_DIR}")

# Quick plot of cumulative USD PnL per retrace frac
plt.figure(figsize=(10,5))
for retr in sorted(trades_df['retrace_frac'].unique()):
    series = trades_df[trades_df['retrace_frac']==retr]['pnl_usd'].cumsum()
    if len(series)==0: continue
    plt.plot(series.index, series.values, label=f"retr={retr}")
plt.legend()
plt.title("Intraday cumulative USD PnL per retrace frac (5-min validation)")
plt.xlabel("Date")
plt.ylabel("Cumulative PnL (USD)")
plt.grid(True)
plt.tight_layout()
plt.show()

# Display most recent trades
print("\nMost recent intraday trades (sample):")
display(trades_df[['retrace_frac','entry_price','exit_price','win','pnl_price','pnl_usd']].iloc[::-1].head(20))

# End of intraday validation cell.
```

```python
# ===== Colab-ready: Intraday 5m validation — compare ATR stop vs original stop, exit-at-close =====
!pip install --quiet yfinance pandas matplotlib tqdm

import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime, timedelta, timezone
import os, time
from tqdm import tqdm

# ================= USER PARAMETERS =================
TICKER = "GC=F"                # Yahoo continuous gold futures
LOOKBACK_DAYS_INTRADAY = 60    # keep <= ~60 for yfinance 5m
RETRACE_FRAC = 0.5             # use 50% retrace only
STOP_MULT = 0.25               # original half-to-opposite multiplier
STOP_BUFFER = 1.0              # original buffer
ATR_MULT = 1.0                 # ATR multiple for ATR-based stop (try 1.0; you can tweak)
ATR_PERIOD = 14                # ATR periods on 5m
SLIPPAGE = 0.0
COMMISSION = 0.0
OUTPUT_CSV = "/mnt/data/gc_intraday_atr_vs_original.csv"
OUTPUT_SUMMARY = "/mnt/data/gc_intraday_atr_vs_original_summary.csv"
LOOKBACK_FETCH_BUFFER = 1.3    # multiplier for fetch window
SLEEP_BETWEEN_REQUESTS = 0.4
# ==================================================

def compute_atr(df_ohlc, period=14):
    """Compute ATR (True Range) on ohlc DataFrame. Returns series aligned with df index."""
    high = df_ohlc['High']
    low = df_ohlc['Low']
    close = df_ohlc['Close']
    tr1 = high - low
    tr2 = (high - close.shift(1)).abs()
    tr3 = (low - close.shift(1)).abs()
    tr = pd.concat([tr1, tr2, tr3], axis=1).max(axis=1)
    atr = tr.rolling(window=period, min_periods=1).mean()  # simple rolling mean ATR
    return atr

# Prepare dates
end_utc = datetime.now(timezone.utc)
start_utc = end_utc - timedelta(days=int(LOOKBACK_DAYS_INTRADAY * LOOKBACK_FETCH_BUFFER))
start_date = start_utc.strftime("%Y-%m-%d")
end_date = end_utc.strftime("%Y-%m-%d")

print(f"Fetching daily data from {start_date} to {end_date} and intraday for last ~{LOOKBACK_DAYS_INTRADAY} days.")

# Download daily data for PDH/PDL
daily = yf.download(TICKER, start=start_date, end=end_date, interval="1d", progress=True, auto_adjust=False)
if daily.empty:
    raise SystemExit("No daily data downloaded. Check ticker/dates and internet connectivity.")

# Flatten columns if necessary and keep essentials
if isinstance(daily.columns, pd.MultiIndex):
    daily.columns = [c[0] if isinstance(c, tuple) else c for c in daily.columns]
daily = daily[['Open','High','Low','Close','Volume']].copy()
daily.index = pd.to_datetime(daily.index).tz_localize(None)

# Prepare recent days with PDH/PDL
daily['PDH'] = daily['High'].shift(1)
daily['PDL'] = daily['Low'].shift(1)
daily['Open_today'] = daily['Open']
daily = daily.dropna(subset=['PDH','PDL'])
recent_days = daily.iloc[-LOOKBACK_DAYS_INTRADAY:].copy()
print(f"Prepared {len(recent_days)} recent days for intraday validation.")

# Helper to fetch intraday 5m for a calendar day
def fetch_intraday_day(ticker, day_timestamp):
    start = pd.Timestamp(day_timestamp.date(), tz=timezone.utc)
    end = start + pd.Timedelta(days=1)
    try:
        df = yf.download(ticker, start=start.strftime("%Y-%m-%d"), end=end.strftime("%Y-%m-%d"), interval="5m", progress=False, auto_adjust=False)
    except Exception as e:
        print(f"Error fetching intraday for {day_timestamp.date()}: {e}")
        return pd.DataFrame()
    if df.empty:
        return df
    if isinstance(df.columns, pd.MultiIndex):
        df.columns = [c[0] if isinstance(c, tuple) else c for c in df.columns]
    df = df[['Open','High','Low','Close','Volume']].copy()
    df.index = pd.to_datetime(df.index).tz_localize(None)
    return df

# Intraday simulation for one day: returns dicts for ATR-stop and original-stop variants (or None)
def simulate_day_with_stops(day_row, intraday_df):
    entry_open = float(day_row['Open_today'])
    # decide which prev extreme is closer
    if abs(day_row['PDH'] - entry_open) <= abs(entry_open - day_row['PDL']):
        target_side = 'PDH'
        target_price = float(day_row['PDH'])
        intended_side = 'LONG'
    else:
        target_side = 'PDL'
        target_price = float(day_row['PDL'])
        intended_side = 'SHORT'
    dist_to_chosen = abs(target_price - entry_open)

    # entry price for retrace_frac
    if RETRACE_FRAC == 0.0:
        entry_price = entry_open
    else:
        entry_price = entry_open - RETRACE_FRAC * dist_to_chosen if intended_side=='LONG' else entry_open + RETRACE_FRAC * dist_to_chosen

    # compute original stop (half-to-opposite * STOP_MULT + buffer)
    opposite_price = float(day_row['PDL'] if target_side=='PDH' else day_row['PDH'])
    half_to_opposite = abs(entry_open - opposite_price) / 2.0
    orig_stop_dist = half_to_opposite * STOP_MULT + STOP_BUFFER
    orig_stop_price = entry_price - orig_stop_dist if intended_side=='LONG' else entry_price + orig_stop_dist

    # compute intraday ATR series (we'll use ATR computed on intraday bars up to the moment of entry)
    atr_series = compute_atr(intraday_df, period=ATR_PERIOD)

    # Now scan intraday bars for realistic order:
    entry_filled = False
    filled_at = None
    # we need to iterate bars in time order
    bars = intraday_df.itertuples()
    # We'll accumulate bars to compute ATR up to current index if needed
    intraday_indexed = intraday_df  # alias
    # store results per stop type
    res_orig = None
    res_atr = None

    for idx, bar in intraday_df.iterrows():
        if not entry_filled:
            # check if entry would be filled at this bar
            if intended_side == 'LONG':
                if bar['Low'] <= entry_price:
                    entry_filled = True
                    filled_at = idx
                    # compute ATR value up to and including this bar
                    atr_val = atr_series.loc[:idx].iloc[-1] if len(atr_series.loc[:idx])>0 else np.nan
                    # compute ATR stop distance
                    if not np.isnan(atr_val) and atr_val>0:
                        atr_stop_dist = atr_val * ATR_MULT
                        atr_stop_price = entry_price - atr_stop_dist
                    else:
                        atr_stop_dist = np.nan
                        atr_stop_price = np.nan
            else:  # SHORT
                if bar['High'] >= entry_price:
                    entry_filled = True
                    filled_at = idx
                    atr_val = atr_series.loc[:idx].iloc[-1] if len(atr_series.loc[:idx])>0 else np.nan
                    if not np.isnan(atr_val) and atr_val>0:
                        atr_stop_dist = atr_val * ATR_MULT
                        atr_stop_price = entry_price + atr_stop_dist
                    else:
                        atr_stop_dist = np.nan
                        atr_stop_price = np.nan
            # if entry isn't filled yet, continue scanning
            if not entry_filled:
                continue
            # once entry is filled, we initialize tracking for both stop types and continue to next bars to detect target/stop
            # note: we assume entry fill at entry_price
            continue  # go to next bar to check outcome
        else:
            # After entry, outcome detection
            if intended_side == 'LONG':
                # target hit?
                if bar['High'] >= target_price:
                    # win for both stop rules
                    res_orig = {'entry_price': entry_price, 'exit_price': target_price, 'win': True, 'pnl_price': target_price - entry_price, 'filled_at': filled_at}
                    res_atr = {'entry_price': entry_price, 'exit_price': target_price, 'win': True, 'pnl_price': target_price - entry_price, 'filled_at': filled_at}
                    return res_orig, res_atr  # both win
                # original stop hit?
                if bar['Low'] <= orig_stop_price:
                    res_orig = {'entry_price': entry_price, 'exit_price': orig_stop_price, 'win': False, 'pnl_price': orig_stop_price - entry_price, 'filled_at': filled_at}
                    # ATR-stop may or may not have triggered yet; check ATR stop if defined
                    if not np.isnan(atr_stop_price):
                        if bar['Low'] <= atr_stop_price:
                            res_atr = {'entry_price': entry_price, 'exit_price': atr_stop_price, 'win': False, 'pnl_price': atr_stop_price - entry_price, 'filled_at': filled_at}
                        else:
                            res_atr = None  # may still be open for ATR stop
                    else:
                        res_atr = None
                    # original stopped; but ATR-stop may still be open — continue scanning for ATR-stop outcome
                    # we don't return yet; we keep iterating so ATR-stop result can be detected
                # ATR stop hit?
                if not np.isnan(atr_stop_price) and bar['Low'] <= atr_stop_price:
                    res_atr = {'entry_price': entry_price, 'exit_price': atr_stop_price, 'win': False, 'pnl_price': atr_stop_price - entry_price, 'filled_at': filled_at}
                    # original stop may have been hit already (res_orig set). We keep both results.
                    if res_orig is not None:
                        return res_orig, res_atr
            else:  # SHORT
                if bar['Low'] <= target_price:
                    res_orig = {'entry_price': entry_price, 'exit_price': target_price, 'win': True, 'pnl_price': entry_price - target_price, 'filled_at': filled_at}
                    res_atr = {'entry_price': entry_price, 'exit_price': target_price, 'win': True, 'pnl_price': entry_price - target_price, 'filled_at': filled_at}
                    return res_orig, res_atr
                if bar['High'] >= orig_stop_price:
                    res_orig = {'entry_price': entry_price, 'exit_price': orig_stop_price, 'win': False, 'pnl_price': entry_price - orig_stop_price, 'filled_at': filled_at}
                if not np.isnan(atr_stop_price) and bar['High'] >= atr_stop_price:
                    res_atr = {'entry_price': entry_price, 'exit_price': atr_stop_price, 'win': False, 'pnl_price': entry_price - atr_stop_price, 'filled_at': filled_at}
                    if res_orig is not None:
                        return res_orig, res_atr
            # continue scanning until end of day
            continue

    # End of day reached: determine exit-at-close behavior
    # If entry was never filled, return (no trade)
    if not entry_filled:
        return None, None

    # If after scanning day one or both stops weren't hit and target wasn't hit, exit at close for both (per request)
    close_price = float(intraday_df['Close'].iloc[-1])
    # For original stop result: if res_orig is None or it was not a win, mark exit at close (if target not hit)
    if res_orig is None:
        pnl_orig_price = (target_price - entry_price) if intended_side=='LONG' else (entry_price - target_price)
        # But target not hit, so use close exit:
        pnl_orig_price = (close_price - entry_price) if intended_side=='LONG' else (entry_price - close_price)
        res_orig = {'entry_price': entry_price, 'exit_price': close_price, 'win': False, 'pnl_price': pnl_orig_price, 'filled_at': filled_at}
    if res_atr is None:
        # If ATR stop couldn't be computed (nan), we'll mark ATR path as None to indicate skip
        if np.isnan(atr_stop_price):
            res_atr = None
        else:
            pnl_atr_price = (close_price - entry_price) if intended_side=='LONG' else (entry_price - close_price)
            res_atr = {'entry_price': entry_price, 'exit_price': close_price, 'win': False, 'pnl_price': pnl_atr_price, 'filled_at': filled_at}
    return res_orig, res_atr

# Loop over recent days and fetch intraday data
all_results = []
days = list(recent_days.index)
for day in tqdm(days, desc="Processing days"):
    day_row = recent_days.loc[day]
    intraday = fetch_intraday_day(TICKER, day)
    time.sleep(SLEEP_BETWEEN_REQUESTS)
    if intraday.empty:
        continue
    # simulate
    res_orig, res_atr = simulate_day_with_stops(day_row, intraday)
    # record results with conversion to USD per contract (gold multiplier = 100 oz)
    if res_orig is not None:
        pnl_usd_orig = res_orig['pnl_price'] * 100.0 - SLIPPAGE - COMMISSION
        all_results.append({'date': day, 'variant': 'original_stop_exit_at_close', 'entry_price': res_orig['entry_price'], 'exit_price': res_orig['exit_price'], 'win': res_orig['win'], 'pnl_price': res_orig['pnl_price'], 'pnl_usd': pnl_usd_orig})
    if res_atr is not None:
        pnl_usd_atr = res_atr['pnl_price'] * 100.0 - SLIPPAGE - COMMISSION
        all_results.append({'date': day, 'variant': 'atr_stop_exit_at_close', 'entry_price': res_atr['entry_price'], 'exit_price': res_atr['exit_price'], 'win': res_atr['win'], 'pnl_price': res_atr['pnl_price'], 'pnl_usd': pnl_usd_atr})

# Build summary
if len(all_results) == 0:
    raise SystemExit("No intraday trades simulated — maybe intraday data was not available for recent days.")

results_df = pd.DataFrame(all_results)
results_df['date'] = pd.to_datetime(results_df['date'])
results_df = results_df.set_index('date').sort_index()

summary = results_df.groupby('variant').agg(trades=('win','count'), wins=('win','sum'), win_rate=('win','mean'), avg_pnl_price=('pnl_price','mean'), avg_pnl_usd=('pnl_usd','mean'), cumulative_usd=('pnl_usd','sum'))
print("\nIntraday ATR vs Original stop summary (exit at close if neither hit):")
display(summary)

# Save detailed results and summary
os.makedirs(os.path.dirname(OUTPUT_CSV), exist_ok=True)
results_df.to_csv(OUTPUT_CSV)
summary.to_csv(OUTPUT_SUMMARY)
print(f"\nSaved detailed results to: {OUTPUT_CSV}")
print(f"Saved summary to: {OUTPUT_SUMMARY}")

# Quick plot: cumulative USD PnL per variant
plt.figure(figsize=(10,5))
for variant in results_df['variant'].unique():
    s = results_df[results_df['variant']==variant]['pnl_usd'].cumsum()
    if len(s) > 0:
        plt.plot(s.index, s.values, label=variant)
plt.legend()
plt.title("Cumulative USD PnL: ATR-stop vs original-stop (exit at close)")
plt.xlabel("Date")
plt.ylabel("Cumulative PnL (USD)")
plt.grid(True)
plt.tight_layout()
plt.show()

# Show sample trades
print("\nSample most recent results:")
display(results_df[['variant','entry_price','exit_price','win','pnl_price','pnl_usd']].iloc[::-1].head(30))

# End of cell
```

```python
# ===== Colab-ready: Enhanced intraday backtest (ATR stop + dynamic targets + metrics) =====
!pip install --quiet yfinance pandas matplotlib tqdm numpy scipy

import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime, timedelta, timezone
import os, time
from tqdm import tqdm
from scipy import stats

# ================= USER PARAMETERS (tweak these) =================
TICKER = "GC=F"
LOOKBACK_DAYS_INTRADAY = 60    # <= ~60 for yfinance 5m
RETRACE_FRAC = 0.5             # 0.5 retrace (limit)
ATR_PERIOD = 14                # ATR on 5m bars
ATR_STOP_MULT = 1.0            # ATR * this = stop distance
ATR_TARGET_MULT = 1.5          # ATR * this = target distance (if target_mode='atr')
TARGET_MODE = 'atr'            # options: 'atr' (ATR-based target) or 'prev_extreme' (PDH/PDL)
EXIT_AT_CLOSE = True           # if neither target nor stop hit, exit at close
SLIPPAGE = 0.0                 # USD per contract per trade (adjust)
COMMISSION = 0.0               # USD per contract per trade (adjust)
GOLD_MULTIPLIER = 100.0        # USD per 1 price unit
OUTPUT_DIR = "/mnt/data/gc_enhanced_intraday"
SLEEP_BETWEEN_REQUESTS = 0.25
# =================================================================

def compute_atr(df_ohlc, period=14):
    high = df_ohlc['High']
    low = df_ohlc['Low']
    close = df_ohlc['Close']
    tr1 = high - low
    tr2 = (high - close.shift(1)).abs()
    tr3 = (low - close.shift(1)).abs()
    tr = pd.concat([tr1, tr2, tr3], axis=1).max(axis=1)
    atr = tr.rolling(window=period, min_periods=1).mean()
    return atr

# prepare date window
end_utc = datetime.now(timezone.utc)
start_utc = end_utc - timedelta(days=int(LOOKBACK_DAYS_INTRADAY * 1.3))
start_date = start_utc.strftime("%Y-%m-%d")
end_date = end_utc.strftime("%Y-%m-%d")
print(f"Fetching daily data from {start_date} to {end_date} and intraday for last ~{LOOKBACK_DAYS_INTRADAY} days.")

# daily data for PDH/PDL
daily = yf.download(TICKER, start=start_date, end=end_date, interval="1d", progress=True, auto_adjust=False)
if daily.empty:
    raise SystemExit("No daily data downloaded. Check ticker/dates and internet connectivity.")
if isinstance(daily.columns, pd.MultiIndex):
    daily.columns = [c[0] if isinstance(c, tuple) else c for c in daily.columns]
daily = daily[['Open','High','Low','Close','Volume']].copy()
daily.index = pd.to_datetime(daily.index).tz_localize(None)
daily['PDH'] = daily['High'].shift(1)
daily['PDL'] = daily['Low'].shift(1)
daily['Open_today'] = daily['Open']
daily = daily.dropna(subset=['PDH','PDL']).copy()
recent_days = daily.iloc[-LOOKBACK_DAYS_INTRADAY:].copy()
print(f"Prepared {len(recent_days)} recent days for intraday validation.")

# helper to fetch intraday 5-min for a given calendar day
def fetch_intraday_day(ticker, day_timestamp):
    start = pd.Timestamp(day_timestamp.date(), tz=timezone.utc)
    end = start + pd.Timedelta(days=1)
    try:
        df = yf.download(ticker, start=start.strftime("%Y-%m-%d"), end=end.strftime("%Y-%m-%d"), interval="5m", progress=False, auto_adjust=False)
    except Exception as e:
        # return empty DataFrame on error
        print(f"Fetch error for {day_timestamp.date()}: {e}")
        return pd.DataFrame()
    if df.empty:
        return df
    if isinstance(df.columns, pd.MultiIndex):
        df.columns = [c[0] if isinstance(c, tuple) else c for c in df.columns]
    df = df[['Open','High','Low','Close','Volume']].copy()
    df.index = pd.to_datetime(df.index).tz_localize(None)
    return df

# simulate day realistically with sequence
def simulate_day(day_row, intraday_df):
    """
    Returns a dict with results or None if entry never filled.
    Calculates entry price (limit), ATR at entry, ATR-stop, ATR-target or prev-extreme target.
    Sequence: wait for entry fill, then watch bars for target or stop. If none, exit at close if EXIT_AT_CLOSE True.
    """
    if intraday_df.empty:
        return None
    entry_open = float(day_row['Open_today'])
    # choose closer prev extreme
    pdh = float(day_row['PDH'])
    pdl = float(day_row['PDL'])
    if abs(pdh - entry_open) <= abs(entry_open - pdl):
        target_side = 'PDH'
        prev_extreme = pdh
        intended_side = 'LONG'
    else:
        target_side = 'PDL'
        prev_extreme = pdl
        intended_side = 'SHORT'
    dist_to_prev_extreme = abs(prev_extreme - entry_open)
    # entry (limit) price
    if RETRACE_FRAC == 0.0:
        entry_price = entry_open
    else:
        entry_price = entry_open - RETRACE_FRAC * dist_to_prev_extreme if intended_side=='LONG' else entry_open + RETRACE_FRAC * dist_to_prev_extreme

    # compute ATR series for day (over intraday bars)
    atr_series = compute_atr(intraday_df, period=ATR_PERIOD)
    # baseline ATR value at any time will be atr_series.loc[:bar_time].iloc[-1]
    entry_filled = False
    entry_fill_time = None
    atr_val_at_fill = np.nan
    # compute original stop (half-to-opposite)
    opposite = pdl if target_side=='PDH' else pdh
    half_to_opposite = abs(entry_open - opposite) / 2.0
    orig_stop_dist = half_to_opposite * STOP_MULT + STOP_BUFFER
    orig_stop_price = entry_price - orig_stop_dist if intended_side=='LONG' else entry_price + orig_stop_dist

    # iterate intraday bars (chronological)
    for ts, bar in intraday_df.iterrows():
        if not entry_filled:
            # check fill (limit)
            if intended_side == 'LONG':
                if bar['Low'] <= entry_price:
                    entry_filled = True
                    entry_fill_time = ts
                    # ATR at fill
                    atr_val_at_fill = atr_series.loc[:ts].iloc[-1] if len(atr_series.loc[:ts])>0 else np.nan
            else: # SHORT
                if bar['High'] >= entry_price:
                    entry_filled = True
                    entry_fill_time = ts
                    atr_val_at_fill = atr_series.loc[:ts].iloc[-1] if len(atr_series.loc[:ts])>0 else np.nan
            if not entry_filled:
                continue
            # once filled, compute ATR-stop and ATR-target if possible
            if not np.isnan(atr_val_at_fill) and atr_val_at_fill > 0:
                atr_stop_dist = atr_val_at_fill * ATR_STOP_MULT
                atr_stop_price = entry_price - atr_stop_dist if intended_side=='LONG' else entry_price + atr_stop_dist
                atr_target_dist = atr_val_at_fill * ATR_TARGET_MULT
                atr_target_price = entry_price + atr_target_dist if intended_side=='LONG' else entry_price - atr_target_dist
            else:
                atr_stop_price = np.nan
                atr_target_price = np.nan
            # define prev_extreme target price (cap)
            prev_ext_target_price = prev_extreme
            # Now continue to next bars and check outcome
            continue
        else:
            # entry filled; check target/stop hits
            if intended_side == 'LONG':
                # target: if using ATR target, check atr_target_price; if using prev_extreme, use prev_ext_target_price
                target_hit = False
                if TARGET_MODE == 'atr' and not np.isnan(atr_target_price):
                    if bar['High'] >= atr_target_price:
                        return {'entry_price': entry_price, 'exit_price': atr_target_price, 'win': True, 'pnl_price': atr_target_price - entry_price, 'variant': 'atr_target_atr_stop', 'atr_at_fill': atr_val_at_fill}
                if TARGET_MODE == 'prev_extreme':
                    if bar['High'] >= prev_ext_target_price:
                        return {'entry_price': entry_price, 'exit_price': prev_ext_target_price, 'win': True, 'pnl_price': prev_ext_target_price - entry_price, 'variant': 'prev_ext_target_atr_stop', 'atr_at_fill': atr_val_at_fill}
                # stop checks
                if not np.isnan(atr_stop_price) and bar['Low'] <= atr_stop_price:
                    return {'entry_price': entry_price, 'exit_price': atr_stop_price, 'win': False, 'pnl_price': atr_stop_price - entry_price, 'variant': 'atr_target_atr_stop', 'atr_at_fill': atr_val_at_fill}
                if bar['Low'] <= orig_stop_price:
                    return {'entry_price': entry_price, 'exit_price': orig_stop_price, 'win': False, 'pnl_price': orig_stop_price - entry_price, 'variant': 'any_target_orig_stop', 'atr_at_fill': atr_val_at_fill}
            else: # SHORT
                if TARGET_MODE == 'atr' and not np.isnan(atr_target_price):
                    if bar['Low'] <= atr_target_price:
                        return {'entry_price': entry_price, 'exit_price': atr_target_price, 'win': True, 'pnl_price': entry_price - atr_target_price, 'variant': 'atr_target_atr_stop', 'atr_at_fill': atr_val_at_fill}
                if TARGET_MODE == 'prev_extreme':
                    if bar['Low'] <= prev_ext_target_price:
                        return {'entry_price': entry_price, 'exit_price': prev_ext_target_price, 'win': True, 'pnl_price': entry_price - prev_ext_target_price, 'variant': 'prev_ext_target_atr_stop', 'atr_at_fill': atr_val_at_fill}
                if not np.isnan(atr_stop_price) and bar['High'] >= atr_stop_price:
                    return {'entry_price': entry_price, 'exit_price': atr_stop_price, 'win': False, 'pnl_price': entry_price - atr_stop_price, 'variant': 'atr_target_atr_stop', 'atr_at_fill': atr_val_at_fill}
                if bar['High'] >= orig_stop_price:
                    return {'entry_price': entry_price, 'exit_price': orig_stop_price, 'win': False, 'pnl_price': entry_price - orig_stop_price, 'variant': 'any_target_orig_stop', 'atr_at_fill': atr_val_at_fill}
            # keep scanning
            continue

    # End of day reached
    if not entry_filled:
        return None
    # exit at close (if requested)
    close_price = float(intraday_df['Close'].iloc[-1])
    if intended_side == 'LONG':
        pnl_price_close = close_price - entry_price
    else:
        pnl_price_close = entry_price - close_price
    # we will return two variants: "atr_target_atr_stop" or "prev_ext_target_atr_stop" depending on TARGET_MODE,
    # but label as 'exit_at_close' for clarity
    return {'entry_price': entry_price, 'exit_price': close_price, 'win': False, 'pnl_price': pnl_price_close, 'variant': 'exit_at_close', 'atr_at_fill': atr_val_at_fill}

# run simulation over recent days
results = []
days = list(recent_days.index)
for day in tqdm(days, desc="Days"):
    row = recent_days.loc[day]
    intraday = fetch_intraday_day(TICKER, day)
    time.sleep(SLEEP_BETWEEN_REQUESTS)
    if intraday.empty:
        continue
    out = simulate_day(row, intraday)
    if out is None:
        continue
    # convert pnl price -> USD per contract and apply fees
    pnl_usd = out['pnl_price'] * GOLD_MULTIPLIER - SLIPPAGE - COMMISSION
    results.append({
        'date': day,
        'entry_price': out['entry_price'],
        'exit_price': out['exit_price'],
        'win': out['win'],
        'pnl_price': out['pnl_price'],
        'pnl_usd': pnl_usd,
        'variant': out.get('variant','unknown'),
        'atr_at_fill': out.get('atr_at_fill', np.nan)
    })

if len(results) == 0:
    raise SystemExit("No trades simulated (maybe intraday data unavailable).")

trades_df = pd.DataFrame(results).set_index('date').sort_index()

# Metrics per-variant
def compute_metrics(series):
    arr = np.array(series)
    n = len(arr)
    mean = np.mean(arr)
    std = np.std(arr, ddof=1) if n>1 else np.nan
    # trade-level Sharpe (per-trade): mean / std * sqrt(n). (Not a perfect annualization, but useful relative metric.)
    sharpe = (mean / std * np.sqrt(n)) if (n>1 and std>0) else np.nan
    # Sortino: use downside deviation
    downside = arr[arr < 0]
    dd = np.sqrt(np.mean(downside**2)) if len(downside)>0 else 0.0
    sortino = (mean / dd * np.sqrt(n)) if (n>0 and dd>0) else np.nan
    return {'n': n, 'mean': mean, 'std': std, 'sharpe_trade': sharpe, 'sortino_trade': sortino, 'total': arr.sum()}

summary_rows = []
for variant, g in trades_df.groupby('variant'):
    wins = g['win'].sum()
    n = len(g)
    win_rate = wins / n
    m = compute_metrics(g['pnl_usd'])
    avg_win = g[g['win']==True]['pnl_usd'].mean() if wins>0 else np.nan
    avg_loss = g[g['win']==False]['pnl_usd'].mean() if n-wins>0 else np.nan
    expectancy = m['mean']
    summary_rows.append({
        'variant': variant,
        'trades': n,
        'wins': int(wins),
        'win_rate': win_rate,
        'avg_pnl_usd': m['mean'],
        'avg_win_usd': avg_win,
        'avg_loss_usd': avg_loss,
        'expectancy_per_trade_usd': expectancy,
        'cumulative_usd': m['total'],
        'sharpe_trade': m['sharpe_trade'],
        'sortino_trade': m['sortino_trade']
    })

summary_df = pd.DataFrame(summary_rows).sort_values('cumulative_usd', ascending=False).reset_index(drop=True)

# Print results
pd.set_option('display.float_format', lambda x: f"{x:.2f}")
print("Detailed trades head (most recent 20):")
display(trades_df[['variant','entry_price','exit_price','win','pnl_price','pnl_usd','atr_at_fill']].iloc[::-1].head(20))

print("\nSummary by variant:")
display(summary_df)

# Save outputs
os.makedirs(OUTPUT_DIR, exist_ok=True)
trades_df.to_csv(os.path.join(OUTPUT_DIR, "gc_enhanced_trades.csv"))
summary_df.to_csv(os.path.join(OUTPUT_DIR, "gc_enhanced_summary.csv"), index=False)
print(f"\nSaved trades to: {os.path.join(OUTPUT_DIR, 'gc_enhanced_trades.csv')}")
print(f"Saved summary to: {os.path.join(OUTPUT_DIR, 'gc_enhanced_summary.csv')}")

# Plots: equity curve and PnL histogram
plt.figure(figsize=(11,5))
for variant in trades_df['variant'].unique():
    s = trades_df[trades_df['variant']==variant]['pnl_usd'].cumsum()
    if len(s)>0:
        plt.plot(s.index, s.values, label=variant)
plt.legend()
plt.title('Cumulative USD PnL by variant')
plt.xlabel('Date')
plt.ylabel('Cumulative PnL (USD)')
plt.grid(True)
plt.tight_layout()
plt.show()

plt.figure(figsize=(8,4))
plt.hist(trades_df['pnl_usd'], bins=30)
plt.title('Distribution of per-trade PnL (USD)')
plt.xlabel('PnL (USD)')
plt.ylabel('Count')
plt.grid(True)
plt.tight_layout()
plt.show()

# Quick stats print
total_trades = len(trades_df)
total_pnl = trades_df['pnl_usd'].sum()
print(f"Total trades: {total_trades}, Total PnL (USD): {total_pnl:.2f}")
print("End of enhanced backtest cell.")
# ===== End cell =====
```

```python
# Colab cell: Grid-search intraday strategy for GC=F, pick historically-positive config (if any)
!pip install --quiet yfinance pandas matplotlib tqdm numpy scipy

import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime, timedelta, timezone
import os, time
from tqdm import tqdm
from itertools import product

# ---------------- USER TWEAKABLE PARAMETERS ----------------
TICKER = "GC=F"
LOOKBACK_DAYS_INTRADAY = 60    # <= ~60 for yfinance 5m
DISTANCE_QUANTILES = [0.2, 0.3, 0.4]   # only trade days with dist_to_chosen <= q
RETRACE_FRACS = [0.5, 0.25, 0.0]
ATR_STOP_MULTS = [0.5, 0.75, 1.0]
ATR_TARGET_MULTS = [0.75, 1.0, 1.5]
EARLY_BARS_ALLOWED = [6, 12, None]    # None = allow entry anytime during day, else limit to first N 5-min bars
MIN_TRADES_REQUIRED = 8               # require at least this many trades to consider a config "valid"
ATR_PERIOD = 14
GOLD_MULTIPLIER = 100.0
STOP_BUFFER = 1.0
SLIPPAGE = 0.0
COMMISSION = 0.0
OUTPUT_DIR = "/mnt/data/gc_grid_search_results"
SLEEP_BETWEEN_REQUESTS = 0.15
# ----------------------------------------------------------

def compute_atr(df, period=14):
    high = df['High']; low = df['Low']; close = df['Close']
    tr1 = high - low
    tr2 = (high - close.shift(1)).abs()
    tr3 = (low - close.shift(1)).abs()
    tr = pd.concat([tr1, tr2, tr3], axis=1).max(axis=1)
    return tr.rolling(window=period, min_periods=1).mean()

# prepare dates
end_utc = datetime.now(timezone.utc)
start_utc = end_utc - timedelta(days=int(LOOKBACK_DAYS_INTRADAY * 1.3))
start_date = start_utc.strftime("%Y-%m-%d")
end_date = end_utc.strftime("%Y-%m-%d")
print("Fetching daily data from", start_date, "to", end_date)

# fetch daily data
daily = yf.download(TICKER, start=start_date, end=end_date, interval="1d", progress=True, auto_adjust=False)
if daily.empty:
    raise SystemExit("No daily data downloaded. Check ticker/dates or internet.")
if isinstance(daily.columns, pd.MultiIndex):
    daily.columns = [c[0] if isinstance(c, tuple) else c for c in daily.columns]
daily = daily[['Open','High','Low','Close','Volume']].copy()
daily.index = pd.to_datetime(daily.index).tz_localize(None)
daily['PDH'] = daily['High'].shift(1)
daily['PDL'] = daily['Low'].shift(1)
daily['Open_today'] = daily['Open']
daily = daily.dropna(subset=['PDH','PDL']).copy()
recent_days = daily.iloc[-LOOKBACK_DAYS_INTRADAY:].copy()
print(f"Prepared {len(recent_days)} recent days")

# helper: fetch intraday for a day (yfinance 5m)
def fetch_intraday_day(ticker, day_timestamp):
    start = pd.Timestamp(day_timestamp.date(), tz=timezone.utc)
    end = start + pd.Timedelta(days=1)
    try:
        df = yf.download(ticker, start=start.strftime("%Y-%m-%d"), end=end.strftime("%Y-%m-%d"),
                         interval="5m", progress=False, auto_adjust=False)
    except Exception as e:
        print(f"Intraday fetch error for {day_timestamp.date()}: {e}")
        return pd.DataFrame()
    if df.empty:
        return df
    if isinstance(df.columns, pd.MultiIndex):
        df.columns = [c[0] if isinstance(c, tuple) else c for c in df.columns]
    df = df[['Open','High','Low','Close','Volume']].copy()
    df.index = pd.to_datetime(df.index).tz_localize(None)
    return df

# realistic intraday simulation for a single day with given params
def simulate_day_intraday(day_row, intraday_df, retrace_frac, atr_stop_mult, atr_target_mult, early_bars_limit):
    """
    Return trade dict or None if entry didn't fill or ATR not available for ATR variant as needed.
    Entry: limit at open +/- retrace_frac * dist_to_chosen (0 => market/open)
    Stop: ATR_stop = ATR_at_fill * atr_stop_mult; original fallback: half_to_opposite*... not used here (we focus on ATR stops)
    Target: atr_target_mult * ATR_at_fill (or prev_extreme if chosen outside grid)
    Early_bars_limit: if not None, only accept entry if entry is filled within first N bars
    """
    # compute chosen target (closest of PDH/PDL)
    entry_open = float(day_row['Open_today'])
    pdh = float(day_row['PDH']); pdl = float(day_row['PDL'])
    if abs(pdh - entry_open) <= abs(entry_open - pdl):
        target_side = 'PDH'; target_price_prev = pdh; intended_side = 'LONG'
    else:
        target_side = 'PDL'; target_price_prev = pdl; intended_side = 'SHORT'
    dist_to_chosen = abs(target_price_prev - entry_open)
    # entry price
    if retrace_frac == 0.0:
        entry_price = entry_open
    else:
        entry_price = entry_open - retrace_frac * dist_to_chosen if intended_side=='LONG' else entry_open + retrace_frac * dist_to_chosen

    # compute ATR series on intraday bars
    atr_series = compute_atr(intraday_df, period=ATR_PERIOD)
    entry_filled = False
    entry_fill_time = None
    atr_at_fill = np.nan
    # iterate bars
    bars = intraday_df.iterrows()
    bar_index = 0
    for ts, bar in bars:
        bar_index += 1
        if not entry_filled:
            # limit fill conditions
            if intended_side == 'LONG':
                if bar['Low'] <= entry_price:
                    entry_filled = True
                    entry_fill_time = ts
                    atr_at_fill = atr_series.loc[:ts].iloc[-1] if len(atr_series.loc[:ts])>0 else np.nan
            else:
                if bar['High'] >= entry_price:
                    entry_filled = True
                    entry_fill_time = ts
                    atr_at_fill = atr_series.loc[:ts].iloc[-1] if len(atr_series.loc[:ts])>0 else np.nan
            if not entry_filled:
                continue
            # if early_bars_limit is set and bar_index > limit -> reject trade (entry too late)
            if (early_bars_limit is not None) and (bar_index > early_bars_limit):
                return None
            # need ATR to compute stops/targets; if nan, skip this day
            if np.isnan(atr_at_fill) or atr_at_fill <= 0:
                return None
            # compute atr stop & target
            atr_stop_dist = atr_at_fill * atr_stop_mult
            atr_stop_price = entry_price - atr_stop_dist if intended_side=='LONG' else entry_price + atr_stop_dist
            atr_target_dist = atr_at_fill * atr_target_mult
            atr_target_price = entry_price + atr_target_dist if intended_side=='LONG' else entry_price - atr_target_dist
            # now continue scanning for outcome (starting with the same bar)
        else:
            # after entry filled: check if target or stop hit on this bar
            if intended_side == 'LONG':
                if bar['High'] >= atr_target_price:
                    pnl_price = atr_target_price - entry_price
                    return {'win': True, 'pnl_price': pnl_price, 'entry_price': entry_price, 'exit_price': atr_target_price, 'atr_at_fill': atr_at_fill}
                if bar['Low'] <= atr_stop_price:
                    pnl_price = atr_stop_price - entry_price
                    return {'win': False, 'pnl_price': pnl_price, 'entry_price': entry_price, 'exit_price': atr_stop_price, 'atr_at_fill': atr_at_fill}
            else:
                if bar['Low'] <= atr_target_price:
                    pnl_price = entry_price - atr_target_price
                    return {'win': True, 'pnl_price': pnl_price, 'entry_price': entry_price, 'exit_price': atr_target_price, 'atr_at_fill': atr_at_fill}
                if bar['High'] >= atr_stop_price:
                    pnl_price = entry_price - atr_stop_price
                    return {'win': False, 'pnl_price': pnl_price, 'entry_price': entry_price, 'exit_price': atr_stop_price, 'atr_at_fill': atr_at_fill}
            continue

    # end of day reached
    if not entry_filled:
        return None
    # exit at close (conservative choice)
    close_price = float(intraday_df['Close'].iloc[-1])
    pnl_price = (close_price - entry_price) if intended_side=='LONG' else (entry_price - close_price)
    return {'win': False, 'pnl_price': pnl_price, 'entry_price': entry_price, 'exit_price': close_price, 'atr_at_fill': atr_at_fill}

# Grid search
configs = list(product(DISTANCE_QUANTILES, RETRACE_FRACS, ATR_STOP_MULTS, ATR_TARGET_MULTS, EARLY_BARS_ALLOWED))
print(f"Total configs to test: {len(configs)} (may take a minute)")

results = []
# cache intraday per day to avoid re-fetching repeatedly
intraday_cache = {}

days_list = list(recent_days.index)
for i, cfg in enumerate(configs):
    dist_q, retrace_frac, atr_stop_mult, atr_target_mult, early_bars = cfg
    # prepare candidate days filtered by distance quantile
    # compute dist_to_chosen for recent_days
    tmp = recent_days.copy()
    tmp['dist_to_PDH'] = (tmp['PDH'] - tmp['Open_today']).abs()
    tmp['dist_to_PDL'] = (tmp['Open_today'] - tmp['PDL']).abs()
    tmp['dist_to_chosen'] = np.where(tmp['dist_to_PDH'] <= tmp['dist_to_PDL'], tmp['dist_to_PDH'], tmp['dist_to_PDL'])
    cutoff = tmp['dist_to_chosen'].quantile(dist_q)
    candidate_days = tmp[tmp['dist_to_chosen'] <= cutoff].copy()
    trades = []
    # loop candidate days
    for day in candidate_days.index:
        # fetch intraday once per day
        if day not in intraday_cache:
            intraday_cache[day] = fetch_intraday_day(TICKER, day)
            time.sleep(SLEEP_BETWEEN_REQUESTS)
        intraday = intraday_cache[day]
        if intraday.empty:
            continue
        out = simulate_day_intraday(candidate_days.loc[day], intraday, retrace_frac, atr_stop_mult, atr_target_mult, early_bars)
        if out is None:
            continue
        pnl_usd = out['pnl_price'] * GOLD_MULTIPLIER - SLIPPAGE - COMMISSION
        trades.append({'date': day, 'pnl_usd': pnl_usd, 'win': out['win'], 'entry_price': out['entry_price'], 'exit_price': out['exit_price'], 'atr_at_fill': out.get('atr_at_fill', np.nan)})
    # aggregate results for this config
    n = len(trades)
    total_pnl = sum(t['pnl_usd'] for t in trades) if n>0 else 0.0
    wins = sum(1 for t in trades if t['win'])
    win_rate = wins/n if n>0 else 0.0
    results.append({
        'dist_q': dist_q,
        'retrace_frac': retrace_frac,
        'atr_stop_mult': atr_stop_mult,
        'atr_target_mult': atr_target_mult,
        'early_bars': early_bars if early_bars is not None else -1,
        'trades': n,
        'wins': wins,
        'win_rate': win_rate,
        'total_pnl_usd': total_pnl
    })

# build results DataFrame
res_df = pd.DataFrame(results)
# filter configs with minimum trades
valid = res_df[res_df['trades'] >= MIN_TRADES_REQUIRED].copy()
# among valid, find any positive total_pnl_usd configs
positive_configs = valid[valid['total_pnl_usd'] > 0].sort_values('total_pnl_usd', ascending=False)
os.makedirs(OUTPUT_DIR, exist_ok=True)
res_df.to_csv(os.path.join(OUTPUT_DIR, "grid_search_all_configs.csv"), index=False)
valid.to_csv(os.path.join(OUTPUT_DIR, "grid_search_valid_configs.csv"), index=False)
positive_configs.to_csv(os.path.join(OUTPUT_DIR, "grid_search_positive_configs.csv"), index=False)

print("\nGrid search complete. Summary:")
print("Total configs tested:", len(configs))
print("Configs with >= min trades:", len(valid))
print("Configs with positive cumulative USD PnL and >= min trades:", len(positive_configs))

if len(positive_configs) > 0:
    best = positive_configs.iloc[0]
    print("\nBest positive config found (historical sample):")
    display(best)
    # Reconstruct trades for the best config and save details
    best_cfg = (best['dist_q'], best['retrace_frac'], best['atr_stop_mult'], best['atr_target_mult'], (None if best['early_bars']==-1 else int(best['early_bars'])))
    print("Rebuilding detailed trades for best config:", best_cfg)
    # recompute candidate days and trades
    dist_q, retrace_frac, atr_stop_mult, atr_target_mult, early_bars = best_cfg
    tmp = recent_days.copy()
    tmp['dist_to_PDH'] = (tmp['PDH'] - tmp['Open_today']).abs()
    tmp['dist_to_PDL'] = (tmp['Open_today'] - tmp['PDL']).abs()
    tmp['dist_to_chosen'] = np.where(tmp['dist_to_PDH'] <= tmp['dist_to_PDL'], tmp['dist_to_PDH'], tmp['dist_to_PDL'])
    cutoff = tmp['dist_to_chosen'].quantile(dist_q)
    candidate_days = tmp[tmp['dist_to_chosen'] <= cutoff].copy()
    detailed = []
    for day in candidate_days.index:
        intraday = intraday_cache.get(day, fetch_intraday_day(TICKER, day))
        if intraday.empty:
            continue
        out = simulate_day_intraday(candidate_days.loc[day], intraday, retrace_frac, atr_stop_mult, atr_target_mult, early_bars)
        if out is None: continue
        pnl_usd = out['pnl_price'] * GOLD_MULTIPLIER - SLIPPAGE - COMMISSION
        detailed.append({'date': day, 'entry_price': out['entry_price'], 'exit_price': out['exit_price'], 'win': out['win'], 'pnl_usd': pnl_usd, 'atr_at_fill': out.get('atr_at_fill',np.nan)})
    detailed_df = pd.DataFrame(detailed).set_index('date').sort_index()
    detailed_df.to_csv(os.path.join(OUTPUT_DIR, "best_config_trades.csv"))
    print(f"Saved detailed trades for best config to {os.path.join(OUTPUT_DIR, 'best_config_trades.csv')}")
    display(detailed_df)
else:
    print("\nNo config produced positive cumulative PnL on this recent sample with the chosen grid and min trades requirement.")
    # show top 5 configs by total_pnl_usd regardless of sign (for inspection)
    top5 = valid.sort_values('total_pnl_usd', ascending=False).head(10)
    print("Top configs (even if negative) — you can inspect them and expand grid:")
    display(top5)
    top5.to_csv(os.path.join(OUTPUT_DIR, "top5_configs_inspection.csv"))
    print(f"Saved inspection table to {os.path.join(OUTPUT_DIR, 'top5_configs_inspection.csv')}")

print("\nAll outputs saved to:", OUTPUT_DIR)
print("Warning: this selects a historically-positive config on the recent sample if one exists. It is NOT a guarantee of live profitability.")
```

```python
# Colab cell: Grid search + walk-forward validation to pick historically-positive config (intraday GC=F 5m)
# NOTE: Yahoo 5m is limited to recent ~60 days. This script tries to find a config that shows positive
# OOS (test) PnL across multiple rolling folds to reduce overfitting risk.
!pip install --quiet yfinance pandas numpy matplotlib tqdm

import yfinance as yf
import pandas as pd
import numpy as np
from datetime import datetime, timedelta, timezone
import time, os
from itertools import product
from tqdm import tqdm

# ---------------- USER PARAMETERS ----------------
TICKER = "GC=F"
LOOKBACK_DAYS_INTRADAY = 60    # <= ~60 for yfinance 5m
DISTANCE_QUANTILES = [0.2, 0.3, 0.4]   # day filter
RETRACE_FRACS = [0.5, 0.25, 0.0]
ATR_STOP_MULTS = [0.5, 0.75, 1.0]
ATR_TARGET_MULTS = [0.75, 1.0, 1.5]
EARLY_BARS_ALLOWED = [6, 12, None]
ATR_PERIOD = 14
RETRACE_FRAC_CHOICE = RETRACE_FRACS  # kept for clarity
GOLD_MULTIPLIER = 100.0
STOP_BUFFER = 1.0
SLIPPAGE = 0.0
COMMISSION = 0.0
SLEEP_BETWEEN_REQUESTS = 0.12
OUTPUT_DIR = "/mnt/data/gc_grid_wf_results"
# Walk-forward parameters
WF_N_FOLDS = 3           # number of rolling folds: train on fold_i, test on fold_i+1 for i in 0..N-2
WF_TRAIN_DAYS = 20       # days used for training each fold
WF_TEST_DAYS = 10        # days used for testing each fold
# --------------------------------------------------

os.makedirs(OUTPUT_DIR, exist_ok=True)

def compute_atr(df, period=14):
    high = df['High']; low = df['Low']; close = df['Close']
    tr1 = high - low
    tr2 = (high - close.shift(1)).abs()
    tr3 = (low - close.shift(1)).abs()
    tr = pd.concat([tr1, tr2, tr3], axis=1).max(axis=1)
    return tr.rolling(window=period, min_periods=1).mean()

# 1) download daily to build days list
end_utc = datetime.now(timezone.utc)
start_utc = end_utc - timedelta(days=int(LOOKBACK_DAYS_INTRADAY * 1.3))
start_date = start_utc.strftime("%Y-%m-%d")
end_date = end_utc.strftime("%Y-%m-%d")
print("Fetching daily data:", start_date, "->", end_date)
daily = yf.download(TICKER, start=start_date, end=end_date, interval="1d", progress=True, auto_adjust=False)
if daily.empty:
    raise SystemExit("No daily data. Check ticker or internet.")
if isinstance(daily.columns, pd.MultiIndex):
    daily.columns = [c[0] if isinstance(c, tuple) else c for c in daily.columns]
daily = daily[['Open','High','Low','Close','Volume']].copy()
daily.index = pd.to_datetime(daily.index).tz_localize(None)
daily['PDH'] = daily['High'].shift(1)
daily['PDL'] = daily['Low'].shift(1)
daily['Open_today'] = daily['Open']
daily = daily.dropna(subset=['PDH','PDL']).copy()
recent_days = daily.iloc[-LOOKBACK_DAYS_INTRADAY:].copy()
print("Prepared", len(recent_days), "recent days for intraday work.")

# helper to fetch intraday 5m for a day (yfinance)
def fetch_intraday_day(ticker, day_timestamp):
    start = pd.Timestamp(day_timestamp.date(), tz=timezone.utc)
    end = start + pd.Timedelta(days=1)
    try:
        df = yf.download(ticker, start=start.strftime("%Y-%m-%d"), end=end.strftime("%Y-%m-%d"),
                         interval="5m", progress=False, auto_adjust=False)
    except Exception as e:
        # return empty df on error
        print("Intraday fetch error for", day_timestamp.date(), "->", e)
        return pd.DataFrame()
    if df.empty:
        return df
    if isinstance(df.columns, pd.MultiIndex):
        df.columns = [c[0] if isinstance(c, tuple) else c for c in df.columns]
    df = df[['Open','High','Low','Close','Volume']].copy()
    df.index = pd.to_datetime(df.index).tz_localize(None)
    return df

# realistic intraday simulation for a day (ATR rules) - returns None or trade dict
def simulate_day_intraday(day_row, intraday_df, retrace_frac, atr_stop_mult, atr_target_mult, early_bars):
    entry_open = float(day_row['Open_today'])
    pdh = float(day_row['PDH']); pdl = float(day_row['PDL'])
    if abs(pdh - entry_open) <= abs(entry_open - pdl):
        target_side = 'PDH'; intended_side = 'LONG'
    else:
        target_side = 'PDL'; intended_side = 'SHORT'
    dist_to_chosen = abs((pdh if target_side=='PDH' else pdl) - entry_open)
    # entry price (limit)
    if retrace_frac == 0.0:
        entry_price = entry_open
    else:
        entry_price = entry_open - retrace_frac * dist_to_chosen if intended_side=='LONG' else entry_open + retrace_frac * dist_to_chosen
    atr_series = compute_atr(intraday_df, period=ATR_PERIOD)
    filled = False
    fill_time = None
    atr_at_fill = np.nan
    bar_index = 0
    for ts, bar in intraday_df.iterrows():
        bar_index += 1
        if not filled:
            if intended_side=='LONG':
                if bar['Low'] <= entry_price:
                    filled = True
                    fill_time = ts
                    atr_at_fill = atr_series.loc[:ts].iloc[-1] if len(atr_series.loc[:ts])>0 else np.nan
            else:
                if bar['High'] >= entry_price:
                    filled = True
                    fill_time = ts
                    atr_at_fill = atr_series.loc[:ts].iloc[-1] if len(atr_series.loc[:ts])>0 else np.nan
            if not filled:
                continue
            # enforce early_bars limit
            if (early_bars is not None) and (bar_index > early_bars):
                return None
            if np.isnan(atr_at_fill) or atr_at_fill <= 0:
                return None
            atr_stop_dist = atr_at_fill * atr_stop_mult
            atr_target_dist = atr_at_fill * atr_target_mult
            atr_stop_price = entry_price - atr_stop_dist if intended_side=='LONG' else entry_price + atr_stop_dist
            atr_target_price = entry_price + atr_target_dist if intended_side=='LONG' else entry_price - atr_target_dist
            continue
        else:
            if intended_side=='LONG':
                if bar['High'] >= atr_target_price:
                    return {'win': True, 'pnl_price': atr_target_price - entry_price}
                if bar['Low'] <= atr_stop_price:
                    return {'win': False, 'pnl_price': atr_stop_price - entry_price}
            else:
                if bar['Low'] <= atr_target_price:
                    return {'win': True, 'pnl_price': entry_price - atr_target_price}
                if bar['High'] >= atr_stop_price:
                    return {'win': False, 'pnl_price': entry_price - atr_stop_price}
            continue
    # end of day
    if not filled:
        return None
    close_price = float(intraday_df['Close'].iloc[-1])
    pnl_price = (close_price - entry_price) if intended_side=='LONG' else (entry_price - close_price)
    return {'win': False, 'pnl_price': pnl_price}

# prepare list of configs
configs = list(product(DISTANCE_QUANTILES, RETRACE_FRACS, ATR_STOP_MULTS, ATR_TARGET_MULTS, EARLY_BARS_ALLOWED))
print("Configs to test:", len(configs))

# cache intraday per day
intraday_cache = {}
days = list(recent_days.index)

# Build walk-forward folds (chronological)
# We'll build folds of consecutive days: train window -> test window, then advance by test window
# e.g., days[0:20] train, days[20:30] test -> days[10:30] train? To keep it simple: non-overlapping sequential folds
sorted_days = sorted(days)
folds = []
i = 0
while i + WF_TRAIN_DAYS + WF_TEST_DAYS <= len(sorted_days):
    train_days = sorted_days[i:i+WF_TRAIN_DAYS]
    test_days = sorted_days[i+WF_TRAIN_DAYS:i+WF_TRAIN_DAYS+WF_TEST_DAYS]
    folds.append({'train': train_days, 'test': test_days})
    i += WF_TEST_DAYS  # roll forward by test window
print("Walk-forward folds:", len(folds))
if len(folds) < 1:
    print("Not enough days for walk-forward. Consider reducing WF_TRAIN_DAYS/WF_TEST_DAYS or increasing LOOKBACK_DAYS_INTRADAY.")

# Evaluate each config with walk-forward: for each fold, pick days in train that pass distance cutoff and compute pnl,
# then compute PnL on test days (using same params)
grid_results = []
for cfg in tqdm(configs, desc="Grid"):
    dist_q, retrace_frac, atr_stop_mult, atr_target_mult, early_bars = cfg
    test_fold_pnls = []  # store cumulative pnl (USD) for each test fold
    valid_fold_count = 0
    for f in folds:
        # compute distance cutoff from train days
        train_df = recent_days.loc[f['train']].copy()
        train_df['dist_to_PDH'] = (train_df['PDH'] - train_df['Open_today']).abs()
        train_df['dist_to_PDL'] = (train_df['Open_today'] - train_df['PDL']).abs()
        train_df['dist_to_chosen'] = np.where(train_df['dist_to_PDH'] <= train_df['dist_to_PDL'], train_df['dist_to_PDH'], train_df['dist_to_PDL'])
        cutoff = train_df['dist_to_chosen'].quantile(dist_q)
        # candidate test days meeting cutoff (apply cutoff derived from train)
        test_df = recent_days.loc[f['test']].copy()
        test_df['dist_to_PDH'] = (test_df['PDH'] - test_df['Open_today']).abs()
        test_df['dist_to_PDL'] = (test_df['Open_today'] - test_df['PDL']).abs()
        test_df['dist_to_chosen'] = np.where(test_df['dist_to_PDH'] <= test_df['dist_to_PDL'], test_df['dist_to_PDH'], test_df['dist_to_PDL'])
        candidate_days = test_df[test_df['dist_to_chosen'] <= cutoff].copy()
        if candidate_days.empty:
            # no candidate days in test -> treat as zero pnl for fold (still counts)
            test_fold_pnls.append(0.0)
            valid_fold_count += 1
            continue
        # simulate each candidate day and sum usd pnl
        fold_pnl = 0.0
        for day in candidate_days.index:
            if day not in intraday_cache:
                intraday_cache[day] = fetch_intraday_day(TICKER, day)
                time.sleep(SLEEP_BETWEEN_REQUESTS)
            intraday = intraday_cache[day]
            if intraday.empty:
                continue
            out = simulate_day_intraday(candidate_days.loc[day], intraday, retrace_frac, atr_stop_mult, atr_target_mult, early_bars)
            if out is None:
                continue
            pnl_usd = out['pnl_price'] * GOLD_MULTIPLIER - SLIPPAGE - COMMISSION
            fold_pnl += pnl_usd
        test_fold_pnls.append(fold_pnl)
        valid_fold_count += 1
    # compute mean test fold pnl (this is the walk-forward generalization metric)
    if valid_fold_count == 0:
        mean_test_pnl = np.nan
    else:
        mean_test_pnl = np.mean(test_fold_pnls)
    grid_results.append({
        'dist_q': dist_q,
        'retrace_frac': retrace_frac,
        'atr_stop_mult': atr_stop_mult,
        'atr_target_mult': atr_target_mult,
        'early_bars': (early_bars if early_bars is not None else -1),
        'mean_test_pnl': mean_test_pnl,
        'sum_test_pnl': np.sum(test_fold_pnls),
        'folds_counted': valid_fold_count
    })

grid_df = pd.DataFrame(grid_results).sort_values('mean_test_pnl', ascending=False).reset_index(drop=True)
grid_df.to_csv(os.path.join(OUTPUT_DIR, "grid_walkforward_summary.csv"), index=False)
print("Grid walk-forward summary saved to:", os.path.join(OUTPUT_DIR, "grid_walkforward_summary.csv"))

# Pick top candidate(s)
top = grid_df.head(5).copy()
print("\nTop 5 configs by walk-forward mean test PnL:")
display(top)

# If the very top config has positive mean_test_pnl, rebuild detailed trades across all test folds and save
best = grid_df.iloc[0]
print("\nBest candidate (by mean test PnL):")
display(best)
if not np.isnan(best['mean_test_pnl']) and best['mean_test_pnl'] > 0:
    # reconstruct detailed test trades for each fold (save CSV)
    dist_q = best['dist_q']; retrace_frac = best['retrace_frac']; atr_stop_mult = best['atr_stop_mult']
    atr_target_mult = best['atr_target_mult']; early_bars = None if best['early_bars']==-1 else int(best['early_bars'])
    detailed_records = []
    for f in folds:
        # compute cutoff from train
        train_df = recent_days.loc[f['train']].copy()
        train_df['dist_to_PDH'] = (train_df['PDH'] - train_df['Open_today']).abs()
        train_df['dist_to_PDL'] = (train_df['Open_today'] - train_df['PDL']).abs()
        train_df['dist_to_chosen'] = np.where(train_df['dist_to_PDH'] <= train_df['dist_to_PDL'], train_df['dist_to_PDH'], train_df['dist_to_PDL'])
        cutoff = train_df['dist_to_chosen'].quantile(dist_q)
        test_df = recent_days.loc[f['test']].copy()
        test_df['dist_to_PDH'] = (test_df['PDH'] - test_df['Open_today']).abs()
        test_df['dist_to_PDL'] = (test_df['Open_today'] - test_df['PDL']).abs()
        test_df['dist_to_chosen'] = np.where(test_df['dist_to_PDH'] <= test_df['dist_to_PDL'], test_df['dist_to_PDH'], test_df['dist_to_PDL'])
        candidate_days = test_df[test_df['dist_to_chosen'] <= cutoff].copy()
        for day in candidate_days.index:
            if day not in intraday_cache:
                intraday_cache[day] = fetch_intraday_day(TICKER, day)
                time.sleep(SLEEP_BETWEEN_REQUESTS)
            intraday = intraday_cache[day]
            if intraday.empty:
                continue
            out = simulate_day_intraday(candidate_days.loc[day], intraday, retrace_frac, atr_stop_mult, atr_target_mult, early_bars)
            if out is None:
                continue
            pnl_usd = out['pnl_price'] * GOLD_MULTIPLIER - SLIPPAGE - COMMISSION
            detailed_records.append({'date': day, 'pnl_usd': pnl_usd, 'win': out['win'], 'entry_price': out.get('entry_price', np.nan), 'exit_price': out.get('exit_price', np.nan)})
    if detailed_records:
        detailed_df = pd.DataFrame(detailed_records).set_index('date').sort_index()
        detailed_df.to_csv(os.path.join(OUTPUT_DIR, "best_candidate_test_trades.csv"))
        print("Saved best candidate test trades to:", os.path.join(OUTPUT_DIR, "best_candidate_test_trades.csv"))
        display(detailed_df)
    else:
        print("Best candidate produced no detailed trades in the test folds (maybe intraday missing).")
else:
    print("\nNo candidate had positive mean walk-forward test PnL. Top configs were saved to CSV for inspection.")
    # save top 10 for inspection
    grid_df.head(10).to_csv(os.path.join(OUTPUT_DIR, "top10_configs_for_inspection.csv"))
    print("Saved top10 configs to:", os.path.join(OUTPUT_DIR, "top10_configs_for_inspection.csv"))

print("\nAll files saved under:", OUTPUT_DIR)
print("IMPORTANT: If no positive candidate is found, the correct next step is more/better data (broker CSV) or rethinking the signal.")

```

```python
# === Re-run grid search with trade count filter ===

import pandas as pd
from tqdm import tqdm
import numpy as np
import yfinance as yf

symbol = "GC=F"
interval = "5m"

# 60 days of 5m data (max allowed by Yahoo)
data = yf.download(symbol, period="60d", interval=interval)
data.index = data.index.tz_localize(None)

print(f"Fetched {len(data)} bars of {symbol} ({interval})")

# Add basic intraday OHLC structure
data['date'] = data.index.date
daily_groups = data.groupby('date')

# Function to simulate “mean reversion after trend” setup
def backtest_day(df, retrace_frac=0.5, atr_stop_mult=1.0, atr_target_mult=1.5):
    if len(df) < 20:
        return None

    o = df['Open'].iloc[0]
    h = df['High'].max()
    l = df['Low'].min()
    atr = (h - l) / 14  # simple rough ATR estimate

    # Assume previous day bias — fade direction after large move
    bias = "short" if h - o > o - l else "long"
    retrace_price = o + (h - o)*retrace_frac if bias=="short" else o - (o - l)*retrace_frac
    stop = retrace_price + atr*atr_stop_mult if bias=="short" else retrace_price - atr*atr_stop_mult
    target = retrace_price - atr*atr_target_mult if bias=="short" else retrace_price + atr*atr_target_mult

    hit_target = ((df['Low'] <= target) if bias=="short" else (df['High'] >= target)).any()
    hit_stop   = ((df['High'] >= stop) if bias=="short" else (df['Low'] <= stop)).any()

    pnl = 0
    if hit_target and not hit_stop:
        pnl = abs(target - retrace_price)
    elif hit_stop and not hit_target:
        pnl = -abs(stop - retrace_price)
    elif hit_stop and hit_target:
        # whichever came first
        t_hit = df.index[(df['Low'] <= target) if bias=="short" else (df['High'] >= target)][0]
        s_hit = df.index[(df['High'] >= stop) if bias=="short" else (df['Low'] <= stop)][0]
        pnl = abs(target - retrace_price) if t_hit < s_hit else -abs(stop - retrace_price)

    return pnl * (200)  # gold contract multiplier approx

# Grid search
retrace_fracs = [0.25, 0.5, 0.75]
stop_mults = [0.5, 1.0, 1.5]
target_mults = [1.0, 1.5, 2.0]
results = []

for retr in retrace_fracs:
    for stop in stop_mults:
        for tgt in target_mults:
            pnls = []
            for d, df_day in daily_groups:
                r = backtest_day(df_day, retr, stop, tgt)
                if r is not None:
                    pnls.append(r)
            if len(pnls) >= 5:  # <- minimum trades filter
                results.append({
                    'retrace_frac': retr,
                    'atr_stop_mult': stop,
                    'atr_target_mult': tgt,
                    'trades': len(pnls),
                    'mean_pnl': np.mean(pnls),
                    'total_pnl': np.sum(pnls)
                })

summary = pd.DataFrame(results)
if not summary.empty:
    print("Top configs by total PnL:")
    display(summary.sort_values('total_pnl', ascending=False).head(10))
else:
    print("No configs met the minimum trade requirement.")
```

```python
# Fixed and robust intraday daily-group grid test (no pandas truth-value errors)
# Paste into Colab and run after previous yfinance fetch (or run standalone; it will fetch itself).
!pip install --quiet yfinance pandas numpy tqdm

import yfinance as yf
import pandas as pd
import numpy as np
from tqdm import tqdm
from datetime import datetime, timezone
import time

# ---------- parameters ----------
symbol = "GC=F"
interval = "5m"
period = "60d"     # last ~60 days (yfinance limit)
min_trades = 5     # require at least this many trades per config to consider it
retrace_fracs = [0.25, 0.5, 0.75]
stop_mults = [0.5, 1.0, 1.5]
target_mults = [1.0, 1.5, 2.0]
contract_mult = 100.0   # $ per 1 price unit for GC
# --------------------------------

print("Downloading intraday bars (may take a moment)...")
bars = yf.download(symbol, period=period, interval=interval, progress=True, auto_adjust=False)
if bars.empty:
    raise SystemExit("No intraday data downloaded. Check ticker/period/interval.")
bars.index = bars.index.tz_localize(None)  # remove tz info if present
print(f"Got {len(bars)} bars")

# group into calendar days
bars['cal_date'] = bars.index.date
grouped = bars.groupby('cal_date')
print("Days available:", len(grouped))

# helper: conservative ATR estimate for the day using high-low range across intraday bars
def day_atr_estimate(df_day):
    # Use True Range across the day's bars (simple)
    highs = df_day['High'].max()
    lows = df_day['Low'].min()
    # If close previous available, could use that; here use high-low range as rough proxy
    tr = highs - lows
    # smoothing by 14 -> but we only have intraday day-level, so return fraction
    # We'll return tr/14 as a simple ATR estimate for this lightweight test
    return tr / 14.0 if tr > 0 else np.nan

# simulate trade for a single day DataFrame (intraday bars)
def backtest_day(df_day, retrace_frac=0.5, atr_stop_mult=1.0, atr_target_mult=1.5):
    if df_day.empty:
        return None
    # construct daily-like values from intraday (open of day, max high, min low)
    day_open = df_day['Open'].iloc[0]
    day_high = df_day['High'].max()
    day_low = df_day['Low'].min()

    # simple bias: if high - open > open - low => bias is up (we'll fade -> SHORT); else bias down -> LONG
    up_move = day_high - day_open
    down_move = day_open - day_low
    bias_short = up_move > down_move  # True => fade by shorting
    bias_long = not bias_short

    # entry (limit) defined as retrace_frac of the day's move toward extreme
    if bias_short:
        # market moved up; entry = open + retrace_frac * up_move (i.e. short a pullback)
        entry_price = day_open + retrace_frac * up_move
    else:
        # market moved down; entry = open - retrace_frac * down_move (long on pullback)
        entry_price = day_open - retrace_frac * down_move

    # ATR (rough) and stop/target distances
    atr = day_atr_estimate(df_day)
    if np.isnan(atr) or atr == 0:
        return None
    stop_dist = atr * atr_stop_mult
    target_dist = atr * atr_target_mult

    if bias_short:
        stop_price = entry_price + stop_dist
        target_price = entry_price - target_dist
        # find bars where target/stop touched
        target_hits_idx = df_day.index[df_day['Low'] <= target_price].tolist()
        stop_hits_idx   = df_day.index[df_day['High'] >= stop_price].tolist()
    else:
        stop_price = entry_price - stop_dist
        target_price = entry_price + target_dist
        target_hits_idx = df_day.index[df_day['High'] >= target_price].tolist()
        stop_hits_idx   = df_day.index[df_day['Low'] <= stop_price].tolist()

    # no hits -> exit at close
    if not target_hits_idx and not stop_hits_idx:
        exit_price = df_day['Close'].iloc[-1]
        pnl_price = (entry_price - exit_price) if bias_short else (exit_price - entry_price)
        return pnl_price * contract_mult

    # if both happened, need to see which timestamp came first
    if target_hits_idx and not stop_hits_idx:
        # first target time
        t_time = target_hits_idx[0]
        pnl_price = (entry_price - target_price) if bias_short else (target_price - entry_price)
        return pnl_price * contract_mult
    if stop_hits_idx and not target_hits_idx:
        s_time = stop_hits_idx[0]
        pnl_price = (entry_price - stop_price) if bias_short else (stop_price - entry_price)
        return pnl_price * contract_mult

    # both lists non-empty -> compare first timestamps
    t_time = target_hits_idx[0]
    s_time = stop_hits_idx[0]
    if t_time < s_time:
        pnl_price = (entry_price - target_price) if bias_short else (target_price - entry_price)
    else:
        pnl_price = (entry_price - stop_price) if bias_short else (stop_price - entry_price)
    return pnl_price * contract_mult

# Run grid and require minimum trades
results = []
for retr in retrace_fracs:
    for stop_mult in stop_mults:
        for tgt_mult in target_mults:
            pnls = []
            days_with_trade = 0
            for day, df_day in grouped:
                pnl = backtest_day(df_day, retrace_frac=retr, atr_stop_mult=stop_mult, atr_target_mult=tgt_mult)
                if pnl is not None:
                    pnls.append(pnl)
                    days_with_trade += 1
            if days_with_trade >= min_trades:
                total = float(np.sum(pnls))
                mean = float(np.mean(pnls)) if pnls else 0.0
                results.append({
                    'retrace_frac': retr,
                    'atr_stop_mult': stop_mult,
                    'atr_target_mult': tgt_mult,
                    'trades': days_with_trade,
                    'mean_pnl': mean,
                    'total_pnl': total
                })

res_df = pd.DataFrame(results)
if res_df.empty:
    print("No configs met the minimum-trades requirement (min_trades =", min_trades, ").")
else:
    res_df = res_df.sort_values('total_pnl', ascending=False).reset_index(drop=True)
    print("Top configs (by total PnL) that have >= {} trades:".format(min_trades))
    display(res_df.head(10))
    # save
    res_df.to_csv("/mnt/data/grid_fixed_results.csv", index=False)
    print("Saved /mnt/data/grid_fixed_results.csv")

# quick diagnostic: how many days produced any trade under at least one config?
day_trade_count = {}
for day, df_day in grouped:
    any_trade = False
    for retr in retrace_fracs:
        for stop_mult in stop_mults:
            for tgt_mult in target_mults:
                if backtest_day(df_day, retr, stop_mult, tgt_mult) is not None:
                    any_trade = True
                    break
            if any_trade: break
        if any_trade: break
    day_trade_count[day] = any_trade

n_days_with_any_trade = sum(1 for v in day_trade_count.values() if v)
print("Days that would produce at least one trade under any config:", n_days_with_any_trade, "out of", len(day_trade_count))
```

```python
# Robust intraday grid search (fixed ambiguous-series issues + diagnostics)
!pip install --quiet yfinance pandas numpy tqdm

import yfinance as yf
import pandas as pd
import numpy as np
from tqdm import tqdm
from datetime import datetime
import time, os

# ---------- parameters ----------
symbol = "GC=F"
interval = "5m"
period = "60d"
min_trades = 3   # lowered to 3 so we can inspect candidates (change back to 5 if needed)
retrace_fracs = [0.25, 0.5, 0.75]
stop_mults = [0.5, 1.0, 1.5]
target_mults = [1.0, 1.5, 2.0]
contract_mult = 100.0
OUTPUT_CSV = "/mnt/data/grid_fixed_results_final.csv"
# --------------------------------

print("Downloading intraday bars (may take a moment)...")
bars = yf.download(symbol, period=period, interval=interval, progress=True, auto_adjust=False)
if bars.empty:
    raise SystemExit("No intraday data downloaded. Check ticker/period/interval.")
bars.index = bars.index.tz_localize(None)
print(f"Got {len(bars)} bars")

# group into calendar days
bars['cal_date'] = bars.index.date
grouped = bars.groupby('cal_date')
print("Days available:", len(grouped))

def day_atr_estimate(df_day):
    """Return scalar ATR estimate for the day (safe)."""
    if df_day.empty:
        return np.nan
    highs = df_day['High'].max()
    lows = df_day['Low'].min()
    tr = float(highs - lows)
    return float(tr / 14.0) if tr > 0 else float(np.nan)

def first_hit_time(mask_series):
    """Return first index (timestamp) where mask True, or None."""
    if mask_series.sum() == 0:
        return None
    # use idxmax on boolean-as-int to get first True (works because False->0, True->1)
    # but must ensure mask_series is boolean Series
    s_int = mask_series.astype(int)
    first_pos = int(np.argmax(s_int.values))  # position of first True
    return mask_series.index[first_pos]

def simulate_day(df_day, retrace_frac, atr_stop_mult, atr_target_mult):
    """
    Returns: float pnl_usd or None if no trade executed.
    Deterministic, uses explicit boolean masks and .any()/.sum().
    """
    if df_day.empty:
        return None

    day_open = float(df_day['Open'].iloc[0])
    day_high = float(df_day['High'].max())
    day_low = float(df_day['Low'].min())

    up_move = day_high - day_open
    down_move = day_open - day_low

    # bias: if up_move larger -> fade up (SHORT), else LONG
    bias_short = (up_move > down_move)
    bias_long = not bias_short

    # entry
    if bias_short:
        entry_price = day_open + float(retrace_frac) * up_move
    else:
        entry_price = day_open - float(retrace_frac) * down_move

    atr = day_atr_estimate(df_day)
    if np.isnan(atr) or atr <= 0:
        return None

    stop_dist = atr * float(atr_stop_mult)
    target_dist = atr * float(atr_target_mult)

    if bias_short:
        stop_price = entry_price + stop_dist
        target_price = entry_price - target_dist
        # explicit boolean masks
        target_mask = df_day['Low'].le(float(target_price))
        stop_mask = df_day['High'].ge(float(stop_price))
    else:
        stop_price = entry_price - stop_dist
        target_price = entry_price + target_dist
        target_mask = df_day['High'].ge(float(target_price))
        stop_mask = df_day['Low'].le(float(stop_price))

    # no hits -> exit at close
    if (not target_mask.any()) and (not stop_mask.any()):
        exit_price = float(df_day['Close'].iloc[-1])
        pnl_price = (entry_price - exit_price) if bias_short else (exit_price - entry_price)
        return float(pnl_price * contract_mult)

    # determine which hit first when both occur, else handle single hit
    if target_mask.any() and not stop_mask.any():
        # target hits exist; choose first occurrence
        t_time = first_hit_time(target_mask)
        pnl_price = (entry_price - float(target_price)) if bias_short else (float(target_price) - entry_price)
        return float(pnl_price * contract_mult)
    if stop_mask.any() and not target_mask.any():
        s_time = first_hit_time(stop_mask)
        pnl_price = (entry_price - float(stop_price)) if bias_short else (float(stop_price) - entry_price)
        return float(pnl_price * contract_mult)
    # both true: compare first times
    t_time = first_hit_time(target_mask)
    s_time = first_hit_time(stop_mask)
    # if either is None (shouldn't be), handle gracefully
    if (t_time is None) and (s_time is None):
        return None
    if t_time is None:
        pnl_price = (entry_price - float(stop_price)) if bias_short else (float(stop_price) - entry_price)
        return float(pnl_price * contract_mult)
    if s_time is None:
        pnl_price = (entry_price - float(target_price)) if bias_short else (float(target_price) - entry_price)
        return float(pnl_price * contract_mult)
    # compare timestamps
    if pd.to_datetime(t_time) < pd.to_datetime(s_time):
        pnl_price = (entry_price - float(target_price)) if bias_short else (float(target_price) - entry_price)
    else:
        pnl_price = (entry_price - float(stop_price)) if bias_short else (float(stop_price) - entry_price)
    return float(pnl_price * contract_mult)

# Run grid, collect only configs with >= min_trades
results = []
days_with_any_trade = set()

for retr in retrace_fracs:
    for stop_mult in stop_mults:
        for tgt_mult in target_mults:
            pnls = []
            for day, df_day in grouped:
                try:
                    res = simulate_day(df_day, retr, stop_mult, tgt_mult)
                except Exception as e:
                    # log and continue (but do not crash)
                    print("Sim error for day", day, "cfg", (retr, stop_mult, tgt_mult), ":", e)
                    res = None
                if res is not None:
                    pnls.append(res)
                    days_with_any_trade.add(day)
            if len(pnls) >= min_trades:
                results.append({
                    'retrace_frac': retr,
                    'atr_stop_mult': stop_mult,
                    'atr_target_mult': tgt_mult,
                    'trades': len(pnls),
                    'mean_pnl': float(np.mean(pnls)),
                    'total_pnl': float(np.sum(pnls))
                })

res_df = pd.DataFrame(results)
if res_df.empty:
    print(f"No configs met the minimum-trades requirement (min_trades = {min_trades}).")
else:
    res_df = res_df.sort_values('total_pnl', ascending=False).reset_index(drop=True)
    print("Top configs (by total PnL) that have >= {} trades:".format(min_trades))
    display(res_df.head(10))
    res_df.to_csv(OUTPUT_CSV, index=False)
    print("Saved", OUTPUT_CSV)

# diagnostic: how many days would produce at least one trade under any config?
print("Days that would produce at least one trade under any tested config:", len(days_with_any_trade), "out of", len(grouped))
if len(days_with_any_trade) > 0:
    sampled = sorted(list(days_with_any_trade))[:20]
    print("Sample days with any trade (first 20):", sampled)
```

```python
# Robust grid run with debug logging for days that throw exceptions
!pip install --quiet yfinance pandas numpy tqdm

import yfinance as yf
import pandas as pd
import numpy as np
from tqdm import tqdm
from datetime import datetime

# ---------- parameters (tweak as needed) ----------
symbol = "GC=F"
interval = "5m"
period = "60d"
min_trades = 5
retrace_fracs = [0.25, 0.5, 0.75]
stop_mults = [0.5, 1.0, 1.5]
target_mults = [1.0, 1.5, 2.0]
contract_mult = 100.0
# --------------------------------------------------

print("Downloading intraday bars (may take a moment)...")
bars = yf.download(symbol, period=period, interval=interval, progress=True, auto_adjust=False)
if bars.empty:
    raise SystemExit("No intraday data downloaded. Check ticker/period/interval.")
bars.index = bars.index.tz_localize(None)
print(f"Got {len(bars)} bars")

# group into calendar days
bars['cal_date'] = bars.index.date
grouped = bars.groupby('cal_date')
print("Days available:", len(grouped))

def day_atr_estimate(df_day):
    """Return a scalar float ATR estimate for the day (safe)."""
    if df_day.empty:
        return np.nan
    highs = df_day['High'].max()
    lows = df_day['Low'].min()
    tr = highs - lows
    return float(tr / 14.0) if tr > 0 else float(np.nan)

def backtest_day_safe(df_day, retrace_frac=0.5, atr_stop_mult=1.0, atr_target_mult=1.5):
    """
    Robust per-day simulation. Returns float PnL in USD, or None if no trade.
    """
    try:
        if df_day.empty:
            return None

        day_open = float(df_day['Open'].iloc[0])
        day_high = float(df_day['High'].max())
        day_low = float(df_day['Low'].min())

        up_move = day_high - day_open
        down_move = day_open - day_low

        # decide bias
        bias_short = (up_move > down_move)
        bias_long = not bias_short

        # entry price as retrace fraction
        if bias_short:
            entry_price = day_open + float(retrace_frac) * up_move
        else:
            entry_price = day_open - float(retrace_frac) * down_move

        atr = day_atr_estimate(df_day)
        if np.isnan(atr) or atr <= 0:
            return None

        stop_dist = atr * float(atr_stop_mult)
        target_dist = atr * float(atr_target_mult)

        if bias_short:
            stop_price = entry_price + stop_dist
            target_price = entry_price - target_dist
            # build boolean masks explicitly
            target_mask = df_day['Low'] <= target_price
            stop_mask = df_day['High'] >= stop_price
            target_hits_idx = df_day.index[target_mask].tolist()
            stop_hits_idx   = df_day.index[stop_mask].tolist()
        else:
            stop_price = entry_price - stop_dist
            target_price = entry_price + target_dist
            target_mask = df_day['High'] >= target_price
            stop_mask = df_day['Low'] <= stop_price
            target_hits_idx = df_day.index[target_mask].tolist()
            stop_hits_idx   = df_day.index[stop_mask].tolist()

        # no hits -> exit at close
        if (len(target_hits_idx) == 0) and (len(stop_hits_idx) == 0):
            exit_price = float(df_day['Close'].iloc[-1])
            pnl_price = (entry_price - exit_price) if bias_short else (exit_price - entry_price)
            return float(pnl_price * contract_mult)

        # only target hit
        if (len(target_hits_idx) > 0) and (len(stop_hits_idx) == 0):
            pnl_price = (entry_price - target_price) if bias_short else (target_price - entry_price)
            return float(pnl_price * contract_mult)

        # only stop hit
        if (len(stop_hits_idx) > 0) and (len(target_hits_idx) == 0):
            pnl_price = (entry_price - stop_price) if bias_short else (stop_price - entry_price)
            return float(pnl_price * contract_mult)

        # both hit: whichever timestamp is earlier
        t_time = target_hits_idx[0]
        s_time = stop_hits_idx[0]
        if t_time < s_time:
            pnl_price = (entry_price - target_price) if bias_short else (target_price - entry_price)
        else:
            pnl_price = (entry_price - stop_price) if bias_short else (stop_price - entry_price)
        return float(pnl_price * contract_mult)

    except Exception as e:
        # propagate exception so caller can log but return None to avoid crash
        raise RuntimeError(f"Error simulating day {df_day.index[0].date()}: {e}") from e

# Run grid safely and log any problematic days
results = []
error_log = []

for retr in retrace_fracs:
    for stop_mult in stop_mults:
        for tgt_mult in target_mults:
            pnls = []
            days_with_trade = 0
            for day, df_day in grouped:
                try:
                    pnl = backtest_day_safe(df_day, retrace_frac=retr, atr_stop_mult=stop_mult, atr_target_mult=tgt_mult)
                except Exception as e:
                    # log and continue
                    error_log.append({'date': day, 'retrace_frac': retr, 'stop_mult': stop_mult, 'tgt_mult': tgt_mult, 'error': str(e)})
                    # do not add pnl for this day
                    continue
                if pnl is not None:
                    pnls.append(pnl)
                    days_with_trade += 1
            if days_with_trade >= min_trades:
                total = float(np.sum(pnls)) if pnls else 0.0
                mean = float(np.mean(pnls)) if pnls else 0.0
                results.append({
                    'retrace_frac': retr,
                    'atr_stop_mult': stop_mult,
                    'atr_target_mult': tgt_mult,
                    'trades': days_with_trade,
                    'mean_pnl': mean,
                    'total_pnl': total
                })

res_df = pd.DataFrame(results)
if res_df.empty:
    print("No configs met the minimum-trades requirement (min_trades =", min_trades, ").")
else:
    res_df = res_df.sort_values('total_pnl', ascending=False).reset_index(drop=True)
    print("Top configs (by total PnL) that have >= {} trades:".format(min_trades))
    display(res_df.head(10))
    res_df.to_csv("/mnt/data/grid_fixed_results_safe.csv", index=False)
    print("Saved /mnt/data/grid_fixed_results_safe.csv")

# print any logged errors (first 10)
if len(error_log) > 0:
    print("\nSome days/configs raised exceptions during simulation (showing up to 10):")
    for e in error_log[:10]:
        print(e)
else:
    print("\nNo per-day simulation errors detected.")

# diagnostic: how many days would produce any trade under at least one config?
day_trade_any = {}
for day, df_day in grouped:
    any_trade = False
    for retr in retrace_fracs:
        for stop_mult in stop_mults:
            for tgt_mult in target_mults:
                try:
                    if backtest_day_safe(df_day, retr, stop_mult, tgt_mult) is not None:
                        any_trade = True
                        break
                except Exception as e:
                    # ignore day-specific exceptions here
                    any_trade = any_trade or False
            if any_trade: break
        if any_trade: break
    day_trade_any[day] = any_trade
n_days_with_any_trade = sum(1 for v in day_trade_any.values() if v)
print("Days that would produce at least one trade under any config:", n_days_with_any_trade, "out of", len(day_trade_any))
```

```python
# Colab-ready: counter-trend (fade the trend) strategy backtest on GC=F (daily)
# Explanation (short):
# 1) Detect "significant trend" if price has moved by at least TREND_RET_THRESHOLD over the last TREND_WINDOW days AND EMA short > EMA long.
# 2) When trend is detected (up or down), wait for an exhaustion signal:
#      - For an uptrend: RSI14 > RSI_ENTRY (overbought) -> SHORT at next open.
#      - For a downtrend: RSI14 < (100 - RSI_ENTRY) (oversold) -> LONG at next open.
# 3) Target: move halfway (50%) from entry back to the 20-day SMA (mean reversion).
# 4) Stop: ATR(14) * STOP_ATR_MULT away from entry.
# 5) Use daily OHLC to simulate intraday fills: entry at next open; check if day reached target (win) or stop (loss); if neither, exit at close.
#
# Tweak parameters in the PARAMETERS block below.

!pip install --quiet yfinance pandas matplotlib numpy ta

import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime, timedelta
import os

# ---------------- PARAMETERS (tweak these) ----------------
TICKER = "GC=F"
START_DATE = "2015-01-01"
END_DATE = None  # None => up to today
TREND_WINDOW = 10             # days to measure short-term trend momentum
TREND_RET_THRESHOLD = 0.025   # 2.5% move over TREND_WINDOW qualifies as "significant trend"
EMA_SHORT = 10
EMA_LONG = 50
RSI_PERIOD = 14
RSI_ENTRY = 70                # RSI above this in an uptrend -> short (exhaustion)
STOP_ATR_MULT = 1.0           # stop distance in ATR multiples
TARGET_RETRACE_PCT = 0.5      # aim for 50% retrace toward 20-day SMA
SMA_TARGET_LEN = 20
SLIPPAGE = 0.0                # price units per trade
COMMISSION = 0.0              # price units per trade
CONTRACT_MULTIPLIER = 100.0   # Gold: $ per 1 price unit
OUTPUT_DIR = "/mnt/data/gold_countertrend_results"
MIN_DAYS_REQUIRED = 250       # avoid tiny run (optional)
# --------------------------------------------------------

# helper functions
def compute_indicators(df):
    # EMAs
    df['ema_short'] = df['Close'].ewm(span=EMA_SHORT, adjust=False).mean()
    df['ema_long']  = df['Close'].ewm(span=EMA_LONG, adjust=False).mean()
    # SMA target
    df['sma20'] = df['Close'].rolling(SMA_TARGET_LEN).mean()
    # RSI (classic)
    delta = df['Close'].diff()
    up = delta.clip(lower=0)
    down = -1 * delta.clip(upper=0)
    ma_up = up.rolling(RSI_PERIOD).mean()
    ma_down = down.rolling(RSI_PERIOD).mean()
    rs = ma_up / ma_down
    df['rsi'] = 100 - (100 / (1 + rs))
    # ATR(14)
    high = df['High']; low = df['Low']; close = df['Close']
    tr = pd.concat([high - low, (high - close.shift(1)).abs(), (low - close.shift(1)).abs()], axis=1).max(axis=1)
    df['atr'] = tr.rolling(14, min_periods=1).mean()
    # short-term return over TREND_WINDOW
    df['ret_trend'] = df['Close'].pct_change(periods=TREND_WINDOW)
    return df

def generate_signals(df):
    """
    Create rows where a trade is signaled:
    - For uptrend (ret_trend >= threshold and ema_short > ema_long): when rsi > RSI_ENTRY -> SHORT next open
    - For downtrend (ret_trend <= -threshold and ema_short < ema_long): when rsi < (100-RSI_ENTRY) -> LONG next open
    """
    signals = []
    for i in range(len(df)-1):  # we need next day for entry
        row = df.iloc[i]
        next_row = df.iloc[i+1]
        # ensure indicators present
        if np.isnan(row['rsi']) or np.isnan(row['ret_trend']) or np.isnan(row['atr']) or np.isnan(row['sma20']):
            continue
        # uptrend detection
        if (row['ret_trend'] >= TREND_RET_THRESHOLD) and (row['ema_short'] > row['ema_long']):
            # exhaustion: overbought
            if row['rsi'] >= RSI_ENTRY:
                signals.append({
                    'signal_date': df.index[i],
                    'entry_date': df.index[i+1],
                    'side': 'SHORT',
                    'reason': 'uptrend_exhaustion',
                    'rsi': row['rsi'],
                    'ret_trend': row['ret_trend']
                })
        # downtrend detection
        if (row['ret_trend'] <= -TREND_RET_THRESHOLD) and (row['ema_short'] < row['ema_long']):
            # exhaustion: oversold
            if row['rsi'] <= (100 - RSI_ENTRY):
                signals.append({
                    'signal_date': df.index[i],
                    'entry_date': df.index[i+1],
                    'side': 'LONG',
                    'reason': 'downtrend_exhaustion',
                    'rsi': row['rsi'],
                    'ret_trend': row['ret_trend']
                })
    return pd.DataFrame(signals)

def run_backtest(df, signals):
    trades = []
    for _, s in signals.iterrows():
        entry_date = s['entry_date']
        # find the daily row for entry_date (it must exist)
        if entry_date not in df.index:
            continue
        row_entry = df.loc[entry_date]
        entry_price = row_entry['Open']  # assume entry at open
        side = s['side']
        atr = row_entry['atr'] if not np.isnan(row_entry['atr']) else df.loc[:entry_date]['atr'].iloc[-1]
        sma20 = row_entry['sma20'] if not np.isnan(row_entry['sma20']) else df.loc[:entry_date]['sma20'].iloc[-1]
        # target: halfway from entry back to sma20 (mean reversion)
        if side == 'SHORT':
            # target is lower than entry: halfway toward sma20 but not beyond
            target_price = entry_price - TARGET_RETRACE_PCT * (entry_price - sma20)
            # stop above entry
            stop_price = entry_price + STOP_ATR_MULT * atr
        else:  # LONG
            target_price = entry_price + TARGET_RETRACE_PCT * (sma20 - entry_price)
            stop_price = entry_price - STOP_ATR_MULT * atr

        # simulate day: use entry day OHLC to see whether target or stop is hit intraday
        day_high = row_entry['High']
        day_low = row_entry['Low']
        win = False
        exit_price = None
        pnl_price = None

        # When side SHORT: we win if Low <= target_price (price moved down to target)
        # If High >= stop_price -> stop hit (loss)
        if side == 'SHORT':
            # check which happened first — we cannot know intra-bar order with daily OHLC. We'll use priority:
            # If both target and stop are within day's range, assume the one closer to the Open (entry) happened first.
            # This is a heuristic; it's conservative-ish.
            possible_target = (day_low <= target_price)
            possible_stop = (day_high >= stop_price)
            if possible_target and not possible_stop:
                win = True; exit_price = target_price
            elif possible_stop and not possible_target:
                win = False; exit_price = stop_price
            elif possible_target and possible_stop:
                # compare distances from entry: whichever is closer likely hit first — assume that
                dist_target = abs(entry_price - target_price)
                dist_stop = abs(stop_price - entry_price)
                if dist_target <= dist_stop:
                    win = True; exit_price = target_price
                else:
                    win = False; exit_price = stop_price
            else:
                # neither touched: exit at close
                exit_price = row_entry['Close']
                win = (exit_price <= target_price)  # unlikely but formal
            pnl_price = entry_price - exit_price  # SHORT pnl = entry - exit
        else:  # LONG
            possible_target = (day_high >= target_price)
            possible_stop = (day_low <= stop_price)
            if possible_target and not possible_stop:
                win = True; exit_price = target_price
            elif possible_stop and not possible_target:
                win = False; exit_price = stop_price
            elif possible_target and possible_stop:
                dist_target = abs(entry_price - target_price)
                dist_stop = abs(stop_price - entry_price)
                if dist_target <= dist_stop:
                    win = True; exit_price = target_price
                else:
                    win = False; exit_price = stop_price
            else:
                exit_price = row_entry['Close']
                win = (exit_price >= target_price)
            pnl_price = exit_price - entry_price  # LONG pnl = exit - entry

        pnl_price_after = pnl_price - SLIPPAGE - COMMISSION
        pnl_usd = pnl_price_after * CONTRACT_MULTIPLIER

        trades.append({
            'signal_date': s['signal_date'],
            'entry_date': s['entry_date'],
            'side': side,
            'entry_price': entry_price,
            'exit_price': exit_price,
            'target_price': target_price,
            'stop_price': stop_price,
            'win': win,
            'pnl_price': pnl_price_after,
            'pnl_usd': pnl_usd,
            'rsi_at_signal': s['rsi'],
            'ret_trend_at_signal': s['ret_trend'],
            'atr_at_entry': atr,
            'sma20_at_entry': sma20
        })
    trades_df = pd.DataFrame(trades).set_index('entry_date').sort_index()
    return trades_df

# === Main run ===
print("Downloading historical daily data for", TICKER, "...")
end_dt = datetime.today().strftime("%Y-%m-%d") if END_DATE is None else END_DATE
df = yf.download(TICKER, start=START_DATE, end=end_dt, interval="1d", progress=True, auto_adjust=False)
if df.empty:
    raise SystemExit("No data downloaded. Check ticker or dates.")
# flatten MultiIndex if any
if isinstance(df.columns, pd.MultiIndex):
    df.columns = [c[0] if isinstance(c, tuple) else c for c in df.columns]
df = df[['Open','High','Low','Close','Adj Close','Volume']].copy()
df.index = pd.to_datetime(df.index)
df = compute_indicators(df)

# quick sanity check
if len(df) < MIN_DAYS_REQUIRED:
    print("Warning: short history (< MIN_DAYS_REQUIRED). Results will be noisy.")

signals = generate_signals(df)
print("Signals generated:", len(signals))

trades_df = run_backtest(df, signals)
print("Trades simulated:", len(trades_df))

# basic performance printout
if len(trades_df) == 0:
    print("No trades generated with current parameters. Try loosening thresholds.")
else:
    total_trades = len(trades_df)
    wins = trades_df['win'].sum()
    win_rate = wins / total_trades
    avg_pnl_price = trades_df['pnl_price'].mean()
    avg_win = trades_df[trades_df['win']==True]['pnl_price'].mean() if wins>0 else np.nan
    avg_loss = trades_df[trades_df['win']==False]['pnl_price'].mean() if wins<total_trades else np.nan
    expectancy = avg_pnl_price
    cumulative_pnl_usd = trades_df['pnl_usd'].sum()

    print("\n=== Strategy results ===")
    print(f"Trades: {total_trades}, Wins: {int(wins)}, Win rate: {win_rate:.2%}")
    print(f"Avg PnL (price units): {avg_pnl_price:.4f}, Avg win: {avg_win:.4f}, Avg loss: {avg_loss:.4f}")
    print(f"Expectancy per trade (price units): {expectancy:.4f}")
    print(f"Cumulative PnL (USD, 1 contract): ${cumulative_pnl_usd:,.2f}")

    # Save results
    os.makedirs(OUTPUT_DIR, exist_ok=True)
    trades_df.to_csv(os.path.join(OUTPUT_DIR, "countertrend_trades.csv"))
    print("Saved trades to:", os.path.join(OUTPUT_DIR, "countertrend_trades.csv"))

    # equity curve
    trades_df['cum_pnl_usd'] = trades_df['pnl_usd'].cumsum()
    plt.figure(figsize=(10,5))
    plt.plot(trades_df.index, trades_df['cum_pnl_usd'])
    plt.title("Equity curve (USD) - Countertrend strategy on " + TICKER)
    plt.xlabel("Entry date")
    plt.ylabel("Cumulative PnL (USD) per 1 contract")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    # trades summary head
    display(trades_df[['side','entry_price','exit_price','target_price','stop_price','win','pnl_price','pnl_usd','rsi_at_signal','ret_trend_at_signal']].iloc[::-1].head(40))

    # simple stats by weekday
    trades_df['weekday'] = trades_df.index.weekday
    print("\nPerformance by weekday (entry day):")
    print(trades_df.groupby('weekday')['pnl_usd'].agg(['count','sum','mean']).sort_values('count', ascending=False))
```

```python
# Grid-search (daily) for counter-trend configs that produce positive cumulative PnL and decent expectancy.
# Paste and run in Google Colab.
!pip install --quiet yfinance pandas numpy matplotlib tqdm

import yfinance as yf
import pandas as pd
import numpy as np
import os
from itertools import product
from datetime import datetime
import matplotlib.pyplot as plt

# ---------- User params ----------
TICKER = "GC=F"
START_DATE = "2010-01-01"
END_DATE = None
OUTPUT_DIR = "/mnt/data/gold_countertrend_grid_search"
os.makedirs(OUTPUT_DIR, exist_ok=True)

MIN_TRADES = 40        # require at least this many trades to consider config
CONTRACT_MULT = 100.0

TREND_WINDOW_LIST = [5, 10, 15]
TREND_RET_LIST = [0.01, 0.015, 0.025, 0.04]    # smaller values -> more trades
EMA_SHORT_LIST = [5, 10]
EMA_LONG_LIST = [20, 50]
RSI_ENTRY_LIST = [65, 70, 75]
ATR_STOP_MULT_LIST = [0.5, 0.75, 1.0, 1.5]
TARGET_RETRACE_LIST = [0.4, 0.5, 0.6]
SKIP_WEEKDAYS = []   # e.g. [0] to skip Monday; leave [] to allow all

# ---------- helpers ----------
def compute_indicators(df, ema_short, ema_long, sma_len=20, rsi_period=14):
    df = df.copy()
    df['ema_short'] = df['Close'].ewm(span=ema_short, adjust=False).mean()
    df['ema_long']  = df['Close'].ewm(span=ema_long, adjust=False).mean()
    df['sma20'] = df['Close'].rolling(window=sma_len, min_periods=1).mean()
    # RSI
    delta = df['Close'].diff()
    up = delta.clip(lower=0)
    down = -delta.clip(upper=0)
    ma_up = up.rolling(window=rsi_period, min_periods=1).mean()
    ma_down = down.rolling(window=rsi_period, min_periods=1).mean()
    rs = ma_up / ma_down.replace(0, np.nan)
    df['rsi'] = 100 - (100 / (1 + rs))
    # ATR14
    high = df['High']; low = df['Low']; close = df['Close']
    tr = pd.concat([high-low, (high - close.shift(1)).abs(), (low - close.shift(1)).abs()], axis=1).max(axis=1)
    df['atr14'] = tr.rolling(window=14, min_periods=1).mean()
    return df

def gen_signals(df, trend_window, trend_ret, ema_s, ema_l, rsi_e):
    df = df.copy()
    df['ret_trend'] = df['Close'].pct_change(periods=trend_window)
    rows = []
    for i in range(len(df)-1):
        row = df.iloc[i]
        # skip if not enough data
        if pd.isna(row.get('rsi')) or pd.isna(row.get('ret_trend')) or pd.isna(row.get('atr14')) or pd.isna(row.get('sma20')):
            continue
        # uptrend exhaustion -> SHORT
        if (row['ret_trend'] >= trend_ret) and (row['ema_short'] > row['ema_long']) and (row['rsi'] >= rsi_e):
            rows.append({'entry_date': df.index[i+1], 'side': 'SHORT', 'signal_date': df.index[i]})
        # downtrend exhaustion -> LONG
        if (row['ret_trend'] <= -trend_ret) and (row['ema_short'] < row['ema_long']) and (row['rsi'] <= (100 - rsi_e)):
            rows.append({'entry_date': df.index[i+1], 'side': 'LONG', 'signal_date': df.index[i]})
    return pd.DataFrame(rows)

def backtest(df, signals, stop_atr_mult, target_retrace, skip_weekdays):
    trades = []
    for _, s in signals.iterrows():
        entry_date = s['entry_date']
        if entry_date not in df.index: continue
        if entry_date.weekday() in skip_weekdays:
            continue
        r = df.loc[entry_date]
        entry = r['Open']
        atr = r['atr14']
        sma = r['sma20']
        if np.isnan(atr) or atr <= 0 or np.isnan(sma):
            continue
        side = s['side']
        if side == 'SHORT':
            target = entry - target_retrace * (entry - sma)
            stop = entry + stop_atr_mult * atr
            possible_target = (r['Low'] <= target)
            possible_stop = (r['High'] >= stop)
            if possible_target and not possible_stop:
                exit_price = target; win = True
            elif possible_stop and not possible_target:
                exit_price = stop; win = False
            elif possible_target and possible_stop:
                # choose closer first (heuristic)
                if abs(entry - target) <= abs(stop - entry):
                    exit_price = target; win = True
                else:
                    exit_price = stop; win = False
            else:
                exit_price = r['Close']; win = exit_price <= target
            pnl_price = entry - exit_price
        else:  # LONG
            target = entry + target_retrace * (sma - entry)
            stop = entry - stop_atr_mult * atr
            possible_target = (r['High'] >= target)
            possible_stop = (r['Low'] <= stop)
            if possible_target and not possible_stop:
                exit_price = target; win = True
            elif possible_stop and not possible_target:
                exit_price = stop; win = False
            elif possible_target and possible_stop:
                if abs(entry - target) <= abs(entry - stop):
                    exit_price = target; win = True
                else:
                    exit_price = stop; win = False
            else:
                exit_price = r['Close']; win = exit_price >= target
            pnl_price = exit_price - entry
        pnl_usd = pnl_price * CONTRACT_MULT
        trades.append({'entry_date': entry_date, 'side': side, 'entry': entry, 'exit': exit_price, 'win': win, 'pnl_price': pnl_price, 'pnl_usd': pnl_usd})
    trades_df = pd.DataFrame(trades).set_index('entry_date').sort_index()
    return trades_df

# ---------- load data ----------
end_str = datetime.today().strftime("%Y-%m-%d") if END_DATE is None else END_DATE
raw = yf.download(TICKER, start=START_DATE, end=end_str, interval="1d", auto_adjust=False, progress=True)
if isinstance(raw.columns, pd.MultiIndex): raw.columns = [c[0] if isinstance(c, tuple) else c for c in raw.columns]
df = raw[['Open','High','Low','Close','Volume']].copy()
df.index = pd.to_datetime(df.index)

# ---------- grid search ----------
grid = list(product(TREND_WINDOW_LIST, TREND_RET_LIST, EMA_SHORT_LIST, EMA_LONG_LIST, RSI_ENTRY_LIST, ATR_STOP_MULT_LIST, TARGET_RETRACE_LIST))
results = []
for (trend_w, trend_ret, ema_s, ema_l, rsi_e, atr_stop, targ_ret) in grid:
    df_ind = compute_indicators(df, ema_s, ema_l)
    signals = gen_signals(df_ind, trend_w, trend_ret, ema_s, ema_l, rsi_e)
    trades_df = backtest(df_ind, signals, atr_stop, targ_ret, SKIP_WEEKDAYS)
    n = len(trades_df)
    if n < MIN_TRADES:
        continue
    total = trades_df['pnl_usd'].sum()
    mean = trades_df['pnl_usd'].mean()
    win_rate = trades_df['win'].mean() if n>0 else 0.0
    results.append({'trend_w':trend_w,'trend_ret':trend_ret,'ema_s':ema_s,'ema_l':ema_l,'rsi_e':rsi_e,'atr_stop':atr_stop,'targ_ret':targ_ret,'n_trades':n,'total_pnl':float(total),'mean_pnl':float(mean),'win_rate':float(win_rate)})

res_df = pd.DataFrame(results).sort_values(['total_pnl','mean_pnl'], ascending=False).reset_index(drop=True)
res_df.to_csv(os.path.join(OUTPUT_DIR,'grid_search_results.csv'), index=False)
print("Grid search complete. Saved to", OUTPUT_DIR)
display(res_df.head(10))
```

```python
# Colab cell: visualize + forward-test the top counter-trend config (daily GC=F)
!pip install --quiet yfinance pandas numpy matplotlib ta

import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
from datetime import datetime

# ------------------ CONFIG / user-tweak ------------------
TICKER = "GC=F"
START_DATE = "2010-01-01"
END_DATE = None      # None -> today
# Default "best" config from your grid results (change if you want)
BEST_CONFIG = {
    'trend_w': 15,
    'trend_ret': 0.01,
    'ema_s': 5,
    'ema_l': 20,
    'rsi_e': 75,
    'atr_stop': 0.5,
    'targ_ret': 0.6
}
# Optionally load first row from grid CSV if it exists
GRID_CSV = "/mnt/data/gold_countertrend_grid_search/best_configs.csv"  # your earlier path
# Forward-test split date: everything before this is "train", from this date (inclusive) is OOS/test
OOS_START = "2021-01-01"
CONTRACT_MULT = 100.0   # $ per 1 price unit (1 contract)
OUTPUT_DIR = "/mnt/data/gc_visual_forward_test"
os.makedirs(OUTPUT_DIR, exist_ok=True)
# ---------------------------------------------------------

def maybe_load_best_from_csv():
    try:
        if os.path.exists(GRID_CSV):
            df = pd.read_csv(GRID_CSV)
            if not df.empty:
                # pick the top row (first)
                top = df.iloc[0].to_dict()
                cfg = {
                    'trend_w': int(top.get('trend_w', BEST_CONFIG['trend_w'])),
                    'trend_ret': float(top.get('trend_ret', BEST_CONFIG['trend_ret'])),
                    'ema_s': int(top.get('ema_s', BEST_CONFIG['ema_s'])),
                    'ema_l': int(top.get('ema_l', BEST_CONFIG['ema_l'])),
                    'rsi_e': int(top.get('rsi_e', BEST_CONFIG['rsi_e'])),
                    'atr_stop': float(top.get('atr_stop', BEST_CONFIG['atr_stop'])),
                    'targ_ret': float(top.get('targ_ret', BEST_CONFIG['targ_ret']))
                }
                print("Loaded best config from csv:", GRID_CSV)
                return cfg
    except Exception as e:
        print("Could not load best config from CSV:", e)
    return BEST_CONFIG

# indicator & signal functions (same logic as earlier)
def compute_indicators(df, ema_s, ema_l, sma_len=20, rsi_period=14):
    df = df.copy()
    df['ema_short'] = df['Close'].ewm(span=ema_s, adjust=False).mean()
    df['ema_long']  = df['Close'].ewm(span=ema_l, adjust=False).mean()
    df['sma20'] = df['Close'].rolling(window=sma_len, min_periods=1).mean()
    delta = df['Close'].diff()
    up = delta.clip(lower=0)
    down = -delta.clip(upper=0)
    ma_up = up.rolling(window=rsi_period, min_periods=1).mean()
    ma_down = down.rolling(window=rsi_period, min_periods=1).mean()
    rs = ma_up / ma_down.replace(0, np.nan)
    df['rsi'] = 100 - (100 / (1 + rs))
    # ATR14
    high = df['High']; low = df['Low']; close = df['Close']
    tr = pd.concat([high-low, (high - close.shift(1)).abs(), (low - close.shift(1)).abs()], axis=1).max(axis=1)
    df['atr14'] = tr.rolling(window=14, min_periods=1).mean()
    return df

def generate_signals(df, trend_window, trend_ret_threshold, rsi_entry):
    df2 = df.copy()
    df2['ret_trend'] = df2['Close'].pct_change(periods=trend_window)
    rows = []
    for i in range(len(df2)-1):
        row = df2.iloc[i]
        if pd.isna(row.get('rsi')) or pd.isna(row.get('ret_trend')) or pd.isna(row.get('atr14')) or pd.isna(row.get('sma20')):
            continue
        # uptrend exhaustion -> SHORT
        if (row['ret_trend'] >= trend_ret_threshold) and (row['ema_short'] > row['ema_long']) and (row['rsi'] >= rsi_entry):
            rows.append({'entry_date': df2.index[i+1], 'side': 'SHORT', 'signal_date': df2.index[i]})
        # downtrend exhaustion -> LONG
        if (row['ret_trend'] <= -trend_ret_threshold) and (row['ema_short'] < row['ema_long']) and (row['rsi'] <= (100 - rsi_entry)):
            rows.append({'entry_date': df2.index[i+1], 'side': 'LONG', 'signal_date': df2.index[i]})
    return pd.DataFrame(rows)

def backtest_signals(df, signals, stop_atr_mult, target_retrace):
    trades = []
    for _, s in signals.iterrows():
        entry_date = s['entry_date']
        if entry_date not in df.index:
            continue
        r = df.loc[entry_date]
        entry = float(r['Open'])
        atr = float(r['atr14'])
        sma = float(r['sma20'])
        if atr <= 0 or np.isnan(atr) or np.isnan(sma):
            continue
        side = s['side']
        if side == 'SHORT':
            target = entry - target_retrace * (entry - sma)
            stop = entry + stop_atr_mult * atr
            possible_target = (r['Low'] <= target)
            possible_stop = (r['High'] >= stop)
            if possible_target and not possible_stop:
                exit_price = target; win=True
            elif possible_stop and not possible_target:
                exit_price = stop; win=False
            elif possible_target and possible_stop:
                # heuristic: closer first
                if abs(entry - target) <= abs(stop - entry):
                    exit_price = target; win=True
                else:
                    exit_price = stop; win=False
            else:
                exit_price = float(r['Close']); win = exit_price <= target
            pnl_price = entry - exit_price
        else:  # LONG
            target = entry + target_retrace * (sma - entry)
            stop = entry - stop_atr_mult * atr
            possible_target = (r['High'] >= target)
            possible_stop = (r['Low'] <= stop)
            if possible_target and not possible_stop:
                exit_price = target; win=True
            elif possible_stop and not possible_target:
                exit_price = stop; win=False
            elif possible_target and possible_stop:
                if abs(entry - target) <= abs(entry - stop):
                    exit_price = target; win=True
                else:
                    exit_price = stop; win=False
            else:
                exit_price = float(r['Close']); win = exit_price >= target
            pnl_price = exit_price - entry
        pnl_usd = pnl_price * CONTRACT_MULT
        trades.append({
            'entry_date': entry_date, 'side': side, 'entry_price': entry, 'exit_price': exit_price,
            'target_price': target, 'stop_price': stop, 'win': win, 'pnl_price': pnl_price, 'pnl_usd': pnl_usd,
            'rsi_at_signal': float(s.get('rsi', np.nan)) if 'rsi' in s else np.nan,
            'signal_date': s.get('signal_date')
        })
    trades_df = pd.DataFrame(trades).set_index('entry_date').sort_index()
    return trades_df

def trade_stats(trades):
    if trades.empty:
        return {}
    total = float(trades['pnl_usd'].sum())
    n = len(trades)
    wins = trades['win'].sum()
    win_rate = wins / n
    avg_price = trades['pnl_price'].mean()
    avg_win = trades[trades['win']==True]['pnl_price'].mean() if wins>0 else np.nan
    avg_loss = trades[trades['win']==False]['pnl_price'].mean() if wins<n else np.nan
    expectancy = avg_price
    # trade-level Sharpe (mean/std)
    std = trades['pnl_usd'].std(ddof=0)
    sharpe = (trades['pnl_usd'].mean() / std) if std and std>0 else np.nan
    # cum pnl series and max drawdown
    cum = trades['pnl_usd'].cumsum()
    running_max = cum.cummax()
    drawdown = cum - running_max
    max_dd = drawdown.min() if not drawdown.empty else 0.0
    return {'n_trades': n, 'wins': int(wins), 'win_rate': win_rate, 'total_pnl_usd': total,
            'avg_pnl_price': avg_price, 'avg_win_price': avg_win, 'avg_loss_price': avg_loss,
            'expectancy_price': expectancy, 'trade_sharpe': sharpe, 'max_drawdown_usd': max_dd,
            'cum_series': cum}

# ------------------ run ------------------
cfg = maybe_load_best_from_csv()
print("Using config:", cfg)

end_str = datetime.today().strftime("%Y-%m-%d") if END_DATE is None else END_DATE
raw = yf.download(TICKER, start=START_DATE, end=end_str, interval="1d", progress=True, auto_adjust=False)
if raw.empty:
    raise SystemExit("No data downloaded.")
if isinstance(raw.columns, pd.MultiIndex):
    raw.columns = [c[0] if isinstance(c, tuple) else c for c in raw.columns]
price = raw[['Open','High','Low','Close','Volume']].copy()
price.index = pd.to_datetime(price.index)

# indicators
price_ind = compute_indicators(price, cfg['ema_s'], cfg['ema_l'])

# signals & trades (all history)
signals_all = generate_signals(price_ind, trend_window=cfg['trend_w'], trend_ret_threshold=cfg['trend_ret'], rsi_entry=cfg['rsi_e'])
trades_all = backtest_signals(price_ind, signals_all, stop_atr_mult=cfg['atr_stop'], target_retrace=cfg['targ_ret'])
trades_all.to_csv(os.path.join(OUTPUT_DIR, "trades_all.csv"))
print("Total trades (all history):", len(trades_all))

# forward-test split
oos_start = pd.to_datetime(OOS_START)
train_df = price_ind[price_ind.index < oos_start]
test_df  = price_ind[price_ind.index >= oos_start]

signals_test = signals_all[signals_all['entry_date'] >= oos_start]
trades_test = trades_all[trades_all.index >= oos_start].copy()
trades_test.to_csv(os.path.join(OUTPUT_DIR, "trades_oos.csv"))

# stats
stats_all = trade_stats(trades_all)
stats_oos = trade_stats(trades_test)

print("\n=== All-history stats ===")
for k,v in stats_all.items():
    if k=='cum_series': continue
    print(f"{k}: {v}")
print("\n=== OOS (from", OOS_START, ") stats ===")
for k,v in stats_oos.items():
    if k=='cum_series': continue
    print(f"{k}: {v}")

# Plot equity curve (OOS)
if not trades_test.empty:
    plt.figure(figsize=(12,5))
    plt.plot(trades_test.index, trades_test['pnl_usd'].cumsum(), marker='o')
    plt.title(f"Equity Curve (OOS) - {TICKER} - config: {cfg}")
    plt.xlabel("Entry date")
    plt.ylabel("Cumulative PnL (USD)")
    plt.grid(True)
    plt.tight_layout()
    plt.savefig(os.path.join(OUTPUT_DIR, "equity_oos.png"))
    plt.show()
else:
    print("No OOS trades to plot.")

# Plot price with entry/exit markers for OOS trades (sample up to last 200 trades)
if not trades_test.empty:
    # plot price (Close) over OOS period
    fig, ax = plt.subplots(2,1, figsize=(14,8), gridspec_kw={'height_ratios':[3,1]}, sharex=True)
    ax0 = ax[0]; ax1 = ax[1]
    ax0.plot(test_df.index, test_df['Close'], label='Close', linewidth=1)
    # annotate trades
    for idx, tr in trades_test.tail(200).iterrows():
        entry_date = idx
        entry_price = tr['entry_price']
        exit_price = tr['exit_price']
        side = tr['side']
        color = 'green' if tr['win'] else 'red'
        # entry marker
        ax0.scatter(entry_date, entry_price, marker='^' if side=='LONG' else 'v', color=color, s=60, zorder=5)
        # exit marker
        ax0.scatter(entry_date, exit_price, marker='s', color=color, s=30, zorder=5)
    ax0.set_title("Price and OOS trades (markers: entry=triangle, exit=square)")
    ax0.legend()
    ax0.grid(True)
    # volume
    ax1.bar(test_df.index, test_df['Volume'], width=1.0)
    ax1.set_ylabel("Volume")
    plt.tight_layout()
    plt.savefig(os.path.join(OUTPUT_DIR, "price_trades_oos.png"))
    plt.show()

# save summary
summary = {
    'config': cfg,
    'all_stats': {k:v for k,v in stats_all.items() if k!='cum_series'},
    'oos_stats': {k:v for k,v in stats_oos.items() if k!='cum_series'}
}
pd.Series(summary).to_json(os.path.join(OUTPUT_DIR, "summary.json"))
print("\nSaved outputs to", OUTPUT_DIR)
print("Files: trades_all.csv, trades_oos.csv, equity_oos.png, price_trades_oos.png, summary.json")
```

```python
# Resample Dukascopy 5-min to daily OHLC and run forward-test using best config
!pip install --quiet yfinance pandas numpy matplotlib

import pandas as pd, os
from datetime import datetime
import matplotlib.pyplot as plt

IN_5M = "/mnt/data/dukascopy_xauusd_5min.csv"
OUT_DAILY = "/mnt/data/dukascopy_xauusd_daily.csv"
OUT_SUMMARY_DIR = "/mnt/data/gc_duka_test"
os.makedirs(OUT_SUMMARY_DIR, exist_ok=True)

# load 5-minute CSV (created by previous cell)
df5 = pd.read_csv(IN_5M, index_col=0, parse_dates=True)
df5.index = pd.to_datetime(df5.index, utc=True)

# resample to daily (UTC), you may want to align to exchange/session timezone if needed
daily = pd.DataFrame()
daily['Open']  = df5['Open'].resample('1D').first()
daily['High']  = df5['High'].resample('1D').max()
daily['Low']   = df5['Low'].resample('1D').min()
daily['Close'] = df5['Close'].resample('1D').last()
if 'Volume' in df5.columns:
    daily['Volume'] = df5['Volume'].resample('1D').sum()
daily.dropna(how='any', inplace=True)  # remove partial days

daily.to_csv(OUT_DAILY)
print("Saved daily CSV to", OUT_DAILY, "rows:", len(daily))

# --------- Now run the same forward-test code you used earlier ----------
# (this is the forward-test cell adapted to read the daily CSV and run OOS test)
BEST_CONFIG = {
    'trend_w': 15,
    'trend_ret': 0.01,
    'ema_s': 5,
    'ema_l': 20,
    'rsi_e': 75,
    'atr_stop': 0.5,
    'targ_ret': 0.6
}
OOS_START = "2021-01-01"
CONTRACT_MULT = 100.0

def compute_indicators(df, ema_s, ema_l, sma_len=20, rsi_period=14):
    df = df.copy()
    df['ema_short'] = df['Close'].ewm(span=ema_s, adjust=False).mean()
    df['ema_long']  = df['Close'].ewm(span=ema_l, adjust=False).mean()
    df['sma20'] = df['Close'].rolling(window=sma_len, min_periods=1).mean()
    delta = df['Close'].diff()
    up = delta.clip(lower=0)
    down = -delta.clip(upper=0)
    ma_up = up.rolling(window=rsi_period, min_periods=1).mean()
    ma_down = down.rolling(window=rsi_period, min_periods=1).mean()
    rs = ma_up / ma_down.replace(0, pd.NA)
    df['rsi'] = 100 - (100 / (1 + rs))
    high = df['High']; low = df['Low']; close = df['Close']
    tr = pd.concat([high-low, (high - close.shift(1)).abs(), (low - close.shift(1)).abs()], axis=1).max(axis=1)
    df['atr14'] = tr.rolling(window=14, min_periods=1).mean()
    return df

def generate_signals(df, trend_window, trend_ret_threshold, rsi_entry):
    df2 = df.copy()
    df2['ret_trend'] = df2['Close'].pct_change(periods=trend_window)
    rows = []
    for i in range(len(df2)-1):
        row = df2.iloc[i]
        if pd.isna(row.get('rsi')) or pd.isna(row.get('ret_trend')) or pd.isna(row.get('atr14')) or pd.isna(row.get('sma20')):
            continue
        if (row['ret_trend'] >= trend_ret_threshold) and (row['ema_short'] > row['ema_long']) and (row['rsi'] >= rsi_entry):
            rows.append({'entry_date': df2.index[i+1], 'side': 'SHORT', 'signal_date': df2.index[i]})
        if (row['ret_trend'] <= -trend_ret_threshold) and (row['ema_short'] < row['ema_long']) and (row['rsi'] <= (100 - rsi_entry)):
            rows.append({'entry_date': df2.index[i+1], 'side': 'LONG', 'signal_date': df2.index[i]})
    return pd.DataFrame(rows)

def backtest_signals(df, signals, stop_atr_mult, target_retrace):
    trades = []
    for _, s in signals.iterrows():
        entry_date = s['entry_date']
        if entry_date not in df.index:
            continue
        r = df.loc[entry_date]
        entry = float(r['Open'])
        atr = float(r['atr14'])
        sma = float(r['sma20'])
        if atr <= 0 or pd.isna(atr) or pd.isna(sma):
            continue
        side = s['side']
        if side == 'SHORT':
            target = entry - target_retrace * (entry - sma)
            stop = entry + stop_atr_mult * atr
            possible_target = (r['Low'] <= target)
            possible_stop = (r['High'] >= stop)
            if possible_target and not possible_stop:
                exit_price = target; win=True
            elif possible_stop and not possible_target:
                exit_price = stop; win=False
            elif possible_target and possible_stop:
                if abs(entry - target) <= abs(stop - entry):
                    exit_price = target; win=True
                else:
                    exit_price = stop; win=False
            else:
                exit_price = float(r['Close']); win = exit_price <= target
            pnl_price = entry - exit_price
        else:
            target = entry + target_retrace * (sma - entry)
            stop = entry - stop_atr_mult * atr
            possible_target = (r['High'] >= target)
            possible_stop = (r['Low'] <= stop)
            if possible_target and not possible_stop:
                exit_price = target; win=True
            elif possible_stop and not possible_target:
                exit_price = stop; win=False
            elif possible_target and possible_stop:
                if abs(entry - target) <= abs(entry - stop):
                    exit_price = target; win=True
                else:
                    exit_price = stop; win=False
            else:
                exit_price = float(r['Close']); win = exit_price >= target
            pnl_price = exit_price - entry
        pnl_usd = pnl_price * CONTRACT_MULT
        trades.append({'entry_date': entry_date, 'side': side, 'entry_price': entry, 'exit_price': exit_price, 'target_price': target, 'stop_price': stop, 'win': win, 'pnl_price': pnl_price, 'pnl_usd': pnl_usd})
    trades_df = pd.DataFrame(trades).set_index('entry_date').sort_index()
    return trades_df

# run indicators + signals on daily
daily = pd.read_csv(OUT_DAILY, index_col=0, parse_dates=True)
daily.index = pd.to_datetime(daily.index)
daily_ind = compute_indicators(daily, BEST_CONFIG['ema_s'], BEST_CONFIG['ema_l'])
signals_all = generate_signals(daily_ind, trend_window=BEST_CONFIG['trend_w'], trend_ret_threshold=BEST_CONFIG['trend_ret'], rsi_entry=BEST_CONFIG['rsi_e'])
trades_all = backtest_signals(daily_ind, signals_all, stop_atr_mult=BEST_CONFIG['atr_stop'], target_retrace=BEST_CONFIG['targ_ret'])

# forward-test split
oos_start = pd.to_datetime("2021-01-01")
trades_oos = trades_all[trades_all.index >= oos_start].copy()

# basic stats print
def print_stats(trades, label):
    if trades.empty:
        print(label, "no trades")
        return
    total = trades['pnl_usd'].sum()
    n = len(trades)
    wins = trades['win'].sum()
    print(f"\n{label} trades: {n}, wins: {wins}, win_rate: {wins/n:.3f}, total_pnl_usd: {total:.2f}")
print_stats(trades_all, "ALL")
print_stats(trades_oos, "OOS (from 2021)")

# save outputs
trades_all.to_csv(os.path.join(OUT_SUMMARY_DIR,"trades_all_duka.csv"))
trades_oos.to_csv(os.path.join(OUT_SUMMARY_DIR,"trades_oos_duka.csv"))
print("Saved trades CSVs to", OUT_SUMMARY_DIR)
```

```python
# Full Colab-ready backtest cell: flips, HTF confirmation, volatility filter, stop/target grid, walk-forward + spread/slippage
# Save this as one cell and run. Requires dukascopy 5-min CSV at /mnt/data/dukascopy_xauusd_5min.csv

!pip install --quiet pandas numpy tqdm

import pandas as pd, numpy as np, os, math
from tqdm.auto import trange, tqdm
from datetime import datetime, timedelta
import itertools, json

# ---------- USER / FILE PARAMETERS ----------
IN_5M = "/mnt/data/dukascopy_xauusd_5min.csv"   # produced by earlier steps
OUT_DIR = "/mnt/data/gc_model_tweaks_results"
os.makedirs(OUT_DIR, exist_ok=True)

# Walk-forward / optimization params
WALKFWD_FOLDS = 3              # how many walk-forward folds
TRAIN_YEARS = 1.5              # size of each training window in years
TEST_YEARS  = 0.5              # size of test window in years

# Grid to test (stop/target in ATR multiples, holding in bars)
STOP_MULTS   = [0.5, 1.0, 1.5]     # stop = entry +/- stop_mult * ATR
TARGET_MULTS = [0.6, 1.0, 1.5]     # target = entry +/- target_mult * ATR or retrace method (we'll use ATR*mult)
MAX_HOLD_BARS = [1, 5, 10, 20]     # if neither hit, exit after X bars (None = end-of-day)

# Strategy/filters
FLIP_LOGIC = [False, True]     # test original and flipped
DAILY_EMA_LEN = 20             # higher timeframe confirmation EMA length
VOL_ATR_QUANTILE = 0.9         # skip days with daily ATR above this quantile (news/very volatile days)
MIN_TRADES_PER_CONFIG = 10     # discard configs with too few test trades
SPREAD_PTS = 0.3               # spread applied per roundtrip (price units)
SLIPPAGE_PTS = 0.2             # slippage per trade (one-way)
CONTRACT_MULT = 100.0          # dollars per price unit (for USD PnL)

# Misc
RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)
# -----------------------------------------------

# ---------- Helper functions ----------
def load_and_prepare_5m(path):
    df = pd.read_csv(path, index_col=0, parse_dates=True)
    # ensure tz-aware UTC
    if df.index.tz is None:
        df.index = pd.to_datetime(df.index).tz_localize('UTC')
    else:
        df.index = df.index.tz_convert('UTC')
    # normalise column names
    df.columns = [c.title() for c in df.columns]
    # make sure numeric
    df = df[['Open','High','Low','Close']].astype(float)
    return df

def resample_daily_from_5m(df5):
    daily = pd.DataFrame()
    daily['Open']  = df5['Open'].resample('1D').first()
    daily['High']  = df5['High'].resample('1D').max()
    daily['Low']   = df5['Low'].resample('1D').min()
    daily['Close'] = df5['Close'].resample('1D').last()
    # drop days with NaN (weekends / partials)
    daily.dropna(how='any', inplace=True)
    return daily

def compute_indicators(df, ema_short=5, ema_long=20, rsi_period=14, atr_len=14):
    df = df.copy()
    df['ema_s'] = df['Close'].ewm(span=ema_short, adjust=False).mean()
    df['ema_l'] = df['Close'].ewm(span=ema_long, adjust=False).mean()
    # ATR
    high = df['High']; low = df['Low']; close = df['Close']
    tr = pd.concat([high-low, (high - close.shift(1)).abs(), (low - close.shift(1)).abs()], axis=1).max(axis=1)
    df['atr'] = tr.rolling(atr_len, min_periods=1).mean()
    # simple momentum for "trend_ret"
    df['ret_trend'] = df['Close'].pct_change(periods=15)  # 15-bar trend approx (tweakable)
    return df

def generate_signals_from_daily(daily_df, trend_w=15, trend_ret_threshold=0.01, rsi_entry=75):
    # Based on earlier logic: after big trend move + short EMA relation + RSI threshold (we'll approximate RSI with ret)
    # For simplicity, we will use the ret_trend & ema relationship computed earlier
    rows = []
    dd = daily_df.copy()
    for i in range(0, len(dd)-1):
        row = dd.iloc[i]
        next_idx = dd.index[i+1]
        # original SHORT condition: ret_trend >= threshold AND ema_s > ema_l
        if row['ret_trend'] >= trend_ret_threshold and row['ema_s'] > row['ema_l']:
            rows.append({'entry_date': next_idx, 'side':'SHORT', 'signal_date': dd.index[i]})
        # original LONG condition:
        if row['ret_trend'] <= -trend_ret_threshold and row['ema_s'] < row['ema_l']:
            rows.append({'entry_date': next_idx, 'side':'LONG', 'signal_date': dd.index[i]})
    return pd.DataFrame(rows).set_index('entry_date')

def map_daily_signals_to_5m(df5, signals_daily):
    # for each signal with entry_date D (date at 00:00 UTC), find the first 5m bar of that day
    sigs = []
    for idx, s in signals_daily.iterrows():
        # find first 5m bar >= date midnight UTC
        day_start = pd.to_datetime(idx).tz_localize('UTC')
        try:
            first_bar = df5.loc[day_start: day_start + pd.Timedelta(days=1)].iloc[0]
            entry_ts = df5.loc[day_start: day_start + pd.Timedelta(days=1)].index[0]
            sigs.append({'entry_ts': entry_ts, 'side': s['side'], 'signal_date': s['signal_date']})
        except Exception:
            # no bars for that day (holiday) -> skip
            continue
    if len(sigs)==0:
        return pd.DataFrame(columns=['entry_ts','side','signal_date']).set_index('entry_ts')
    df = pd.DataFrame(sigs).set_index('entry_ts').sort_index()
    return df

def is_within_daily_vol_filter(daily_df, date, quantile_threshold):
    # returns False if day should be skipped (i.e. daily atr > quantile)
    thr = daily_df['atr'].quantile(quantile_threshold)
    return daily_df.loc[date]['atr'] <= thr

def simulate_trade_on_5m(df5, entry_idx, side, stop_mult, target_mult, max_hold_bars, spread_pts, slippage_pts, flip=False):
    # returns dict with trade result or None
    # entry_idx is timestamp index in df5
    try:
        entry_row = df5.loc[entry_idx]
    except KeyError:
        return None
    entry_price = float(entry_row['Open'])  # enter at bar open
    # ATR at entry: use 1-day ATR from daily series? We'll use a local ATR by resampling daily earlier;
    # But simpler: use prior-day atr stored in df5_meta (we'll attach daily atr values externally).
    # For simplicity here, assume df5 has column 'daily_atr' prepared.
    if 'daily_atr' in df5.columns and not pd.isna(df5.loc[entry_idx]['daily_atr']):
        atr = float(df5.loc[entry_idx]['daily_atr'])
    else:
        # fallback: estimate ATR as 0.01% * price (tiny) to avoid errors
        atr = max(0.5, abs(entry_price)*0.0001)
    # compute raw price-level target/stop using ATR multiples; direction depends on side & flip flag
    # If flip True, invert side
    eff_side = side
    if flip:
        eff_side = 'LONG' if side=='SHORT' else 'SHORT'
    if eff_side == 'SHORT':
        stop_price = entry_price + stop_mult * atr
        target_price = entry_price - target_mult * atr
    else:
        stop_price = entry_price - stop_mult * atr
        target_price = entry_price + target_mult * atr
    # apply spread/slippage: we'll subtract half spread on entry price direction, slippage assumed adverse on exit
    # Implement simulation progressing through subsequent bars (order of hit)
    bars = df5.loc[entry_idx:]  # inclusive of entry bar
    exit_price = None; exit_ts=None; win=False; exit_reason='close'
    hold_count = 0
    for ts, row in bars.iterrows():
        if hold_count == 0:
            # the entry bar itself; assume entry executed at Open with half spread cost
            if eff_side == 'SHORT':
                effective_entry = entry_price - (spread_pts/2.0)  # for shorts we receive (price - spread/2)
            else:
                effective_entry = entry_price + (spread_pts/2.0)
            # but use the stored entry_price for target/stop comparisions
        # check high/low of current bar for hits (note: compare price levels without spread; we will account at PnL calc)
        H = float(row['High']); L = float(row['Low']); C = float(row['Close'])
        # check stop & target: if both hit in same bar, assume conservative (worse for us) outcome by checking which was hit first approx:
        # we can't know order within bar; use rule: if both hit, assume the one with closer absolute distance to Open happened first.
        hit_target = False; hit_stop=False
        if eff_side == 'SHORT':
            if L <= target_price: hit_target=True
            if H >= stop_price: hit_stop=True
            # if both true, decide by which abs(entry - level) smaller
            if hit_target and hit_stop:
                if abs(entry_price - target_price) <= abs(stop_price - entry_price):
                    hit_stop = False
                else:
                    hit_target = False
        else:
            if H >= target_price: hit_target=True
            if L <= stop_price: hit_stop=True
            if hit_target and hit_stop:
                if abs(entry_price - target_price) <= abs(entry_price - stop_price):
                    hit_stop=False
                else:
                    hit_target=False
        if hit_target or hit_stop:
            # compute exit with slippage applied (adverse)
            if hit_target:
                raw_exit = target_price
                exit_reason = 'target'
                win = True
            else:
                raw_exit = stop_price
                exit_reason = 'stop'
                win = False
            # slippage applied as adverse amount
            if eff_side == 'SHORT':
                # for short: pnl_price = entry - exit  (higher exit is loss)
                # apply slippage that makes exit worse for us:
                if exit_reason=='target':
                    raw_exit = raw_exit + slippage_pts
                else:
                    raw_exit = raw_exit + slippage_pts
            else:
                if exit_reason=='target':
                    raw_exit = raw_exit - slippage_pts
                else:
                    raw_exit = raw_exit - slippage_pts
            exit_price = raw_exit
            exit_ts = ts
            break
        # else increment hold and maybe break due to max hold
        hold_count += 1
        if max_hold_bars is not None and hold_count >= max_hold_bars:
            # exit at close of this bar with slippage and spread effects
            raw_exit = C
            exit_reason = 'time'
            # apply slippage adverse
            if eff_side == 'SHORT':
                raw_exit = raw_exit + slippage_pts
                win = (entry_price - raw_exit) > 0
            else:
                raw_exit = raw_exit - slippage_pts
                win = (raw_exit - entry_price) > 0
            exit_price = raw_exit
            exit_ts = ts
            break
    if exit_price is None:
        # no exit found in dataset (end) -> exit at last close
        last_close = float(bars.iloc[-1]['Close'])
        raw_exit = last_close
        exit_reason='end'
        if eff_side == 'SHORT':
            raw_exit = raw_exit + slippage_pts
            win = (entry_price - raw_exit) > 0
        else:
            raw_exit = raw_exit - slippage_pts
            win = (raw_exit - entry_price) > 0
        exit_price = raw_exit
        exit_ts = bars.index[-1]
    # compute pnl with spread cost accounted (roundtrip)
    # For simplicity: compute gross pnl then subtract SPREAD_PTS and convert to USD
    if eff_side == 'SHORT':
        pnl_price = entry_price - exit_price
    else:
        pnl_price = exit_price - entry_price
    pnl_price = pnl_price - spread_pts  # roundtrip spread cost
    pnl_usd = pnl_price * CONTRACT_MULT
    return {
        'entry_ts': entry_idx, 'exit_ts': exit_ts, 'side': eff_side, 'orig_side': side, 'flip': flip,
        'entry_price': entry_price, 'exit_price': exit_price, 'pnl_price': pnl_price, 'pnl_usd': pnl_usd,
        'win': bool(win), 'exit_reason': exit_reason, 'stop_mult':stop_mult, 'target_mult':target_mult, 'hold_bars': max_hold_bars
    }

# ---------- Load data ----------
print("Loading 5-min CSV:", IN_5M)
df5 = load_and_prepare_5m(IN_5M)
print("Bars:", len(df5), "from", df5.index.min(), "to", df5.index.max())

# build daily DF and daily indicators for HTF confirmation and daily ATR filter
daily = resample_daily_from_5m(df5)
daily = compute_indicators(daily, ema_short=DAILY_EMA_LEN//4 if DAILY_EMA_LEN>4 else 5, ema_long=DAILY_EMA_LEN, rsi_period=14)
print("Daily rows:", len(daily))

# attach daily ATR value to every 5-min bar so simulator can reference it
# Map daily atr to df5 via date
df5 = df5.copy()
df5['date'] = df5.index.date
daily_atr_map = daily['atr'].to_dict()
df5['daily_atr'] = df5.index.normalize().tz_localize('UTC').map(lambda ts: daily_atr_map.get(pd.Timestamp(ts).tz_localize('UTC').date(), np.nan))

# ---------- Walk-forward setup ----------
start_date = daily.index.min()
end_date = daily.index.max()
total_years = (end_date - start_date).days / 365.25
print("Data span (years):", total_years)

# build fold windows (train/test contiguous)
folds = []
cursor = start_date
while True:
    train_end = cursor + pd.DateOffset(years=TRAIN_YEARS)
    test_end  = train_end + pd.DateOffset(years=TEST_YEARS)
    if test_end > end_date:
        break
    folds.append((cursor, train_end, train_end + pd.Timedelta(days=0), test_end))
    # move cursor forward by TEST_YEARS (rolling walk-forward)
    cursor = cursor + pd.DateOffset(years=TEST_YEARS)
print("Walk-forward folds:", len(folds))

# ---------- Main walk-forward grid search ----------
results = []
all_trades_detailed = []

for fold_idx, (train_start, train_end, test_start, test_end) in enumerate(folds):
    print(f"\n=== Fold {fold_idx+1}: train {train_start.date()} -> {train_end.date()} ; test {test_start.date()} -> {test_end.date()} ===")
    # sample daily slice for train
    daily_train = daily.loc[train_start:train_end].copy()
    daily_test  = daily.loc[test_start:test_end].copy()
    if len(daily_train) < 10 or len(daily_test) < 5:
        print("Skipping fold due to insufficient days")
        continue
    # generate signals on daily train + test separately (we'll generate on train to pick params then apply to test)
    signals_train_daily = generate_signals_from_daily(daily_train, trend_w=15, trend_ret_threshold=0.01)
    # generate signals for the whole period to map into 5min entry times
    signals_total_daily = generate_signals_from_daily(daily.loc[train_start:test_end], trend_w=15, trend_ret_threshold=0.01)
    signals_total_5m = map_daily_signals_to_5m(df5, signals_total_daily)
    # restrict to test entries later when scoring
    # grid iterate
    combos = list(itertools.product(STOP_MULTS, TARGET_MULTS, MAX_HOLD_BARS, FLIP_LOGIC))
    print("Configs to test in fold:", len(combos))
    for stop_mult, target_mult, max_hold, flip_flag in combos:
        # For performance we only simulate hits that occur in test period, but we need signals inside test range
        # Build candidate signals in test range
        signals_test_5m = signals_total_5m.loc[(signals_total_5m.index >= test_start) & (signals_total_5m.index <= test_end)]
        trades = []
        # For each signal, apply daily volatility filter and HTF trend confirmation
        for entry_ts, s in signals_test_5m.iterrows():
            # HTF confirmation: only trade if daily ema trend agrees (e.g., short only if daily ema_short > ema_long)
            day = pd.Timestamp(entry_ts).tz_convert('UTC').normalize()
            if day.tzinfo is None:
                day = day.tz_localize('UTC')
            if day.date() not in daily.index.date:
                # map day to nearest daily index
                day_key = pd.Timestamp(day.date()).tz_localize('UTC')
            else:
                day_key = pd.Timestamp(day.date()).tz_localize('UTC')
            # ensure day exists in daily
            if day_key not in daily.index:
                continue
            daily_row = daily.loc[day_key]
            # volatility/news filter
            if daily_row['atr'] > daily['atr'].quantile(VOL_ATR_QUANTILE):
                continue
            # HTF confirmation:
            intended_side = s['side']
            if intended_side == 'SHORT' and not (daily_row['ema_s'] > daily_row['ema_l']):
                continue
            if intended_side == 'LONG' and not (daily_row['ema_s'] < daily_row['ema_l']):
                continue
            # now simulate on 5m bars starting at entry_ts
            res = simulate_trade_on_5m(df5, entry_ts, s['side'], stop_mult, target_mult, max_hold, SPREAD_PTS, SLIPPAGE_PTS, flip=flip_flag)
            if res is not None:
                # attach fold and meta
                res.update({'fold': fold_idx+1, 'test_start': test_start.date().isoformat(), 'test_end': test_end.date().isoformat()})
                trades.append(res)
        # compute stats for this combo
        if len(trades) < MIN_TRADES_PER_CONFIG:
            continue
        tdf = pd.DataFrame(trades)
        total_pnl = float(tdf['pnl_usd'].sum())
        wins = int(tdf['win'].sum())
        win_rate = wins / len(tdf)
        avg_pnl = tdf['pnl_price'].mean()
        sharpe = (tdf['pnl_usd'].mean() / tdf['pnl_usd'].std(ddof=0)) if tdf['pnl_usd'].std(ddof=0) != 0 else np.nan
        results.append({
            'fold': fold_idx+1, 'stop_mult':stop_mult, 'target_mult':target_mult, 'max_hold_bars':max_hold,
            'flip': flip_flag, 'n_trades':len(tdf), 'wins':wins, 'win_rate':win_rate, 'total_pnl_usd':total_pnl,
            'avg_pnl_price': avg_pnl, 'sharpe': sharpe
        })
        # save detailed trades for inspection (append)
        for r in trades:
            all_trades_detailed.append(r)

# ---------- summarize and save ----------
res_df = pd.DataFrame(results).sort_values(by='total_pnl_usd', ascending=False).reset_index(drop=True)
res_df.to_csv(os.path.join(OUT_DIR, "wf_grid_results.csv"), index=False)
pd.DataFrame(all_trades_detailed).to_csv(os.path.join(OUT_DIR, "wf_all_trades_detailed.csv"), index=False)

print("\nTop configs by test cumulative PnL:")
print(res_df.head(10).to_string(index=False))

print("\nSaved results to", OUT_DIR)
print("Files: wf_grid_results.csv, wf_all_trades_detailed.csv")
```

```python
# Full fixed walk-forward backtest cell (loads dukascopy 5-min CSV, maps daily ATR safely, runs grid WF)
!pip install --quiet pandas numpy tqdm

import pandas as pd, numpy as np, os, math, traceback
from tqdm.auto import tqdm
from datetime import datetime
import itertools, json

# ---------- USER / FILE PARAMETERS ----------
IN_5M = "/mnt/data/dukascopy_xauusd_5min.csv"   # must exist
OUT_DIR = "/mnt/data/gc_model_tweaks_results"
os.makedirs(OUT_DIR, exist_ok=True)

# Walk-forward / optimization params
WALKFWD_FOLDS = 3              # not strictly used - folds built by date windows below
TRAIN_YEARS = 1.5
TEST_YEARS  = 0.5

# Grid to test
STOP_MULTS   = [0.5, 1.0, 1.5]
TARGET_MULTS = [0.6, 1.0, 1.5]
MAX_HOLD_BARS = [1, 5, 10, 20]

# Strategy/filters
FLIP_LOGIC = [False, True]
DAILY_EMA_LEN = 20
VOL_ATR_QUANTILE = 0.9
MIN_TRADES_PER_CONFIG = 10
SPREAD_PTS = 0.3
SLIPPAGE_PTS = 0.2
CONTRACT_MULT = 100.0
RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)
# -----------------------------------------------

def load_and_prepare_5m(path):
    df = pd.read_csv(path, index_col=0, parse_dates=True)
    # ensure tz-aware UTC
    if df.index.tz is None:
        df.index = pd.to_datetime(df.index).tz_localize('UTC')
    else:
        df.index = df.index.tz_convert('UTC')
    df.columns = [c.title() for c in df.columns]
    # ensure required columns exist
    for col in ['Open','High','Low','Close']:
        if col not in df.columns:
            raise SystemExit(f"CSV missing required column: {col}")
    df = df[['Open','High','Low','Close']].astype(float)
    return df

def resample_daily_from_5m(df5):
    daily = pd.DataFrame()
    daily['Open']  = df5['Open'].resample('1D').first()
    daily['High']  = df5['High'].resample('1D').max()
    daily['Low']   = df5['Low'].resample('1D').min()
    daily['Close'] = df5['Close'].resample('1D').last()
    daily.dropna(how='any', inplace=True)
    return daily

def compute_indicators(df, ema_short=5, ema_long=20, atr_len=14):
    df = df.copy()
    df['ema_s'] = df['Close'].ewm(span=ema_short, adjust=False).mean()
    df['ema_l'] = df['Close'].ewm(span=ema_long, adjust=False).mean()
    high = df['High']; low = df['Low']; close = df['Close']
    tr = pd.concat([high-low, (high - close.shift(1)).abs(), (low - close.shift(1)).abs()], axis=1).max(axis=1)
    df['atr'] = tr.rolling(atr_len, min_periods=1).mean()
    df['ret_trend'] = df['Close'].pct_change(periods=15)
    return df

def generate_signals_from_daily(daily_df, trend_ret_threshold=0.01):
    rows = []
    dd = daily_df.copy()
    for i in range(0, len(dd)-1):
        row = dd.iloc[i]; next_idx = dd.index[i+1]
        if row['ret_trend'] >= trend_ret_threshold and row['ema_s'] > row['ema_l']:
            rows.append({'entry_date': next_idx, 'side':'SHORT', 'signal_date': dd.index[i]})
        if row['ret_trend'] <= -trend_ret_threshold and row['ema_s'] < row['ema_l']:
            rows.append({'entry_date': next_idx, 'side':'LONG', 'signal_date': dd.index[i]})
    if len(rows)==0:
        return pd.DataFrame(columns=['entry_date','side','signal_date']).set_index('entry_date')
    return pd.DataFrame(rows).set_index('entry_date')

def map_daily_signals_to_5m(df5, signals_daily):
    sigs = []
    for idx, s in signals_daily.iterrows():
        day_start = pd.Timestamp(idx.date()).tz_localize('UTC')
        # take the first bar on that day
        day_bars = df5.loc[day_start: day_start + pd.Timedelta(days=1) - pd.Timedelta(minutes=1)]
        if len(day_bars)==0:
            continue
        entry_ts = day_bars.index[0]
        sigs.append({'entry_ts': entry_ts, 'side': s['side'], 'signal_date': s['signal_date']})
    if len(sigs)==0:
        return pd.DataFrame(columns=['entry_ts','side','signal_date']).set_index('entry_ts')
    return pd.DataFrame(sigs).set_index('entry_ts').sort_index()

def simulate_trade_on_5m(df5, entry_idx, side, stop_mult, target_mult, max_hold_bars, spread_pts, slippage_pts, flip=False):
    try:
        entry_row = df5.loc[entry_idx]
    except KeyError:
        return None
    entry_price = float(entry_row['Open'])
    # use mapped daily_atr column if available
    atr = float(df5.loc[entry_idx].get('daily_atr', np.nan))
    if pd.isna(atr) or atr <= 0:
        atr = max(0.5, abs(entry_price)*0.0001)
    eff_side = 'LONG' if (side=='SHORT' and flip) or (side=='LONG' and not flip) else ('SHORT' if (side=='LONG' and flip) or (side=='SHORT' and not flip) else side)
    # compute targets/stops
    if eff_side == 'SHORT':
        stop_price = entry_price + stop_mult * atr
        target_price = entry_price - target_mult * atr
    else:
        stop_price = entry_price - stop_mult * atr
        target_price = entry_price + target_mult * atr

    bars = df5.loc[entry_idx:]
    exit_price = None; exit_ts=None; win=False; exit_reason='close'
    hold_count=0
    for ts, row in bars.iterrows():
        H=float(row['High']); L=float(row['Low']); C=float(row['Close'])
        hit_target=False; hit_stop=False
        if eff_side=='SHORT':
            if L <= target_price: hit_target=True
            if H >= stop_price: hit_stop=True
            if hit_target and hit_stop:
                if abs(entry_price - target_price) <= abs(stop_price - entry_price):
                    hit_stop=False
                else:
                    hit_target=False
        else:
            if H >= target_price: hit_target=True
            if L <= stop_price: hit_stop=True
            if hit_target and hit_stop:
                if abs(entry_price - target_price) <= abs(entry_price - stop_price):
                    hit_stop=False
                else:
                    hit_target=False
        if hit_target or hit_stop:
            raw_exit = target_price if hit_target else stop_price
            exit_reason = 'target' if hit_target else 'stop'
            # apply slippage adverse
            if eff_side=='SHORT':
                raw_exit = raw_exit + slippage_pts
            else:
                raw_exit = raw_exit - slippage_pts
            exit_price = raw_exit
            exit_ts = ts
            win = hit_target
            break
        hold_count += 1
        if max_hold_bars is not None and hold_count >= max_hold_bars:
            raw_exit = C
            exit_reason = 'time'
            if eff_side=='SHORT':
                raw_exit = raw_exit + slippage_pts
                win = (entry_price - raw_exit) > 0
            else:
                raw_exit = raw_exit - slippage_pts
                win = (raw_exit - entry_price) > 0
            exit_price = raw_exit; exit_ts = ts
            break
    if exit_price is None:
        last_close = float(bars.iloc[-1]['Close'])
        raw_exit = last_close
        if eff_side=='SHORT':
            raw_exit = raw_exit + slippage_pts
            win = (entry_price - raw_exit) > 0
        else:
            raw_exit = raw_exit - slippage_pts
            win = (raw_exit - entry_price) > 0
        exit_price = raw_exit; exit_ts = bars.index[-1]; exit_reason='end'
    if eff_side=='SHORT':
        pnl_price = entry_price - exit_price
    else:
        pnl_price = exit_price - entry_price
    pnl_price = pnl_price - spread_pts
    pnl_usd = pnl_price * CONTRACT_MULT
    return {'entry_ts':entry_idx,'exit_ts':exit_ts,'side':eff_side,'orig_side':side,'flip':flip,
            'entry_price':entry_price,'exit_price':exit_price,'pnl_price':pnl_price,'pnl_usd':pnl_usd,
            'win':bool(win),'exit_reason':exit_reason,'stop_mult':stop_mult,'target_mult':target_mult,'hold_bars':max_hold_bars}

# ---------- Load data ----------
print("Loading 5-min CSV:", IN_5M)
df5 = load_and_prepare_5m(IN_5M)
print("Bars:", len(df5), "from", df5.index.min(), "to", df5.index.max())

# daily data + indicators
daily = resample_daily_from_5m(df5)
daily = compute_indicators(daily, ema_short=max(3, DAILY_EMA_LEN//4), ema_long=DAILY_EMA_LEN)
print("Daily rows:", len(daily))

# Map daily ATR to 5-min bars by date (safe: use date objects)
daily_atr_by_date = {ts.date(): float(val) for ts, val in daily['atr'].items()}
df5 = df5.copy()
df5['date_only'] = [ts.date() for ts in df5.index]
df5['daily_atr'] = df5['date_only'].map(lambda d: daily_atr_by_date.get(d, np.nan))
# drop helper col if you like
df5.drop(columns=['date_only'], inplace=True)
print("Mapped daily ATR to 5-min bars. Example:", df5[['Open','High','Low','Close','daily_atr']].iloc[:3].to_dict('records'))

# ---------- Walk-forward fold windows ----------
start_date = daily.index.min(); end_date = daily.index.max()
folds = []
cursor = start_date
while True:
    train_end = cursor + pd.DateOffset(years=TRAIN_YEARS)
    test_end  = train_end + pd.DateOffset(years=TEST_YEARS)
    if test_end > end_date:
        break
    folds.append((cursor, train_end, train_end + pd.Timedelta(days=0), test_end))
    cursor = cursor + pd.DateOffset(years=TEST_YEARS)
print("Prepared", len(folds), "walk-forward folds")

# ---------- Walk-forward grid search ----------
results=[]; all_trades_detailed=[]
for fold_idx, (train_start, train_end, test_start, test_end) in enumerate(folds):
    print(f"\n=== Fold {fold_idx+1}: train {train_start.date()} -> {train_end.date()} ; test {test_start.date()} -> {test_end.date()} ===")
    if train_end <= train_start or test_end <= test_start:
        print("Invalid fold dates; skipping")
        continue
    # gen signals for whole fold range then map to 5m
    signals_total_daily = generate_signals_from_daily(daily.loc[train_start:test_end], trend_ret_threshold=0.01)
    signals_total_5m = map_daily_signals_to_5m(df5, signals_total_daily)
    signals_test_5m = signals_total_5m.loc[(signals_total_5m.index >= test_start) & (signals_total_5m.index <= test_end)]
    combos = list(itertools.product(STOP_MULTS, TARGET_MULTS, MAX_HOLD_BARS, FLIP_LOGIC))
    print("Testing", len(combos), "configs on", len(signals_test_5m), "test signals")
    for stop_mult, target_mult, max_hold, flip_flag in combos:
        trades=[]
        for entry_ts, s in signals_test_5m.iterrows():
            # determine day key and daily row
            day_key = pd.Timestamp(entry_ts).tz_convert('UTC').normalize()
            if day_key not in daily.index:
                # try mapping by date
                d = day_key.date()
                if d not in daily.index.date:
                    continue
                day_key = [x for x in daily.index if x.date()==d][0]
            daily_row = daily.loc[day_key]
            # volatility filter
            if daily_row['atr'] > daily['atr'].quantile(VOL_ATR_QUANTILE):
                continue
            # HTF confirmation
            intended_side = s['side']
            if intended_side=='SHORT' and not (daily_row['ema_s'] > daily_row['ema_l']): continue
            if intended_side=='LONG' and not (daily_row['ema_s'] < daily_row['ema_l']): continue
            # simulate
            res = simulate_trade_on_5m(df5, entry_ts, s['side'], stop_mult, target_mult, max_hold, SPREAD_PTS, SLIPPAGE_PTS, flip=flip_flag)
            if res is not None:
                res.update({'fold':fold_idx+1,'test_start':test_start.date().isoformat(),'test_end':test_end.date().isoformat()})
                trades.append(res)
        if len(trades) < MIN_TRADES_PER_CONFIG:
            continue
        tdf = pd.DataFrame(trades)
        total_pnl = float(tdf['pnl_usd'].sum())
        wins = int(tdf['win'].sum()); win_rate = wins / len(tdf)
        avg_pnl = float(tdf['pnl_price'].mean()); sharpe = (tdf['pnl_usd'].mean() / tdf['pnl_usd'].std(ddof=0)) if tdf['pnl_usd'].std(ddof=0) != 0 else np.nan
        results.append({'fold':fold_idx+1,'stop_mult':stop_mult,'target_mult':target_mult,'max_hold_bars':max_hold,'flip':flip_flag,
                        'n_trades':len(tdf),'wins':wins,'win_rate':win_rate,'total_pnl_usd':total_pnl,'avg_pnl_price':avg_pnl,'sharpe':sharpe})
        all_trades_detailed += trades

# ---------- save & print ----------
res_df = pd.DataFrame(results).sort_values(by='total_pnl_usd', ascending=False).reset_index(drop=True)
res_df.to_csv(os.path.join(OUT_DIR, "wf_grid_results.csv"), index=False)
pd.DataFrame(all_trades_detailed).to_csv(os.path.join(OUT_DIR, "wf_all_trades_detailed.csv"), index=False)

print("\nTop configs by test cumulative PnL (head):")
print(res_df.head(10).to_string(index=False))
print("\nSaved results to", OUT_DIR)
```

```python
# WALK-FORWARD GRID SEARCH (ready to run)
!pip install --quiet pandas numpy tqdm

import pandas as pd, numpy as np, os, itertools
from tqdm.auto import tqdm
from datetime import datetime
import warnings
warnings.filterwarnings("ignore")

# ---------- USER / FILE PARAMETERS ----------
IN_5M = "/mnt/data/dukascopy_xauusd_5min.csv"
OUT_DIR = "/mnt/data/gc_model_tweaks_results"
os.makedirs(OUT_DIR, exist_ok=True)

# Walk-forward params
TRAIN_YEARS = 1.5
TEST_YEARS  = 0.5

# Grid to test
STOP_MULTS   = [0.5, 1.0, 1.5]
TARGET_MULTS = [0.6, 1.0, 1.5]
MAX_HOLD_BARS = [1, 5, 10, 20]

# Strategy/filters
FLIP_LOGIC = [False, True]
DAILY_EMA_LEN = 20
VOL_ATR_QUANTILE = 0.9
MIN_TRADES_PER_CONFIG = 10
SPREAD_PTS = 0.3
SLIPPAGE_PTS = 0.2
CONTRACT_MULT = 100.0
RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)
# -----------------------------------------------

def load_and_prepare_5m(path):
    df = pd.read_csv(path, index_col=0, parse_dates=True)
    if df.index.tz is None:
        df.index = df.index.tz_localize('UTC')
    else:
        df.index = df.index.tz_convert('UTC')
    df.columns = [c.title() for c in df.columns]
    for col in ['Open','High','Low','Close']:
        if col not in df.columns:
            raise SystemExit(f"CSV missing required column: {col}")
    return df[['Open','High','Low','Close']].astype(float)

def resample_daily_from_5m(df5):
    daily = pd.DataFrame()
    daily['Open']  = df5['Open'].resample('1D').first()
    daily['High']  = df5['High'].resample('1D').max()
    daily['Low']   = df5['Low'].resample('1D').min()
    daily['Close'] = df5['Close'].resample('1D').last()
    daily.dropna(how='any', inplace=True)
    return daily

def compute_indicators(df, ema_short=5, ema_long=20, atr_len=14):
    df = df.copy()
    df['ema_s'] = df['Close'].ewm(span=ema_short, adjust=False).mean()
    df['ema_l'] = df['Close'].ewm(span=ema_long, adjust=False).mean()
    high = df['High']; low = df['Low']; close = df['Close']
    tr = pd.concat([high-low, (high - close.shift(1)).abs(), (low - close.shift(1)).abs()], axis=1).max(axis=1)
    df['atr'] = tr.rolling(atr_len, min_periods=1).mean()
    df['ret_trend'] = df['Close'].pct_change(periods=15)
    return df

def generate_signals_from_daily(daily_df, trend_ret_threshold=0.01):
    rows = []
    dd = daily_df.copy()
    for i in range(0, len(dd)-1):
        row = dd.iloc[i]; next_idx = dd.index[i+1]
        if row['ret_trend'] >= trend_ret_threshold and row['ema_s'] > row['ema_l']:
            rows.append({'entry_date': next_idx, 'side':'SHORT', 'signal_date': dd.index[i]})
        if row['ret_trend'] <= -trend_ret_threshold and row['ema_s'] < row['ema_l']:
            rows.append({'entry_date': next_idx, 'side':'LONG', 'signal_date': dd.index[i]})
    if len(rows)==0:
        return pd.DataFrame(columns=['entry_date','side','signal_date']).set_index('entry_date')
    return pd.DataFrame(rows).set_index('entry_date')

def map_daily_signals_to_5m(df5, signals_daily):
    sigs = []
    for idx, s in signals_daily.iterrows():
        day_start = pd.Timestamp(idx.date()).tz_localize('UTC')
        day_bars = df5.loc[day_start: day_start + pd.Timedelta(days=1) - pd.Timedelta(minutes=1)]
        if len(day_bars)==0:
            continue
        entry_ts = day_bars.index[0]
        sigs.append({'entry_ts': entry_ts, 'side': s['side'], 'signal_date': s['signal_date']})
    if len(sigs)==0:
        return pd.DataFrame(columns=['entry_ts','side','signal_date']).set_index('entry_ts')
    return pd.DataFrame(sigs).set_index('entry_ts').sort_index()

def simulate_trade_on_5m(df5, entry_idx, side, stop_mult, target_mult, max_hold_bars, spread_pts, slippage_pts, flip=False):
    try:
        entry_row = df5.loc[entry_idx]
    except KeyError:
        return None
    entry_price = float(entry_row['Open'])
    atr = float(df5.loc[entry_idx].get('daily_atr', np.nan))
    if pd.isna(atr) or atr <= 0:
        atr = max(0.5, abs(entry_price)*0.0001)
    eff_side = 'LONG' if (side=='SHORT' and flip) or (side=='LONG' and not flip) else ('SHORT' if (side=='LONG' and flip) or (side=='SHORT' and not flip) else side)
    if eff_side == 'SHORT':
        stop_price = entry_price + stop_mult * atr
        target_price = entry_price - target_mult * atr
    else:
        stop_price = entry_price - stop_mult * atr
        target_price = entry_price + target_mult * atr

    bars = df5.loc[entry_idx:]
    exit_price = None; exit_ts=None; win=False; exit_reason='close'
    hold_count=0
    for ts, row in bars.iterrows():
        H=float(row['High']); L=float(row['Low']); C=float(row['Close'])
        hit_target=False; hit_stop=False
        if eff_side=='SHORT':
            if L <= target_price: hit_target=True
            if H >= stop_price: hit_stop=True
            if hit_target and hit_stop:
                if abs(entry_price - target_price) <= abs(stop_price - entry_price):
                    hit_stop=False
                else:
                    hit_target=False
        else:
            if H >= target_price: hit_target=True
            if L <= stop_price: hit_stop=True
            if hit_target and hit_stop:
                if abs(entry_price - target_price) <= abs(entry_price - stop_price):
                    hit_stop=False
                else:
                    hit_target=False
        if hit_target or hit_stop:
            raw_exit = target_price if hit_target else stop_price
            exit_reason = 'target' if hit_target else 'stop'
            if eff_side=='SHORT':
                raw_exit = raw_exit + slippage_pts
            else:
                raw_exit = raw_exit - slippage_pts
            exit_price = raw_exit; exit_ts = ts; win = hit_target
            break
        hold_count += 1
        if max_hold_bars is not None and hold_count >= max_hold_bars:
            raw_exit = C
            exit_reason = 'time'
            if eff_side=='SHORT':
                raw_exit = raw_exit + slippage_pts
                win = (entry_price - raw_exit) > 0
            else:
                raw_exit = raw_exit - slippage_pts
                win = (raw_exit - entry_price) > 0
            exit_price = raw_exit; exit_ts = ts
            break
    if exit_price is None:
        last_close = float(bars.iloc[-1]['Close'])
        raw_exit = last_close
        if eff_side=='SHORT':
            raw_exit = raw_exit + slippage_pts
            win = (entry_price - raw_exit) > 0
        else:
            raw_exit = raw_exit - slippage_pts
            win = (raw_exit - entry_price) > 0
        exit_price = raw_exit; exit_ts = bars.index[-1]; exit_reason='end'
    if eff_side=='SHORT':
        pnl_price = entry_price - exit_price
    else:
        pnl_price = exit_price - entry_price
    pnl_price = pnl_price - spread_pts
    pnl_usd = pnl_price * CONTRACT_MULT
    return {'entry_ts':entry_idx,'exit_ts':exit_ts,'side':eff_side,'orig_side':side,'flip':flip,
            'entry_price':entry_price,'exit_price':exit_price,'pnl_price':pnl_price,'pnl_usd':pnl_usd,
            'win':bool(win),'exit_reason':exit_reason,'stop_mult':stop_mult,'target_mult':target_mult,'hold_bars':max_hold_bars}

# ---------- Load data ----------
print("Loading 5-min CSV:", IN_5M)
df5 = load_and_prepare_5m(IN_5M)
print("Bars:", len(df5), "from", df5.index.min(), "to", df5.index.max())

# daily data + indicators
daily = resample_daily_from_5m(df5)
daily = compute_indicators(daily, ema_short=max(3, DAILY_EMA_LEN//4), ema_long=DAILY_EMA_LEN)
print("Daily rows:", len(daily))

# Map daily ATR to 5-min by date
daily_atr_by_date = {ts.date(): float(val) for ts, val in daily['atr'].items()}
df5 = df5.copy()
df5['date_only'] = [ts.date() for ts in df5.index]
df5['daily_atr'] = df5['date_only'].map(lambda d: daily_atr_by_date.get(d, np.nan))
df5.drop(columns=['date_only'], inplace=True)

# ---------- build walk-forward folds ----------
start_date = daily.index.min(); end_date = daily.index.max()
folds=[]
cursor=start_date
while True:
    train_end = cursor + pd.DateOffset(years=TRAIN_YEARS)
    test_end  = train_end + pd.DateOffset(years=TEST_YEARS)
    if test_end > end_date:
        break
    folds.append((cursor, train_end, train_end + pd.Timedelta(days=0), test_end))
    cursor = cursor + pd.DateOffset(years=TEST_YEARS)
print("Prepared", len(folds), "walk-forward folds")

# ---------- main WF grid ----------
results=[]; all_trades=[]
for fold_idx, (train_start, train_end, test_start, test_end) in enumerate(folds):
    print(f"\n=== Fold {fold_idx+1}: train {train_start.date()} -> {train_end.date()} ; test {test_start.date()} -> {test_end.date()} ===")
    signals_total_daily = generate_signals_from_daily(daily.loc[train_start:test_end], trend_ret_threshold=0.01)
    signals_total_5m = map_daily_signals_to_5m(df5, signals_total_daily)
    signals_test_5m = signals_total_5m.loc[(signals_total_5m.index >= test_start) & (signals_total_5m.index <= test_end)]
    combos = list(itertools.product(STOP_MULTS, TARGET_MULTS, MAX_HOLD_BARS, FLIP_LOGIC))
    print("Testing", len(combos), "configs on", len(signals_test_5m), "test signals")
    for stop_mult, target_mult, max_hold, flip_flag in combos:
        trades=[]
        for entry_ts, s in signals_test_5m.iterrows():
            day_key = pd.Timestamp(entry_ts).tz_convert('UTC').normalize()
            if day_key not in daily.index:
                d = day_key.date()
                if d not in daily.index.date:
                    continue
                day_key = [x for x in daily.index if x.date()==d][0]
            daily_row = daily.loc[day_key]
            if daily_row['atr'] > daily['atr'].quantile(VOL_ATR_QUANTILE):
                continue
            intended_side = s['side']
            if intended_side=='SHORT' and not (daily_row['ema_s'] > daily_row['ema_l']): continue
            if intended_side=='LONG' and not (daily_row['ema_s'] < daily_row['ema_l']): continue
            res = simulate_trade_on_5m(df5, entry_ts, s['side'], stop_mult, target_mult, max_hold, SPREAD_PTS, SLIPPAGE_PTS, flip=flip_flag)
            if res is not None:
                res.update({'fold':fold_idx+1})
                trades.append(res)
        if len(trades) < MIN_TRADES_PER_CONFIG:
            continue
        tdf = pd.DataFrame(trades)
        total_pnl = float(tdf['pnl_usd'].sum())
        wins = int(tdf['win'].sum())
        results.append({'fold':fold_idx+1,'stop_mult':stop_mult,'target_mult':target_mult,'max_hold_bars':max_hold,'flip':flip_flag,
                        'n_trades':len(tdf),'wins':wins,'win_rate':wins/len(tdf),'total_pnl_usd':total_pnl,'avg_pnl_price':float(tdf['pnl_price'].mean())})
        all_trades += trades

# save & print
res_df = pd.DataFrame(results).sort_values(by='total_pnl_usd', ascending=False).reset_index(drop=True)
res_df.to_csv(os.path.join(OUT_DIR, "wf_grid_results.csv"), index=False)
pd.DataFrame(all_trades).to_csv(os.path.join(OUT_DIR, "wf_all_trades_detailed.csv"), index=False)

print("\nTop configs by test cumulative PnL:")
if not res_df.empty:
    print(res_df.head(10).to_string(index=False))
else:
    print("No configs met MIN_TRADES_PER_CONFIG. Inspect signals / lower MIN_TRADES_PER_CONFIG.")
print("\nSaved results to", OUT_DIR, "files: wf_grid_results.csv and wf_all_trades_detailed.csv")
```

```python
# FULL WALK-FORWARD GRID SEARCH - READY FOR GOOGLE COLAB
!pip install --quiet pandas numpy tqdm

import pandas as pd, numpy as np, os, itertools, warnings
from tqdm.auto import tqdm
from datetime import datetime
warnings.filterwarnings("ignore")

# --------- USER / FILE PATHS & HYPERPARAMS ----------
IN_5M = "/mnt/data/dukascopy_xauusd_5min.csv"      # already fetched earlier
OUT_DIR = "/mnt/data/gc_model_tweaks_results"
os.makedirs(OUT_DIR, exist_ok=True)

# walk-forward window sizes (in years, can be fractional)
TRAIN_YEARS = 1.5
TEST_YEARS  = 0.5

# grid to test
STOP_MULTS   = [0.5, 1.0, 1.5]
TARGET_MULTS = [0.6, 1.0, 1.5]
MAX_HOLD_BARS = [1, 5, 10, 20]
FLIP_LOGIC = [False, True]

# filters / misc
DAILY_EMA_LEN = 20
VOL_ATR_QUANTILE = 0.9
MIN_TRADES_PER_CONFIG = 10
SPREAD_PTS = 0.3
SLIPPAGE_PTS = 0.2
CONTRACT_MULT = 100.0
RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)
# -----------------------------------------------------

# ---------- helper functions ----------
def load_and_prepare_5m(path):
    df = pd.read_csv(path, index_col=0, parse_dates=True)
    # ensure tz-aware UTC index
    if df.index.tz is None:
        df.index = pd.to_datetime(df.index).tz_localize('UTC')
    else:
        df.index = pd.to_datetime(df.index).tz_convert('UTC')
    df.columns = [c.title() for c in df.columns]
    for col in ['Open','High','Low','Close']:
        if col not in df.columns:
            raise SystemExit(f"CSV missing required column: {col}")
    return df[['Open','High','Low','Close']].astype(float)

def resample_daily_from_5m(df5):
    daily = pd.DataFrame()
    daily['Open']  = df5['Open'].resample('1D').first()
    daily['High']  = df5['High'].resample('1D').max()
    daily['Low']   = df5['Low'].resample('1D').min()
    daily['Close'] = df5['Close'].resample('1D').last()
    daily.dropna(how='any', inplace=True)
    return daily

def compute_indicators(df, ema_short=5, ema_long=20, atr_len=14):
    df = df.copy()
    df['ema_s'] = df['Close'].ewm(span=ema_short, adjust=False).mean()
    df['ema_l'] = df['Close'].ewm(span=ema_long, adjust=False).mean()
    high = df['High']; low = df['Low']; close = df['Close']
    tr = pd.concat([high-low, (high - close.shift(1)).abs(), (low - close.shift(1)).abs()], axis=1).max(axis=1)
    df['atr'] = tr.rolling(atr_len, min_periods=1).mean()
    df['ret_trend'] = df['Close'].pct_change(periods=15)
    return df

def generate_signals_from_daily(daily_df, trend_ret_threshold=0.01):
    rows=[]
    dd = daily_df.copy()
    for i in range(0, len(dd)-1):
        r = dd.iloc[i]; nxt = dd.index[i+1]
        if r['ret_trend'] >= trend_ret_threshold and r['ema_s'] > r['ema_l']:
            rows.append({'entry_date': nxt, 'side':'SHORT'})
        if r['ret_trend'] <= -trend_ret_threshold and r['ema_s'] < r['ema_l']:
            rows.append({'entry_date': nxt, 'side':'LONG'})
    if not rows:
        return pd.DataFrame(columns=['entry_date','side']).set_index('entry_date')
    return pd.DataFrame(rows).set_index('entry_date')

def map_daily_signals_to_5m(df5, signals_daily):
    sigs=[]
    for idx, s in signals_daily.iterrows():
        day_start = pd.Timestamp(idx.date()).tz_localize('UTC')
        day_bars = df5.loc[day_start : day_start + pd.Timedelta(days=1) - pd.Timedelta(minutes=1)]
        if len(day_bars)==0:
            continue
        entry_ts = day_bars.index[0]
        sigs.append({'entry_ts': entry_ts, 'side': s['side']})
    if not sigs:
        return pd.DataFrame(columns=['entry_ts','side']).set_index('entry_ts')
    return pd.DataFrame(sigs).set_index('entry_ts').sort_index()

def simulate_trade_on_5m(df5, entry_idx, side, stop_mult, target_mult, max_hold_bars, spread_pts, slippage_pts, flip=False):
    try:
        entry_row = df5.loc[entry_idx]
    except KeyError:
        return None
    entry_price = float(entry_row['Open'])
    atr = float(df5.loc[entry_idx].get('daily_atr', np.nan))
    if pd.isna(atr) or atr <= 0:
        atr = max(0.5, abs(entry_price)*0.0001)
    # flip logic: if flip True, reverse intended side
    eff_side = 'LONG' if (side=='SHORT' and flip) or (side=='LONG' and not flip) else ('SHORT' if (side=='LONG' and flip) or (side=='SHORT' and not flip) else side)
    if eff_side == 'SHORT':
        stop_price = entry_price + stop_mult * atr
        target_price = entry_price - target_mult * atr
    else:
        stop_price = entry_price - stop_mult * atr
        target_price = entry_price + target_mult * atr

    bars = df5.loc[entry_idx:]
    exit_price=None; exit_ts=None; win=False; exit_reason='close'
    hold_count=0
    for ts, row in bars.iterrows():
        H=float(row['High']); L=float(row['Low']); C=float(row['Close'])
        hit_target=False; hit_stop=False
        if eff_side=='SHORT':
            if L <= target_price: hit_target=True
            if H >= stop_price: hit_stop=True
            if hit_target and hit_stop:
                # choose the one closer to entry (tie-break)
                if abs(entry_price - target_price) <= abs(stop_price - entry_price):
                    hit_stop=False
                else:
                    hit_target=False
        else:
            if H >= target_price: hit_target=True
            if L <= stop_price: hit_stop=True
            if hit_target and hit_stop:
                if abs(entry_price - target_price) <= abs(entry_price - stop_price):
                    hit_stop=False
                else:
                    hit_target=False
        if hit_target or hit_stop:
            raw_exit = target_price if hit_target else stop_price
            exit_reason = 'target' if hit_target else 'stop'
            # adverse slippage
            raw_exit = raw_exit + slippage_pts if eff_side=='SHORT' else raw_exit - slippage_pts
            exit_price = raw_exit; exit_ts = ts; win = hit_target
            break
        hold_count += 1
        if max_hold_bars is not None and hold_count >= max_hold_bars:
            raw_exit = C
            exit_reason='time'
            raw_exit = raw_exit + slippage_pts if eff_side=='SHORT' else raw_exit - slippage_pts
            win = (entry_price - raw_exit) > 0 if eff_side=='SHORT' else (raw_exit - entry_price) > 0
            exit_price = raw_exit; exit_ts = ts
            break

    if exit_price is None:
        last_close = float(bars.iloc[-1]['Close'])
        raw_exit = last_close + (slippage_pts if eff_side=='SHORT' else -slippage_pts)
        win = (entry_price - raw_exit) > 0 if eff_side=='SHORT' else (raw_exit - entry_price) > 0
        exit_price = raw_exit; exit_ts = bars.index[-1]; exit_reason='end'
    pnl_price = (entry_price - exit_price) if eff_side=='SHORT' else (exit_price - entry_price)
    pnl_price = pnl_price - spread_pts
    pnl_usd = pnl_price * CONTRACT_MULT
    return {'entry_ts':entry_idx,'exit_ts':exit_ts,'side':eff_side,'orig_side':side,'flip':flip,
            'entry_price':entry_price,'exit_price':exit_price,'pnl_price':pnl_price,'pnl_usd':pnl_usd,
            'win':bool(win),'exit_reason':exit_reason,'stop_mult':stop_mult,'target_mult':target_mult,'hold_bars':max_hold_bars}

# ---------- load data ----------
print("Loading 5-min CSV:", IN_5M)
df5 = load_and_prepare_5m(IN_5M)
print("Bars:", len(df5), "from", df5.index.min(), "to", df5.index.max())

# daily + indicators
daily = resample_daily_from_5m(df5)
daily = compute_indicators(daily, ema_short=max(3, DAILY_EMA_LEN//4), ema_long=DAILY_EMA_LEN)
print("Daily rows:", len(daily))

# map daily ATR to 5-min bars using date objects (safe with tz-aware index)
daily_atr_by_date = {ts.date(): float(val) for ts, val in daily['atr'].items()}
df5 = df5.copy()
df5['date_only'] = [ts.date() for ts in df5.index]
df5['daily_atr'] = df5['date_only'].map(lambda d: daily_atr_by_date.get(d, np.nan))
df5.drop(columns=['date_only'], inplace=True)
print("Mapped daily ATR to 5-min bars (example):")
print(df5[['Open','High','Low','Close','daily_atr']].head().to_string(index=True))

# ---------- build month-based walk-forward folds (works with fractional TRAIN_YEARS) ----------
months_train = max(1, int(round(TRAIN_YEARS * 12)))
months_test  = max(1, int(round(TEST_YEARS  * 12)))
start_date = daily.index.min(); end_date = daily.index.max()
folds=[]; cursor = start_date
print(f"\nBuilding folds using {months_train} months train / {months_test} months test")
while True:
    train_end = cursor + pd.DateOffset(months=months_train)
    test_end  = train_end + pd.DateOffset(months=months_test)
    if test_end > end_date:
        break
    folds.append((cursor, train_end, train_end + pd.Timedelta(days=0), test_end))
    cursor = cursor + pd.DateOffset(months=months_test)
print("Prepared", len(folds), "walk-forward folds")
for i, (ts, te, _, te2) in enumerate(folds):
    print(f" Fold {i+1}: train {ts.date()} -> {te.date()}   test -> {te2.date()}")

# ---------- walk-forward grid search ----------
results=[]; all_trades=[]
for fold_idx, (train_start, train_end, test_start, test_end) in enumerate(folds):
    print(f"\n=== Fold {fold_idx+1}: train {train_start.date()} -> {train_end.date()} ; test {test_start.date()} -> {test_end.date()} ===")
    # generate signals on train+test windows to allow mapping to 5-min; evaluate only test portion
    signals_total_daily = generate_signals_from_daily(daily.loc[train_start:test_end], trend_ret_threshold=0.01)
    signals_total_5m = map_daily_signals_to_5m(df5, signals_total_daily)
    signals_test_5m = signals_total_5m.loc[(signals_total_5m.index >= test_start) & (signals_total_5m.index <= test_end)]
    combos = list(itertools.product(STOP_MULTS, TARGET_MULTS, MAX_HOLD_BARS, FLIP_LOGIC))
    print(" Testing", len(combos), "configs on", len(signals_test_5m), "test signals")
    for stop_mult, target_mult, max_hold, flip_flag in combos:
        trades=[]
        for entry_ts, s in signals_test_5m.iterrows():
            # map daily row (safe)
            day_key = pd.Timestamp(entry_ts).tz_convert('UTC').normalize()
            if day_key not in daily.index:
                d = day_key.date()
                if d not in daily.index.date:
                    continue
                day_key = [x for x in daily.index if x.date()==d][0]
            daily_row = daily.loc[day_key]
            # vol filter
            if daily_row['atr'] > daily['atr'].quantile(VOL_ATR_QUANTILE):
                continue
            intended_side = s['side']
            if intended_side=='SHORT' and not (daily_row['ema_s'] > daily_row['ema_l']): continue
            if intended_side=='LONG' and not (daily_row['ema_s'] < daily_row['ema_l']): continue
            res = simulate_trade_on_5m(df5, entry_ts, s['side'], stop_mult, target_mult, max_hold, SPREAD_PTS, SLIPPAGE_PTS, flip=flip_flag)
            if res is not None:
                res.update({'fold':fold_idx+1})
                trades.append(res)
        if len(trades) < MIN_TRADES_PER_CONFIG:
            continue
        tdf = pd.DataFrame(trades)
        total_pnl = float(tdf['pnl_usd'].sum())
        wins = int(tdf['win'].sum()); win_rate = wins/len(tdf)
        results.append({'fold':fold_idx+1,'stop_mult':stop_mult,'target_mult':target_mult,'max_hold_bars':max_hold,'flip':flip_flag,
                        'n_trades':len(tdf),'wins':wins,'win_rate':win_rate,'total_pnl_usd':total_pnl,'avg_pnl_price':float(tdf['pnl_price'].mean())})
        all_trades += trades

# ---------- save & display ----------
res_df = pd.DataFrame(results).sort_values(by='total_pnl_usd', ascending=False).reset_index(drop=True)
res_df.to_csv(os.path.join(OUT_DIR, "wf_grid_results.csv"), index=False)
pd.DataFrame(all_trades).to_csv(os.path.join(OUT_DIR, "wf_all_trades_detailed.csv"), index=False)

print("\nTop configs by test cumulative PnL:")
if not res_df.empty:
    print(res_df.head(10).to_string(index=False))
else:
    print("No configs met MIN_TRADES_PER_CONFIG. Try lowering MIN_TRADES_PER_CONFIG or check signals.")
print("\nSaved results to", OUT_DIR, " (wf_grid_results.csv, wf_all_trades_detailed.csv)")
```
